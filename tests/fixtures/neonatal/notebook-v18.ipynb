## 1. Basis Data Pedoman (Dosis & Interval)

Sel berikut menyimpan seluruh basis rujukan sebagai fungsi Python:

- **Bagian 1a** — nomogram Pedoman RS (usia gestasi/PMA → dosis & interval spesifik).
- **Bagian 1b** — **kriteria kesesuaian** Pedoman RS (rentang dosis & pilihan frekuensi yang dianggap tepat), *berlaku untuk semua neonatus usia pascalahir 0–28 hari*.
- **Bagian 2a** — nomogram Literatur ANMF 2024 — **satu-satunya** dasar angka "Literatur Terkini" pada tabel rekomendasi, evaluasi, PDF, dan batch.
- **Bagian 2b** — nomogram Drugs.com/ASHP (AAP & IDSA) — **tidak dipakai** pada versi aturan 2026.11.01 (`GABUNG_DRUGSCOM_KE_LITERATUR = False`, `TAMPILKAN_DRUGSCOM_TAMBAHAN = False`); kode disimpan agar dapat diaktifkan kembali.
- **Bagian 2c** — catatan kualitatif dari WHO 2024, Swiss Society of Neonatology 2024, JAID/JSC 2021, dan ACOG 2020/AAP.

Bagian ini murni **basis data & logika perhitungan** — tidak perlu diubah, kecuali SPO rumah sakit Anda direvisi (lihat penanda `# EDIT DI SINI`).


```python
# ============================================================================
# KONFIGURASI VERSI & CAKUPAN (revisi 2026.11.01)
# ============================================================================
VERSI_ATURAN = "2026.11.01"
USIA_PASCALAHIR_MAKS_HARI = 28   # kalkulator khusus neonatus
# Literatur Terkini = ANMF 2024 SAJA (keputusan peneliti, revisi 2026.11.01).
# True  = rentang Literatur Terkini menggabungkan ANMF 2024 + Drugs.com/ASHP (perilaku versi 20 Agustus 2026).
# False = Literatur Terkini memakai ANMF 2024 saja.  JANGAN diubah tanpa instruksi peneliti.
GABUNG_DRUGSCOM_KE_LITERATUR = False
# True  = kartu Drugs.com/ASHP tetap ditampilkan sebagai rujukan tambahan (mode Rekomendasi & Evaluasi).
# False = kartu Drugs.com/ASHP disembunyikan (kode tetap disimpan agar bisa diaktifkan kembali).
TAMPILKAN_DRUGSCOM_TAMBAHAN = False


# ============================================================================
# BAGIAN 1a: PEDOMAN RS -- NOMOGRAM (usia gestasi/PMA -> dosis & interval)
# ============================================================================

def hitung_pma(usia_gestasi_lahir_minggu, usia_pascalahir_hari):
    """PMA (postmenstrual age / usia gestasi terkoreksi) = usia gestasi saat
    lahir (minggu) + usia pascalahir (dikonversi ke minggu)."""
    return usia_gestasi_lahir_minggu + (usia_pascalahir_hari / 7.0)


def kriteria_gangguan_ginjal(scr, uop=None):
    """Kriteria gangguan ginjal neonatus (<=44 minggu cGA) menurut UNC
    Vancomycin Dosing & Monitoring Guide (Sep 2023), Tabel 1:
    SCr ≥ 1 mg/dL  ATAU  (SCr ≥ 0,5 mg/dL DAN produksi urin < 1 mL/kg/jam)."""
    if scr is None:
        return False
    if scr >= 1.0:
        return True
    if uop is not None and scr >= 0.5 and uop < 1.0:
        return True
    return False


def rs_gentamisin(ga_lahir_minggu, pna_hari):
    """GENTAMISIN -- Bagian 1a: Pedoman RS.
    Sumber: SPO Sepsis Neonatal RSUP Dr. M. Djamil Padang, No. Dok
    HK.03.04/III/237/2017 (Lini I): gentamisin 5 mg/kg/kali, interval 24-48 jam.
    Konsisten dengan PNPK Sepsis pada Anak, Kemenkes RI 2021 (5-7,5 mg/kg/hari
    dosis tunggal).
    CATATAN PENTING: SPO/PNPK yang dilampirkan TIDAK memuat tabel penyesuaian
    dosis maupun interval berdasarkan usia gestasi / usia pascalahir. Karena itu
    rekomendasi Pedoman RS di sini TIDAK bergantung pada usia gestasi/pascalahir:
    dosis dinyatakan sebagai rentang 5-7,5 mg/kg dan frekuensi sebagai rentang
    24-48 jam untuk semua neonatus.
    # EDIT DI SINI jika SPO RS Anda memiliki tabel dosis/interval spesifik.
    """
    dosis_min_kg, dosis_max_kg = 5.0, 7.5
    interval_min, interval_max = 24, 48
    # Versi "mempertimbangkan usia gestasi dan usia pascalahir": interval spesifik via nomogram aminoglikosida
    # standar, DIBATASI pada jendela 24-48 jam yang dinyatakan SPO. Ini BUKAN dari
    # SPO (SPO tak punya tabel usia), melainkan penerapan nomogram lazim sebagai
    # pembanding "seandainya usia diperhitungkan".
    # Batas pita memakai minggu LENGKAP (mis. 29 minggu 5 hari termasuk pita <30 minggu).
    if ga_lahir_minggu < 30:
        interval_ga = 48 if pna_hari <= 7 else (36 if pna_hari <= 28 else 24)
    elif ga_lahir_minggu < 35:
        interval_ga = 36 if pna_hari <= 7 else 24
    else:
        interval_ga = 24
    catatan = [
        "Dosis 5-7,5 mg/kg dan interval 24-48 jam sesuai SPO RSUP Dr. M. Djamil Padang (Lini I) & PNPK Kemenkes RI 2021.",
        "SPO tidak membedakan dosis maupun interval menurut usia gestasi atau usia pascalahir. Penyesuaian interval menurut usia ditampilkan pada kolom Literatur Terkini.",
    ]
    sumber = "SPO Sepsis Neonatal RSUP Dr. M. Djamil Padang (2017); PNPK Sepsis pada Anak, Kemenkes RI (2021)"
    return dosis_min_kg, dosis_max_kg, interval_min, interval_max, interval_ga, catatan, sumber


def rs_amikasin(ga_lahir_minggu, pna_hari):
    """AMIKASIN -- Bagian 1a: Pedoman RS.
    Sumber: SPO RSUP Dr. M. Djamil Padang (Lini III): amikasin 15 mg/kg/kali,
    interval 24 jam. Konsisten dengan PNPK Kemenkes RI 2021 (15-20 mg/kg/hari
    dosis tunggal; SPO memakai 15 mg/kg).
    CATATAN PENTING: SPO tidak memuat penyesuaian dosis/interval berdasarkan usia
    gestasi / usia pascalahir; rekomendasi RS bersifat TETAP (15 mg/kg, tiap 24 jam)
    untuk semua neonatus.
    # EDIT DI SINI jika SPO RS Anda memiliki tabel dosis/interval spesifik.
    """
    dosis_min_kg, dosis_max_kg = 15.0, 15.0
    interval_min, interval_max = 24, 24
    interval_ga = 24  # SPO tidak memberi variasi interval berdasarkan usia -> sama dengan versi tanpa usia gestasi dan usia pascalahir
    catatan = [
        "Dosis tetap 15 mg/kg dan interval tetap 24 jam sesuai SPO RSUP Dr. M. Djamil Padang (Lini III).",
        "SPO tidak membedakan dosis maupun interval menurut usia gestasi atau usia pascalahir. Penyesuaian menurut usia ditampilkan pada kolom Literatur Terkini.",
    ]
    sumber = "SPO Sepsis Neonatal RSUP Dr. M. Djamil Padang (2017); PNPK Sepsis pada Anak, Kemenkes RI (2021)"
    return dosis_min_kg, dosis_max_kg, interval_min, interval_max, interval_ga, catatan, sumber


def rs_vankomisin(pma_minggu, pna_hari, scr=None, uop=None):
    """VANKOMISIN -- Bagian 1a: Nomogram Pedoman RS.
    Karena SPO RSUP Dr. M. Djamil TIDAK mencantumkan Vankomisin, basis
    'pedoman RS' untuk Vankomisin memakai UNC Medical Center Vancomycin
    Dosing & Monitoring Guide: Neonatal & Pediatric (Sep 2023)."""
    gangguan_ginjal = kriteria_gangguan_ginjal(scr, uop)
    sumber = "UNC Medical Center - Vancomycin Dosing & Monitoring Guide: Neonatal & Pediatric (Sep 2023)"

    if gangguan_ginjal:
        catatan = [
            "PASIEN MEMENUHI KRITERIA GANGGUAN GINJAL (SCr ≥ 1 mg/dL, atau SCr ≥ 0,5 mg/dL dengan produksi urin < 1 mL/kg/jam).",
            "Berikan dosis TUNGGAL 10 mg/kg, periksa kadar 12-24 jam pasca infus (boleh lebih awal, mis. 8 jam, bila indikasi klinis mendesak).",
            "Tahan dosis berikutnya sampai hasil kadar tersedia; sesuaikan dosis/interval lanjutan berdasarkan kadar tersebut.",
        ]
        return 10.0, 10.0, None, None, None, catatan, sumber

    # Interval "mempertimbangkan usia gestasi dan usia pascalahir" (nomogram UNC berbasis PMA & usia pascalahir)
    # Batas pita memakai minggu LENGKAP: '<=29 minggu' berarti PMA < 30 minggu.
    if pma_minggu < 30:
        interval_ga = 18 if pna_hari <= 14 else 12
    elif pma_minggu < 37:
        interval_ga = 12 if pna_hari <= 14 else 8
    elif pma_minggu < 45:
        interval_ga = 12 if pna_hari <= 7 else 8
    else:
        interval_ga = 8
    # Versi "tanpa usia gestasi dan usia pascalahir" (versi utama) = KRITERIA PEDOMAN RS untuk vankomisin:
    # 10-15 mg/kg/dosis; tiap 12 jam pada usia pascalahir 0-14 hari, tiap 8 jam pada 15-28 hari.
    # (Revisi 2026.11.01: sebelumnya 15 mg/kg dengan rentang tiap 8-18 jam.)
    interval_min = interval_max = (12 if pna_hari <= 14 else 8)
    catatan = ["Dosis 10-15 mg/kg/dosis; interval tiap 12 jam (usia pascalahir 0-14 hari) atau tiap 8 jam (15-28 hari) sesuai kriteria Pedoman RS (UNC 2023).",
               "Target kadar trough 10-20 mg/L (8-12 mg/L bila indikasi febrile neutropenia)."]
    return 10.0, 15.0, interval_min, interval_max, interval_ga, catatan, sumber


# ============================================================================
# BAGIAN 1b: PEDOMAN RS -- KRITERIA KESESUAIAN (rentang dosis & frekuensi)
# Berlaku untuk SEMUA neonatus usia pascalahir 0-28 hari (revisi 2026.11.01;
# sebelumnya dibatasi pada bayi cukup bulan dengan ginjal normal).
# Frekuensi dinyatakan sebagai RENTANG interval (interval_min-interval_max jam).
# ============================================================================

def rs_kriteria_kesesuaian(antibiotik, pna_hari):
    """Mengembalikan dict berisi rentang dosis (mg/kg) & rentang interval (jam)
    yang dianggap TEPAT menurut kriteria kesesuaian Pedoman RS."""
    sumber = "Kriteria kesesuaian dosis & frekuensi Pedoman RS (semua neonatus usia pascalahir 0-28 hari)"
    if pna_hari > USIA_PASCALAHIR_MAKS_HARI:
        raise ValueError("Usia pascalahir di luar cakupan kalkulator (0-28 hari).")
    if antibiotik == "Gentamisin":
        return {
            "dosis_min": 5.0, "dosis_max": 7.5, "satuan": "mg/kg per pemberian",
            "interval_min": 24, "interval_max": 48,
            "catatan": "Dosis dianggap tepat bila 5-7,5 mg/kg per pemberian; frekuensi dianggap tepat bila interval 24-48 jam (termasuk 36 jam).",
            "sumber": sumber,
        }
    elif antibiotik == "Amikasin":
        return {
            "dosis_min": 15.0, "dosis_max": 15.0, "satuan": "mg/kg per pemberian",
            "interval_min": 24, "interval_max": 24,
            "catatan": "Dosis dianggap tepat HANYA bila 15 mg/kg per pemberian; frekuensi dianggap tepat HANYA bila tiap 24 jam.",
            "sumber": sumber,
        }
    elif antibiotik == "Vankomisin":
        if pna_hari <= 14:
            return {"dosis_min": 10.0, "dosis_max": 15.0, "satuan": "mg/kg per pemberian",
                    "interval_min": 12, "interval_max": 12,
                    "catatan": "Usia pascalahir 0-14 hari: dosis tepat 10-15 mg/kg per pemberian; frekuensi tepat tiap 12 jam.",
                    "sumber": sumber}
        return {"dosis_min": 10.0, "dosis_max": 15.0, "satuan": "mg/kg per pemberian",
                "interval_min": 8, "interval_max": 8,
                "catatan": "Usia pascalahir 15-28 hari: dosis tepat 10-15 mg/kg per pemberian; frekuensi tepat tiap 8 jam.",
                "sumber": sumber}
    raise ValueError("Antibiotik tidak dikenali.")


# ============================================================================
# BAGIAN 2a: LITERATUR -- ANMF 2024 (nomogram PMA/usia pascalahir)
# Sumber: Australasian Neonatal Medicines Formulary, diakses 8 Juli 2026.
# ============================================================================

def lit_gentamisin(pma_minggu, hipotermia=False, cox_inhibitor=False):
    """GENTAMISIN -- ANMF v4.0, 24 April 2024."""
    dosis_mg_per_kg = 5.0
    if pma_minggu < 30:
        interval = 48
    elif pma_minggu < 35:
        interval = 36
    else:
        interval = 24
    catatan = []
    if hipotermia:
        interval = max(interval, 36)   # tidak pernah memperpendek interval bayi sangat prematur
        catatan.append("Hipotermia terapeutik: gunakan interval 36 jam; periksa kadar trough sebelum SETIAP dosis.")
    elif cox_inhibitor:
        interval += 12
        catatan.append("Terapi indometasin/ibuprofen (COX-inhibitor): interval AWAL diperpanjang 12 jam.")
    catatan.append("Interval di atas adalah interval AWAL; ANMF merekomendasikan penyesuaian berdasarkan kadar gentamisin 22 jam pasca dosis ke-2 (target trough < 2 mg/L, peak 5-12 mg/L).")
    sumber = "Australasian Neonatal Medicines Formulary (ANMF) - Gentamicin, v4.0, 24 April 2024"
    return dosis_mg_per_kg, interval, catatan, sumber


def lit_amikasin(pma_minggu, pna_hari, hipotermia=False, cox_inhibitor=False):
    """AMIKASIN -- ANMF v3.0, 2 Mei 2024."""
    # Batas pita memakai minggu LENGKAP: '<=29 minggu' berarti PMA < 30 minggu.
    if pma_minggu < 30:
        if pna_hari <= 7:
            dosis_mg_per_kg, interval = 14.0, 48
        elif pna_hari <= 28:
            dosis_mg_per_kg, interval = 12.0, 36
        else:
            dosis_mg_per_kg, interval = 12.0, 24
    elif pma_minggu < 35:
        dosis_mg_per_kg, interval = (12.0, 36) if pna_hari <= 7 else (12.0, 24)
    else:
        dosis_mg_per_kg, interval = 12.0, 24
    catatan = []
    if hipotermia or cox_inhibitor:
        interval += 12
        catatan.append("Hipotermia terapeutik dan/atau COX-inhibitor: interval diperpanjang 12 jam.")
    catatan.append("Target peak 20-35 mg/L (hingga 40 mg/L pada infeksi berat), target trough < 5 mg/L.")
    sumber = "Australasian Neonatal Medicines Formulary (ANMF) - Amikacin, v3.0, 2 Mei 2024"
    return dosis_mg_per_kg, interval, catatan, sumber


def lit_vankomisin(pma_minggu, pna_hari, sepsis_berat=False, hipotermia=False):
    """VANKOMISIN -- ANMF v3.0, errata 21 Nov 2024."""
    dosis_mg_per_kg = 15.0
    if pma_minggu < 30:
        interval = 18 if pna_hari <= 2 else 12
    elif pma_minggu < 37:
        interval = 12 if pna_hari <= 14 else 8
    elif pma_minggu < 45:
        interval = 12 if pna_hari <= 7 else 8
    else:
        interval = 6
    catatan = ["Target trough 10-20 mg/L; peak (jika diperiksa) 20-40 mg/L."]
    if sepsis_berat:
        catatan.append("Sepsis berat/dicurigai MRSA, infeksi tulang, meningitis, endokarditis: pertimbangkan loading dose 20 mg/kg (bukti pada neonatus terbatas).")
    if hipotermia:
        catatan.append("Hipotermia terapeutik: periksa trough SEBELUM dosis ke-2, tunggu hasil sebelum dosis berikutnya.")
    sumber = "Australasian Neonatal Medicines Formulary (ANMF) - Vancomycin Intermittent, v3.0 (errata 21 Nov 2024)"
    return dosis_mg_per_kg, interval, catatan, sumber


# ============================================================================
# BAGIAN 2b: LITERATUR -- Drugs.com Professional Monograph (ASHP/AHFS DI,
# dengan rujukan AAP & IDSA). Direpresentasikan sbg RENTANG (dosis_min,
# dosis_max) & (interval_min, interval_max) karena sumber ini sering memberi
# rentang, bukan satu angka. Diakses & diverifikasi 9-10 Juli 2026.
# ============================================================================

def dc_gentamisin(berat_kg, pna_hari):
    """GENTAMISIN -- Drugs.com/ASHP monograph (reviewed 30 Jun 2025),
    rekomendasi AAP berbasis BERAT BADAN."""
    if pna_hari <= 7:
        if berat_kg < 1.2:
            dosis, interval = (2.5, 2.5), (18, 24)
        else:
            dosis, interval = (2.5, 2.5), (12, 12)
    elif pna_hari <= 28:
        if berat_kg < 1.2:
            dosis, interval = (2.5, 2.5), (18, 24)
        elif berat_kg <= 2.0:
            dosis, interval = (2.5, 2.5), (8, 12)
        else:
            dosis, interval = (2.5, 2.5), (8, 8)
    else:
        dosis, interval = (2.5, 2.5), (8, 8)
    catatan = [
        "AAP: dosis berbasis BERAT BADAN (bukan usia gestasi) untuk neonatus <=4 minggu.",
        "Manufaktur (label FDA): 2,5 mg/kg tiap 12 jam (neonatus <=1 minggu), tiap 8 jam (neonatus lebih tua) -- pola dosis lebih sering dibanding pendekatan interval-diperpanjang.",
        "Target peak 4-12 mcg/mL; trough tidak melebihi 1-2 mcg/mL.",
    ]
    sumber = "Drugs.com Professional Monograph (ASHP/AHFS DI Essentials, rev. 30 Jun 2025), rekomendasi AAP"
    return dosis, interval, catatan, sumber


def dc_amikasin(berat_kg, pna_hari):
    """AMIKASIN -- Drugs.com/ASHP monograph, rekomendasi AAP (<1 minggu,
    berbasis berat badan) dan regimen sekali-sehari manufaktur."""
    if pna_hari <= 7:
        if berat_kg < 1.2:
            dosis, interval = (7.5, 7.5), (18, 24)
        elif berat_kg <= 2.0:
            dosis, interval = (7.5, 7.5), (12, 12)
        else:
            dosis, interval = (7.5, 10.0), (12, 12)
        catatan = ["AAP: dosis berbasis BERAT BADAN untuk neonatus <=1 minggu."]
    else:
        dosis, interval = (15.0, 15.0), (24, 24)
        catatan = ["Regimen sekali-sehari (once-daily) manufaktur untuk neonatus cukup bulan; AAP tidak merinci tabel berat badan untuk usia >1 minggu pada sumber ini."]
    catatan.append("Duration of therapy: 7-10 hari; keamanan pemberian >14 hari belum ditetapkan.")
    sumber = "Drugs.com Professional Monograph (ASHP/AHFS DI Essentials), rekomendasi AAP & manufaktur"
    return dosis, interval, catatan, sumber


def dc_vankomisin(ga_lahir_minggu, pna_hari, scr=None):
    """VANKOMISIN -- Drugs.com/ASHP monograph, tabel AAP berbasis usia
    gestasi (cGA) DAN kreatinin serum (SCr), neonatus <=28 hari pascalahir.
    Bila SCr tidak diinput, diasumsikan berada pada kategori SCr terendah
    (paling normal) -- HASIL PERLU DIVERIFIKASI ULANG bila data SCr tersedia.
    """
    catatan = ["AAP: dosis muatan (loading dose) 20 mg/kg IV sekali sebelum dosis rumatan berikut."]
    asumsi_scr = False
    if scr is None:
        asumsi_scr = True
        scr_pakai = 0.3  # asumsi kategori SCr terendah/normal
    else:
        scr_pakai = scr

    if pna_hari <= 28:
        if ga_lahir_minggu <= 28:
            if scr_pakai < 0.5:
                dosis, interval = (15.0, 15.0), (12, 12)
            elif scr_pakai <= 0.7:
                dosis, interval = (20.0, 20.0), (24, 24)
            elif scr_pakai <= 1.0:
                dosis, interval = (15.0, 15.0), (24, 24)
            elif scr_pakai <= 1.4:
                dosis, interval = (10.0, 10.0), (24, 24)
            else:
                dosis, interval = (15.0, 15.0), (48, 48)
        else:
            if scr_pakai < 0.7:
                dosis, interval = (15.0, 15.0), (12, 12)
            elif scr_pakai <= 0.9:
                dosis, interval = (20.0, 20.0), (24, 24)
            elif scr_pakai <= 1.2:
                dosis, interval = (15.0, 15.0), (24, 24)
            elif scr_pakai <= 1.6:
                dosis, interval = (10.0, 10.0), (24, 24)
            else:
                dosis, interval = (15.0, 15.0), (48, 48)
    else:
        # >= 28 hari (di luar tabel GA+SCr AAP untuk neonatus)
        dosis, interval = (11.25, 20.0), (6, 8)
        catatan.append("Usia > 28 hari: AAP merekomendasikan 45-60 mg/kg/hari terbagi 3-4 dosis (di luar tabel usia-gestasi+SCr untuk neonatus).")

    if asumsi_scr:
        catatan.append("Data SCr tidak diinput -- hasil ASUMSI kategori SCr ternormal/terendah. Masukkan data fungsi ginjal untuk hasil yang lebih akurat.")
    sumber = "Drugs.com Professional Monograph (ASHP/AHFS DI Essentials), tabel AAP berbasis usia gestasi & SCr"
    return dosis, interval, catatan, sumber


# ============================================================================
# BAGIAN 2c: CATATAN PUSTAKA TAMBAHAN (kualitatif, tidak dihitung otomatis)
# WHO (2024), Swiss Society of Neonatology (2024), JAID/JSC (2021),
# ACOG (2020) & AAP -- sumber ini menitikberatkan pada PEMILIHAN antibiotik
# dan STRATEGI tata laksana, bukan nomogram dosis neonatal yang rinci untuk
# ketiga obat ini.
# ============================================================================

CATATAN_PUSTAKA_TAMBAHAN = {
    "Gentamisin": [
        ("WHO (2024)", "Ampisilin/Penisilin + Gentamisin adalah regimen empiris lini pertama untuk sepsis pada bayi 0-59 hari; dosis yang dirujuk 5-7,5 mg/kg/kali, durasi minimal 10 hari. Tidak memuat nomogram interval berbasis usia gestasi."),
        ("Swiss Society of Neonatology (2024)", "Amoksisilin + aminoglikosida (gentamisin) tetap pilihan pertama; pedoman ini menitikberatkan KAPAN memulai/menghentikan terapi (pendekatan berbasis probabilitas & observasi klinis berkala), bukan rincian dosis."),
        ("JAID/JSC (2021)", "Pedoman umum sepsis & infeksi aliran darah terkait kateter (predominan dewasa); tidak memuat tabel dosis gentamisin neonatal yang rinci."),
        ("ACOG (2020) / AAP", "ACOG mengatur profilaksis antibiotik intrapartum IBU untuk pencegahan GBS (penisilin G/ampisilin), bukan dosis neonatal. AAP (Puopolo dkk.) merekomendasikan Ampisilin+Gentamisin sebagai regimen empiris tanpa merinci dosis mg/kg (merujuk ke referensi dosis standar)."),
    ],
    "Amikasin": [
        ("WHO (2024)", "Amikasin bukan bagian regimen inti WHO (fokus WHO: Ampisilin/Penisilin + Gentamisin); di beberapa wilayah dengan resistensi gentamisin tinggi, amikasin dipakai sebagai alternatif lini pertama."),
        ("Swiss Society of Neonatology (2024)", "Tidak merinci amikasin secara spesifik; fokus pedoman pada keputusan memulai/menghentikan terapi empiris umum."),
        ("JAID/JSC (2021)", "Tidak memuat tabel dosis amikasin neonatal yang rinci."),
        ("ACOG (2020) / AAP", "Tidak relevan langsung untuk dosis amikasin neonatal (ACOG berfokus pada profilaksis GBS pada ibu)."),
    ],
    "Vankomisin": [
        ("WHO (2024)", "Vankomisin BUKAN bagian regimen inti WHO untuk sepsis neonatal di layanan primer/rujukan terbatas (fokus pada antibiotik esensial biaya-efektif); dipakai selektif untuk kecurigaan MRSA/CoNS di fasilitas rujukan."),
        ("Swiss Society of Neonatology (2024)", "Tidak merinci vankomisin; fokus pedoman pada E. coli dan Group B Streptococcus sebagai patogen utama EOS."),
        ("JAID/JSC (2021)", "Merekomendasikan vankomisin dikombinasikan pada kecurigaan pneumokokus resisten-laktam atau MRSA, tanpa tabel dosis neonatal spesifik."),
        ("ACOG (2020) / AAP", "Vankomisin adalah salah satu pilihan profilaksis intrapartum IBU bila alergi penisilin berat & GBS resisten klindamisin (bukan dosis neonatal). Vankomisin neonatal umumnya dipakai untuk sepsis awitan lambat/dugaan MRSA, di luar fokus utama pedoman EOS awitan dini ACOG/AAP."),
    ],
}

print("Basis data & fungsi perhitungan dosis berhasil dimuat.")
```

```python
# ============================================================================
# FUNGSI BANTU: menggabungkan seluruh basis data & menampilkannya
# ============================================================================

def _rentang_dosis_rs(rs_dosis_per_kg, is_term, ginjal_normal, rs_kriteria):
    """RENTANG dosis (mg/kg) Pedoman RS:
    - Bila Kriteria Kesesuaian berlaku (cukup bulan & ginjal normal): pakai
      rentang [dosis_min, dosis_max] dari kriteria tsb (ini ADALAH rentang
      resmi pedoman RS untuk profil pasien ini).
    - Bila TIDAK berlaku (prematur / gangguan ginjal): kriteria rentang tidak
      tersedia untuk profil ini, jadi dipakai nilai nomogram tunggal sebagai
      rentang berimpit (min==max), ditandai sebagai 'nilai tunggal' pada tampilan.
    """
    if rs_dosis_per_kg is None:
        return None, None, False
    if is_term and ginjal_normal:
        return rs_kriteria["dosis_min"], rs_kriteria["dosis_max"], True
    return rs_dosis_per_kg, rs_dosis_per_kg, False


def _rentang_dosis_lit(anmf_dosis_per_kg, dc_dosis_rentang):
    """RENTANG dosis (mg/kg) Literatur Terkini: gabungan dari titik ANMF dan
    rentang Drugs.com/ASHP -- diambil batas TERLUAS dari keduanya, sehingga
    rentang yang ditampilkan mencerminkan seluruh angka yang dikutip dari
    kedua sumber literatur (bukan memilih salah satu secara sepihak)."""
    dmin, dmax = dc_dosis_rentang
    if anmf_dosis_per_kg is None:
        return dmin, dmax
    if not GABUNG_DRUGSCOM_KE_LITERATUR:
        return anmf_dosis_per_kg, anmf_dosis_per_kg
    return min(anmf_dosis_per_kg, dmin), max(anmf_dosis_per_kg, dmax)


def bangun_rekomendasi(antibiotik, ga_lahir, pna_hari, berat_kg, scr=None, uop=None,
                        hipotermia=False, cox_inhibitor=False, sepsis_berat=False, kode_pasien=""):
    """Menghitung rekomendasi dari SELURUH basis rujukan sekaligus."""
    if pna_hari is None or pna_hari < 0 or pna_hari > USIA_PASCALAHIR_MAKS_HARI:
        raise ValueError("Usia pascalahir di luar cakupan kalkulator (0-28 hari). "
                         "Gunakan pedoman dosis bayi di atas 28 hari.")
    pma = hitung_pma(ga_lahir, pna_hari)
    hasil = {"pma_minggu": round(pma, 1), "ga_lahir": ga_lahir, "pna_hari": pna_hari,
             "berat_kg": berat_kg, "kode_pasien": kode_pasien, "antibiotik": antibiotik}

    if antibiotik == "Gentamisin":
        rd1_min, rd1_max, ri1_min, ri1_max, ri1_ga, c1, s1 = rs_gentamisin(ga_lahir, pna_hari)
        d2, i2, c2, s2 = lit_gentamisin(pma, hipotermia, cox_inhibitor)
        d3, i3, c3, s3 = dc_gentamisin(berat_kg, pna_hari)
        if kriteria_gangguan_ginjal(scr, uop):
            peringatan_ginjal = "Kemungkinan gangguan fungsi ginjal (SCr ≥ 1 mg/dL, atau SCr ≥ 0,5 mg/dL dengan produksi urin < 1 mL/kg/jam). Pertimbangkan pemeriksaan kadar trough sebelum SETIAP dosis dan perpanjangan interval -- kalkulator ini TIDAK menghitung penyesuaian ginjal otomatis untuk Gentamisin."
            hasil["peringatan_ginjal"] = peringatan_ginjal; c3.append(peringatan_ginjal)
    elif antibiotik == "Amikasin":
        rd1_min, rd1_max, ri1_min, ri1_max, ri1_ga, c1, s1 = rs_amikasin(ga_lahir, pna_hari)
        d2, i2, c2, s2 = lit_amikasin(pma, pna_hari, hipotermia, cox_inhibitor)
        d3, i3, c3, s3 = dc_amikasin(berat_kg, pna_hari)
        if kriteria_gangguan_ginjal(scr, uop):
            peringatan_ginjal = "Kemungkinan gangguan fungsi ginjal. Pertimbangkan perpanjangan interval 12-24 jam dan pemeriksaan kadar trough sebelum tiap dosis -- kalkulator ini TIDAK menghitung penyesuaian ginjal otomatis untuk Amikasin."
            hasil["peringatan_ginjal"] = peringatan_ginjal; c3.append(peringatan_ginjal)
    elif antibiotik == "Vankomisin":
        rd1_min, rd1_max, ri1_min, ri1_max, ri1_ga, c1, s1 = rs_vankomisin(pma, pna_hari, scr, uop)
        d2, i2, c2, s2 = lit_vankomisin(pma, pna_hari, sepsis_berat, hipotermia)
        d3, i3, c3, s3 = dc_vankomisin(ga_lahir, pna_hari, scr)
    else:
        raise ValueError("Antibiotik tidak dikenali.")
    if antibiotik == "Vankomisin" and kriteria_gangguan_ginjal(scr, uop):
        hasil["peringatan_ginjal"] = ("Kemungkinan gangguan fungsi ginjal (SCr ≥ 1 mg/dL, atau SCr ≥ 0,5 mg/dL dengan produksi urin < 1 mL/kg/jam). "
                                      "Pedoman RS (UNC) memberikan dosis TUNGGAL 10 mg/kg tanpa interval rutin; dosis berikutnya ditentukan berdasarkan kadar vankomisin dalam darah.")

    # Nilai dosis "representatif" (batas bawah) untuk kolom mg/kg ringkas & kompatibilitas.
    d1 = rd1_min
    # Versi PRIMER = TANPA usia gestasi &amp; pascalahir (rentang interval), sesuai bunyi SPO yang tidak
    # menstratifikasi usia. Frekuensi = rentang [ri1_min, ri1_max].
    hasil["rs"] = {"dosis_mg_per_kg": d1, "dosis_min_kg": rd1_min, "dosis_max_kg": rd1_max,
                    "interval_jam": ri1_min, "interval_min": ri1_min, "interval_max": ri1_max,
                    "catatan": c1, "sumber": s1,
                    "dosis_mg_total": None if d1 is None else round(d1 * berat_kg, 4)}
    # Versi SEKUNDER = DENGAN usia gestasi &amp; pascalahir (interval tunggal via nomogram). Dosis sama
    # (SPO tidak memvariasikan dosis menurut usia); yang berbeda hanya frekuensi.
    hasil["rs_ga"] = {"dosis_min_kg": rd1_min, "dosis_max_kg": rd1_max,
                       "interval_jam": ri1_ga, "sumber": s1,
                       "dosis_mg_total": None if d1 is None else round(d1 * berat_kg, 4)}
    hasil["anmf"] = {"dosis_mg_per_kg": d2, "interval_jam": i2, "catatan": c2, "sumber": s2,
                      "dosis_mg_total": None if d2 is None else round(d2 * berat_kg, 4)}
    hasil["drugscom"] = {"dosis_rentang": d3, "interval_rentang": i3, "catatan": c3, "sumber": s3,
                          "dosis_total_rentang": (round(d3[0]*berat_kg, 4), round(d3[1]*berat_kg, 4))}
    hasil["rs_kriteria"] = rs_kriteria_kesesuaian(antibiotik, pna_hari)
    hasil["pustaka_tambahan"] = CATATAN_PUSTAKA_TAMBAHAN[antibiotik]
    hasil["is_term"] = ga_lahir >= 37
    hasil["ginjal_normal"] = not kriteria_gangguan_ginjal(scr, uop)
    # Revisi 2026.11.01: bila SCr tidak diisi, fungsi ginjal DIASUMSIKAN normal
    # sesuai usia dan hasil diberi catatan (lihat _blok_data_ginjal).
    hasil["data_ginjal_tersedia"] = scr is not None
    hasil["risiko_ginjal_tinggi"] = (ga_lahir < 29 or berat_kg < 1.5 or hipotermia or cox_inhibitor
                                     or antibiotik == "Vankomisin")
    hasil["versi_aturan"] = VERSI_ATURAN
    hasil["scr"], hasil["uop"] = scr, uop
    hasil["hipotermia"], hasil["cox_inhibitor"] = hipotermia, cox_inhibitor

    # RENTANG dosis Pedoman RS diambil langsung dari fungsi rs_* (tidak lagi
    # bergantung pada usia gestasi/pascalahir untuk Gentamisin & Amikasin).
    rs_min_kg, rs_max_kg = rd1_min, rd1_max
    rs_dari_kriteria = (rd1_min is not None and rd1_min != rd1_max)
    lit_min_kg, lit_max_kg = _rentang_dosis_lit(d2, d3)
    hasil["rs_rentang_per_kg"] = (rs_min_kg, rs_max_kg)
    hasil["rs_rentang_dari_kriteria"] = rs_dari_kriteria
    hasil["rs_rentang_mg"] = (None, None) if rs_min_kg is None else (round(rs_min_kg*berat_kg, 4), round(rs_max_kg*berat_kg, 4))
    hasil["lit_rentang_per_kg"] = (lit_min_kg, lit_max_kg)
    hasil["lit_rentang_mg"] = (round(lit_min_kg*berat_kg, 4), round(lit_max_kg*berat_kg, 4))
    # RENTANG FREKUENSI (interval jam) Literatur = gabungan terluas interval ANMF
    # (titik) dan rentang interval Drugs.com/ASHP.
    _dc_imin, _dc_imax = i3
    if i2 is None:
        lit_i_min, lit_i_max = _dc_imin, _dc_imax
    elif not GABUNG_DRUGSCOM_KE_LITERATUR:
        lit_i_min, lit_i_max = i2, i2
    else:
        lit_i_min, lit_i_max = min(i2, _dc_imin), max(i2, _dc_imax)
    hasil["lit_interval_rentang"] = (lit_i_min, lit_i_max)
    return hasil


def _fmt_num(x):
    """Format angka tanpa pembulatan agresif: bilangan bulat tampil tanpa desimal
    (20 bukan 20.0), selain itu hingga 3 angka desimal (presisi gram untuk berat,
    mis. 1,735 kg) dengan angka nol di belakang dibuang (mis. 5.187; 7.5; 8.675).
    Perhitungan tetap memakai nilai penuh; ini hanya format tampilan."""
    if x is None:
        return "-"
    xf = float(x)
    if abs(xf - round(xf)) < 1e-9:
        return str(int(round(xf)))
    s = ("%.3f" % xf).rstrip("0").rstrip(".")
    return s.replace(".", ",")   # tampilan memakai koma desimal (format Indonesia)


def _fmt_usia_minggu(minggu):
    """32,571 minggu -> '32 minggu 4 hari' (minggu lengkap + sisa hari)."""
    total_hari = int(round(float(minggu) * 7))
    w, d = divmod(total_hari, 7)
    return str(w) + " minggu" + ((" " + str(d) + " hari") if d else "")


def _fmt_q(interval_jam):
    """Format interval sebagai notasi farmasi qXh; dosis tunggal ditulis eksplisit."""
    if interval_jam is None:
        return "Dosis tunggal (bukan rutin)"
    return "tiap " + str(interval_jam) + " jam"


def _fmt_q_rentang(imin, imax):
    """Format frekuensi yang bisa berupa RENTANG jam (mis. RS gentamisin 24-48).
    imin==imax -> qXh; keduanya None -> dosis tunggal; selain itu qMIN-qMAXh."""
    if imin is None and imax is None:
        return "Dosis tunggal (bukan rutin)"
    if imin == imax:
        return "tiap " + str(imin) + " jam"
    return "tiap " + str(imin) + "-" + str(imax) + " jam"


def _cmp_interval_rentang(nilai, imin, imax):
    """Bandingkan interval aktual/manual terhadap rentang interval rekomendasi
    (mis. RS gentamisin 24-48). Sesuai bila di dalam rentang; Lebih pendek bila
    < batas bawah; Lebih panjang bila > batas atas."""
    if imin is None or imax is None or nilai is None:
        return "info", "Tidak ada interval rutin untuk dibandingkan."
    if imin <= nilai <= imax:
        if imin == imax:
            return "ok", "Sesuai (" + _fmt_q(nilai) + ")."
        return "ok", "Sesuai (dalam rentang " + _fmt_q_rentang(imin, imax) + ")."
    if nilai < imin:
        return "warn", "LEBIH PENDEK dari " + _fmt_q_rentang(imin, imax) + " -- risiko akumulasi/toksisitas."
    return "warn", "LEBIH PANJANG dari " + _fmt_q_rentang(imin, imax) + " -- risiko kadar subterapeutik."


def _fmt_rentang(a, b, satuan=""):
    if a == b:
        return _fmt_num(a) + (" " + satuan if satuan else "")
    return _fmt_num(a) + "-" + _fmt_num(b) + (" " + satuan if satuan else "")


def _baris(label, nilai):
    return "<tr><td style=\'padding:4px 10px;color:#555;\'>" + label + "</td><td style=\'padding:4px 10px;font-weight:600;\'>" + str(nilai) + "</td></tr>"


TH_STYLE = "padding:8px 12px;text-align:left;background:#e9ecef;border:1px solid #ccc;"
TD_STYLE = "padding:8px 12px;text-align:left;border:1px solid #ccc;"


def _tabel_rekomendasi(hasil):
    """TABEL UTAMA Bagian 2: Parameter | Pedoman Rumah Sakit | Literatur Terkini.
    Dosis ditampilkan sebagai RENTANG TOTAL mg (hasil hitungan x berat badan
    pasien), karena kedua basis rujukan bisa berupa rentang dosis (bukan dosis
    tetap). Bila rentang berimpit (min==max, mis. Amikasin RS pada bayi
    cukup bulan & ginjal normal, atau bayi prematur yang hanya punya nilai
    nomogram tunggal), ditampilkan sebagai satu nilai. Frekuensi tetap
    notasi qXh (titik, dari nomogram/ANMF)."""
    rs, anmf = hasil["rs"], hasil["anmf"]
    rs_min, rs_max = hasil["rs_rentang_mg"]
    lit_min, lit_max = hasil["lit_rentang_mg"]
    rs_min_kg, rs_max_kg = hasil["rs_rentang_per_kg"]
    lit_min_kg, lit_max_kg = hasil["lit_rentang_per_kg"]

    def sel_dosis_rentang(mg_min, mg_max, kg_min, kg_max):
        if mg_min is None:
            return "-"
        if mg_min == mg_max:
            teks_mg, teks_kg = _fmt_num(mg_max) + " mg", _fmt_num(kg_max) + " mg/kg"
        else:
            teks_mg = _fmt_num(mg_min) + "-" + _fmt_num(mg_max) + " mg"
            teks_kg = _fmt_num(kg_min) + "-" + _fmt_num(kg_max) + " mg/kg"
        return ("<b>" + teks_mg + "</b><br>"
                "<span style=\'color:#777;font-size:0.85em;\'>(" + teks_kg + ")</span>")

    def sel_freq_rs():
        return "<b>" + _fmt_q_rentang(rs.get("interval_min"), rs.get("interval_max")) + "</b>"

    def sel_freq(data):
        return "<b>" + _fmt_q(data["interval_jam"]) + "</b>"

    lit_imin, lit_imax = hasil.get("lit_interval_rentang", (None, None))

    def sel_freq_lit():
        return "<b>" + _fmt_q_rentang(lit_imin, lit_imax) + "</b>"

    return ("<table style=\'border-collapse:collapse;width:100%;margin:10px 0;\'>"
            "<tr>"
            "<th style=\'" + TH_STYLE + "\'>Parameter</th>"
            "<th style=\'" + TH_STYLE + "\'>Pedoman Rumah Sakit</th>"
            "<th style=\'" + TH_STYLE + "\'>Literatur Terkini</th>"
            "</tr>"
            "<tr>"
            "<td style=\'" + TD_STYLE + "\'>Dosis per pemberian</td>"
            "<td style=\'" + TD_STYLE + "\'>" + sel_dosis_rentang(rs_min, rs_max, rs_min_kg, rs_max_kg) + "</td>"
            "<td style=\'" + TD_STYLE + "\'>" + sel_dosis_rentang(lit_min, lit_max, lit_min_kg, lit_max_kg) + "</td>"
            "</tr>"
            "<tr>"
            "<td style=\'" + TD_STYLE + "\'>Frekuensi</td>"
            "<td style=\'" + TD_STYLE + "\'>" + sel_freq_rs() + "</td>"
            "<td style=\'" + TD_STYLE + "\'>" + sel_freq_lit() + "</td>"
            "</tr>"
            "<tr>"
            "<td style=\'" + TD_STYLE + "\' colspan=\'3\'><span style=\'color:#777;font-size:0.82em;\'>Pedoman Rumah Sakit (SPO RSUP Dr. M. Djamil; vankomisin: kriteria UNC) tidak membedakan usia gestasi. Literatur Terkini (ANMF 2024) menyesuaikan interval menurut usia pascamenstruasi dan usia pascalahir.</span></td>"
            "</tr>"
            "</table>")


def _blok_rentang(judul, data, warna):
    """Kartu rujukan tambahan RENTANG (dosis_rentang & interval_rentang): Drugs.com/ASHP."""
    dmin, dmax = data["dosis_rentang"]
    imin, imax = data["interval_rentang"]
    tmin, tmax = data["dosis_total_rentang"]
    catatan_html = "".join("<li>" + c + "</li>" for c in data["catatan"])
    baris1 = _baris("Dosis", _fmt_rentang(dmin, dmax, "mg/kg/kali") + " (" + _fmt_rentang(tmin, tmax, "mg") + ")")
    baris2 = _baris("Interval", "setiap " + _fmt_rentang(imin, imax) + " jam")
    return ("<div style=\'border-left:4px solid " + warna + ";background:#f9f9f9;padding:10px 14px;margin:8px 0;border-radius:4px;\'>"
            + "<b>" + judul + "</b>"
            + "<table>" + baris1 + baris2 + "</table>"
            + "<ul style=\'margin:4px 0 0 0;\'>" + catatan_html + "</ul>"
            + "<div style=\'color:#777;font-size:0.85em;margin-top:4px;\'>Sumber: " + data["sumber"] + "</div>"
            + "</div>")


# ============================================================================
# SISTEM PERINGATAN BERWARNA & PENJELASAN SINGKATAN (revisi 2026.11.01)
# ============================================================================
import re as _re_sk

_TINGKAT_PERINGATAN = {
    # tingkat:   (latar,     garis/teks, label)
    "bahaya":    ("#f8d7da", "#b02a37", "PERINGATAN"),
    "perhatian": ("#fff3cd", "#9a6700", "PERHATIAN"),
    "info":      ("#e7f1ff", "#0b5ed7", "INFORMASI"),
    "sesuai":    ("#d1e7dd", "#146c43", "SESUAI"),
}
_URUTAN_TINGKAT = {"bahaya": 0, "perhatian": 1, "info": 2, "sesuai": 3}


def _kotak(tingkat, judul, isi=""):
    """Kotak pesan berwarna. Merah = peringatan, kuning = perhatian, biru = informasi, hijau = sesuai."""
    bg, garis, label = _TINGKAT_PERINGATAN[tingkat]
    return ("<div style='background:" + bg + ";border:1px solid " + garis + ";border-left:8px solid " + garis + ";"
            "padding:10px 14px;margin:8px 0;border-radius:4px;'>"
            "<div style='color:" + garis + ";font-weight:700;font-size:1.05em;'>" + label + ": " + judul + "</div>"
            + (("<div style='margin-top:4px;color:#222;'>" + isi + "</div>") if isi else "")
            + "</div>")


def _urutkan_peringatan(daftar):
    return sorted(daftar, key=lambda p: _URUTAN_TINGKAT[p[0]])


def _legenda_warna():
    def item(t, teks):
        return ("<span style='color:" + _TINGKAT_PERINGATAN[t][1] + ";font-weight:700;'>&#9632; "
                + _TINGKAT_PERINGATAN[t][2] + "</span> " + teks)
    return ("<div style='font-size:0.82em;color:#444;margin:8px 0;'><b>Arti warna:</b> "
            + item("bahaya", "(merah) perlu verifikasi/tindakan segera") + " &middot; "
            + item("perhatian", "(kuning) hasil memakai asumsi atau perlu dipertimbangkan") + " &middot; "
            + item("info", "(biru) informasi") + " &middot; "
            + item("sesuai", "(hijau) sesuai rujukan") + "</div>")


TEKS_ASUMSI_GINJAL = ("Kreatinin serum (SCr) dan produksi urin tidak diisi. Hasil dihitung dengan ASUMSI fungsi ginjal normal sesuai usia, "
                      "sehingga penyesuaian untuk gangguan ginjal tidak dapat dinilai. Rekomendasi ini bersifat sementara; "
                      "periksa kreatinin serum sebelum dosis ke-2.")
TEKS_RISIKO_GINJAL = ("Risiko gangguan ginjal tinggi pada pasien ini (usia gestasi &lt;29 minggu, berat &lt;1,5 kg, hipotermia terapeutik, "
                      "terapi indometasin/ibuprofen, atau vankomisin). Asumsi fungsi ginjal normal mungkin tidak tepat; "
                      "periksa kreatinin serum sesegera mungkin dan sebelum dosis ke-2.")


def _peringatan_pasien(hasil):
    """Daftar (tingkat, judul, isi) tentang kondisi pasien -- dipakai Rekomendasi, Evaluasi, dan PDF."""
    daftar = []
    if hasil.get("peringatan_ginjal"):
        daftar.append(("bahaya", "Kemungkinan gangguan fungsi ginjal", hasil["peringatan_ginjal"]))
    if not hasil.get("data_ginjal_tersedia", True):
        if hasil.get("risiko_ginjal_tinggi"):
            daftar.append(("bahaya", "Data fungsi ginjal tidak diisi pada pasien berisiko tinggi",
                           TEKS_ASUMSI_GINJAL + "<br>" + TEKS_RISIKO_GINJAL))
        else:
            daftar.append(("perhatian", "Data fungsi ginjal tidak diisi", TEKS_ASUMSI_GINJAL))
    if hasil.get("pma_minggu", 0) > 50:
        daftar.append(("bahaya", "Di luar rentang neonatus",
                       "PMA saat ini (" + str(hasil["pma_minggu"]) + " minggu) berada di luar rentang yang dicakup nomogram neonatus. "
                       "Pertimbangkan pedoman dosis pediatrik."))
    scr, uop = hasil.get("scr"), hasil.get("uop")
    if scr is not None and uop is None and 0.5 <= scr < 1.0:
        daftar.append(("perhatian", "Produksi urin tidak diisi",
                       "Kreatinin serum " + _fmt_num(scr) + " mg/dL berada pada rentang 0,5 sampai &lt;1 mg/dL. Menurut kriteria UNC, nilai ini "
                       "menandakan gangguan ginjal bila produksi urin &lt;1 mL/kg/jam. Karena produksi urin tidak diisi, kriteria tersebut "
                       "tidak dapat dinilai; isi produksi urin bila tersedia."))
    if hasil.get("peringatan_ginjal") and hasil.get("pna_hari", 99) <= 3:
        daftar.append(("perhatian", "Kreatinin serum pada 72 jam pertama kehidupan",
                       "Pada 3 hari pertama, kreatinin serum bayi masih dapat mencerminkan kreatinin ibu, sehingga nilai tinggi belum tentu "
                       "menandakan gangguan ginjal bayi. Nilai tren kreatinin serum dan produksi urin, dan diskusikan dengan DPJP."))
    if hasil.get("hipotermia") and hasil.get("ga_lahir", 99) < 35:
        daftar.append(("perhatian", "Periksa isian hipotermia terapeutik",
                       "Hipotermia terapeutik umumnya diberikan pada bayi dengan usia gestasi 35-36 minggu atau lebih. Pastikan isian ini benar."))
    return _urutkan_peringatan(daftar)


def _peringatan_rekomendasi(hasil):
    daftar = list(_peringatan_pasien(hasil))
    rs, anmf = hasil["rs"], hasil["anmf"]
    rs_imin, rs_imax = rs.get("interval_min"), rs.get("interval_max")
    if rs_imin is not None and anmf["interval_jam"] is not None:
        if anmf["interval_jam"] > rs_imax:
            daftar.append(("perhatian", "Interval Pedoman RS lebih pendek dari literatur",
                           "Batas atas interval Pedoman RS (" + _fmt_q(rs_imax) + ") lebih pendek dibanding ANMF ("
                           + _fmt_q(anmf["interval_jam"]) + ") untuk profil pasien ini. Pertimbangkan risiko akumulasi obat dan toksisitas; "
                           "diskusikan dengan DPJP/Tim Farmasi Klinis."))
        elif anmf["interval_jam"] < rs_imin:
            daftar.append(("perhatian", "Interval Pedoman RS lebih panjang dari literatur",
                           "Batas bawah interval Pedoman RS (" + _fmt_q(rs_imin) + ") lebih panjang dibanding ANMF ("
                           + _fmt_q(anmf["interval_jam"]) + "). Pastikan kadar obat tidak di bawah kadar terapi."))
    return _urutkan_peringatan(daftar)


# (singkatan, kepanjangan yang disisipkan di teks, keterangan tambahan untuk daftar singkatan)
SINGKATAN = [
    ("JAID/JSC", "Japanese Association for Infectious Diseases / Japanese Society of Chemotherapy", ""),
    ("RSUP", "Rumah Sakit Umum Pusat", ""),
    ("PNPK", "Pedoman Nasional Pelayanan Kedokteran", ""),
    ("ANMF", "Australasian Neonatal Medicines Formulary", "formularium obat neonatus Australia dan Selandia Baru"),
    ("ASHP", "American Society of Health-System Pharmacists", ""),
    ("AHFS", "American Hospital Formulary Service", ""),
    ("ACOG", "American College of Obstetricians and Gynecologists", ""),
    ("IDSA", "Infectious Diseases Society of America", ""),
    ("MRSA", "Staphylococcus aureus yang resisten metisilin", ""),
    ("CoNS", "stafilokokus koagulase-negatif", ""),
    ("DPJP", "dokter penanggung jawab pasien", ""),
    ("SPO", "Standar Prosedur Operasional", ""),
    ("PPK", "Panduan Praktik Klinis", ""),
    ("PMA", "usia pascamenstruasi", "usia gestasi saat lahir + usia pascalahir, dalam minggu"),
    ("SCr", "kreatinin serum", ""),
    ("UOP", "produksi urin", ""),
    ("HIE", "ensefalopati hipoksik-iskemik", ""),
    ("COX", "siklooksigenase", "indometasin dan ibuprofen adalah penghambat COX"),
    ("UNC", "University of North Carolina Medical Center", ""),
    ("WHO", "World Health Organization", "Organisasi Kesehatan Dunia"),
    ("AAP", "American Academy of Pediatrics", ""),
    ("GBS", "Streptokokus grup B", ""),
    ("EOS", "sepsis awitan dini", ""),
    ("FDA", "Food and Drug Administration", "badan pengawas obat Amerika Serikat"),
    ("TDM", "pemantauan kadar obat dalam darah", ""),
    ("IV", "intravena", ""),
    ("Kemenkes", "Kementerian Kesehatan", ""),
    ("RI", "Republik Indonesia", ""),
    ("RS", "Rumah Sakit", ""),
    ("GA", "usia gestasi", ""),
    ("PNA", "usia pascalahir", ""),
    ("UG", "usia gestasi", ""),
    ("UPL", "usia pascalahir", ""),
    ("BB", "berat badan", ""),
]


def _pola_singkatan(sk):
    return _re_sk.compile(r"(?<![A-Za-z0-9])" + _re_sk.escape(sk) + r"(?![A-Za-z0-9])")


def _legenda_singkatan(daftar=None):
    """Kotak 'Keterangan singkatan' untuk singkatan pada `daftar` (None = semua)."""
    data = [s for s in SINGKATAN if daftar is None or s[0] in daftar]
    if not data:
        return ""
    isi = "; ".join("<b>" + sk + "</b> = " + panjang + ((" (" + ket + ")") if ket else "") for sk, panjang, ket in data)
    return ("<div style='font-size:0.82em;color:#444;background:#f8f9fa;border:1px solid #dee2e6;padding:6px 10px;margin:8px 0;border-radius:4px;'>"
            "<b>Keterangan singkatan:</b> " + isi
            + ". Frekuensi ditulis sebagai &quot;tiap X jam&quot; (interval antardosis).</div>")


def _perjelas_singkatan(html, legenda=True):
    """Menyisipkan kepanjangan pada kemunculan PERTAMA setiap singkatan di teks hasil
    (mis. 'RS' -> 'Rumah Sakit (RS)'), lalu menambahkan kotak 'Keterangan singkatan'
    berisi semua singkatan yang muncul. Hanya teks yang diubah, bukan tag/atribut HTML."""
    bagian = _re_sk.split(r"(<[A-Za-z/!][^>]*>)", html)
    dipakai = []
    for sk, panjang, _ in SINGKATAN:
        pola = _pola_singkatan(sk)
        sudah = False
        for i, seg in enumerate(bagian):
            if not seg or seg.startswith("<") or not pola.search(seg):
                continue
            if sk not in dipakai:
                dipakai.append(sk)
            if sudah:
                continue
            sudah = True
            if _re_sk.search(r"\(\s*" + _re_sk.escape(sk) + r"\s*\)", seg):
                continue  # sudah dijelaskan di teks, mis. 'Kreatinin serum (SCr)'
            bagian[i] = pola.sub(lambda m: panjang + " (" + sk + ")", seg, count=1)
    hasil_html = "".join(bagian)
    if legenda and dipakai:
        hasil_html += _legenda_singkatan(dipakai)
    return hasil_html


def _catatan_cakupan_kriteria(hasil):
    """Revisi 2026.11.01: kriteria RS berlaku untuk semua neonatus 0-28 hari.
    Untuk bayi prematur / indikasi gangguan ginjal hanya ditampilkan catatan."""
    if hasil["is_term"] and hasil["ginjal_normal"]:
        return ""
    kondisi = []
    if not hasil["is_term"]:
        kondisi.append("prematur")
    if not hasil["ginjal_normal"]:
        kondisi.append("dengan indikasi gangguan ginjal")
    profil = " dan ".join(kondisi)
    return ("<div style='color:#856404;font-size:0.85em;margin-top:4px;'>"
            "Catatan: kriteria RS tidak membedakan usia gestasi maupun fungsi ginjal. Untuk pasien " + profil
            + " ini, bandingkan juga dengan kolom Literatur Terkini, yang menyesuaikan interval menurut usia."
            + (" Pada indikasi gangguan ginjal, kolom Pedoman RS vankomisin mengikuti aturan gangguan ginjal UNC (dosis tunggal 10 mg/kg; "
               "dosis berikutnya berdasarkan kadar obat), sehingga berbeda dari kriteria di atas."
               if hasil.get("antibiotik") == "Vankomisin" and not hasil["ginjal_normal"] else "")
            + "</div>")


def _blok_data_ginjal(hasil):
    """Kotak peringatan bila data fungsi ginjal tidak diisi (kuning; merah bila pasien berisiko tinggi)."""
    if hasil.get("data_ginjal_tersedia", True):
        return ""
    if hasil.get("risiko_ginjal_tinggi"):
        return _kotak("bahaya", "Data fungsi ginjal tidak diisi pada pasien berisiko tinggi",
                      TEKS_ASUMSI_GINJAL + "<br>" + TEKS_RISIKO_GINJAL)
    return _kotak("perhatian", "Data fungsi ginjal tidak diisi", TEKS_ASUMSI_GINJAL)


def _blok_kriteria(hasil):
    kr = hasil["rs_kriteria"]
    return ("<div style='border-left:4px solid #0d6efd;background:#eef6ff;padding:10px 14px;margin:8px 0;border-radius:4px;'>"
            + "<b>Kriteria Kesesuaian Pedoman RS</b> (rentang dosis & rentang frekuensi yang dianggap tepat)"
            + "<ul style='margin:4px 0 0 0;'>"
            + "<li>Dosis tepat: " + _fmt_rentang(kr["dosis_min"], kr["dosis_max"]) + " " + kr["satuan"] + "</li>"
            + "<li>Frekuensi tepat: tiap " + _fmt_rentang(kr["interval_min"], kr["interval_max"]) + " jam</li>"
            + "<li>" + kr["catatan"] + "</li>"
            + "</ul>"
            + "<div style='color:#777;font-size:0.85em;margin-top:4px;'>Sumber: " + kr["sumber"] + "</div>"
            + _catatan_cakupan_kriteria(hasil)
            + "</div>")


def _blok_pustaka_tambahan(hasil):
    items = "".join("<li><b>" + nama + ":</b> " + catatan + "</li>" for nama, catatan in hasil["pustaka_tambahan"])
    return ("<div style=\'border-left:4px solid #6c757d;background:#f4f4f4;padding:10px 14px;margin:8px 0;border-radius:4px;\'>"
            "<b>Catatan Pustaka Tambahan</b> <span style=\'color:#777;font-size:0.85em;\'>(referensi kualitatif, bukan nomogram dosis rinci)</span>"
            "<ul style=\'margin:6px 0 0 0;\'>" + items + "</ul></div>")


def _baris_disclaimer():
    return ("<div style='color:#666;font-size:0.8em;margin-top:8px;border-top:1px solid #ddd;padding-top:6px;'>"
            "Kalkulator ini adalah alat bantu keputusan klinis, bukan pengganti penilaian DPJP dan apoteker klinis. "
            "Verifikasi dengan pedoman terbaru, data laboratorium, dan kadar obat dalam darah bila tersedia.</div>")


def _html_rekomendasi(antibiotik, hasil):
    """HTML hasil rekomendasi (belum diberi penjelasan singkatan)."""
    rs, anmf, dc = hasil["rs"], hasil["anmf"], hasil["drugscom"]
    tabel = _tabel_rekomendasi(hasil)
    catatan_rs_html = "".join("<li>" + c + "</li>" for c in rs["catatan"])
    catatan_anmf_html = "".join("<li>" + c + "</li>" for c in anmf["catatan"])

    if antibiotik in ("Gentamisin", "Amikasin"):
        asal_rs = ("Rekomendasi Pedoman RS untuk Gentamisin/Amikasin TIDAK bergantung pada usia gestasi/pascalahir "
                   "(SPO tidak memuat penyesuaian tersebut); dosis &amp; frekuensi ditampilkan sesuai pernyataan SPO/PNPK.")
    elif hasil["rs_rentang_dari_kriteria"]:
        asal_rs = "Rentang dosis Pedoman RS mengikuti rentang yang dinyatakan pedoman untuk profil pasien ini."
    else:
        asal_rs = "Dosis Pedoman RS ditampilkan sebagai nilai tunggal sesuai pedoman untuk profil pasien ini."
    asal_lit = ("Rentang dosis Literatur Terkini = gabungan titik ANMF 2024 dan rentang Drugs.com/ASHP (AAP &amp; IDSA), diambil batas terluas dari keduanya."
                if GABUNG_DRUGSCOM_KE_LITERATUR else
                "Dosis &amp; frekuensi Literatur Terkini = ANMF 2024 (nilai titik dari nomogram PMA &amp; usia pascalahir).")

    peringatan = "".join(_kotak(*p) for p in _peringatan_rekomendasi(hasil))
    _kode = hasil.get("kode_pasien", "")
    _kode_html = ("<div style='font-weight:600;color:#0d6efd;margin-bottom:2px;'>Kode pasien: " + _kode + "</div>") if _kode else ""
    return (_kode_html
            + "<h4>Rekomendasi Dosis " + antibiotik + " -- usia gestasi lahir " + _fmt_usia_minggu(hasil["ga_lahir"]) + ", usia pascalahir " + str(hasil["pna_hari"]) + " hari, PMA " + _fmt_usia_minggu(hasil["ga_lahir"] + hasil["pna_hari"] / 7.0) + ", berat " + _fmt_num(hasil["berat_kg"]) + " kg</h4>"
            + peringatan
            + tabel
            + "<div style='color:#777;font-size:0.85em;margin:2px 0 10px 0;'>" + asal_rs + "<br>" + asal_lit + "</div>"
            + "<b>Catatan Pedoman RS:</b><ul style='margin:2px 0 10px 18px;'>" + catatan_rs_html + "</ul>"
            + "<b>Catatan Literatur (ANMF):</b><ul style='margin:2px 0 10px 18px;'>" + catatan_anmf_html + "</ul>"
            + "<h5 style='margin-top:14px;margin-bottom:2px;'>Rincian Kriteria Kesesuaian &amp; Catatan Rujukan</h5>"
            + _blok_kriteria(hasil)
            + (_blok_rentang("Drugs.com / ASHP (AAP &amp; IDSA)", dc, "#0dcaf0") if TAMPILKAN_DRUGSCOM_TAMBAHAN else "")
            + _blok_pustaka_tambahan(hasil)
            + _legenda_warna()
            + _baris_disclaimer()
            + "<div style='color:#999;font-size:0.8em;margin-top:8px;'>Versi aturan " + hasil.get("versi_aturan", "") + "</div>")


def tampilkan_rekomendasi(antibiotik, hasil):
    display(HTML(_perjelas_singkatan(_html_rekomendasi(antibiotik, hasil))))

print("Fungsi tampilan siap.")
```

## 2. Rekomendasi Dosis

Usia gestasi diisi sebagai **minggu lengkap + tambahan hari** (mis. 32 minggu 4 hari: isi 32 dan 4). Usia pascalahir dihitung dari hari lahir = 0. Berat badan dalam **kilogram**.

Isi data pasien di bawah, lalu klik **Hitung Rekomendasi Dosis**. Hasil ditampilkan sebagai **tabel dosis total (mg) & frekuensi (tiap X jam)**. Peringatan penting tampil di bagian atas hasil dengan warna: **merah** (perlu verifikasi/tindakan segera), **kuning** (hasil memakai asumsi atau perlu dipertimbangkan), **biru** (informasi), **hijau** (sesuai). Semua singkatan dijelaskan pada kemunculan pertama dan dirangkum pada kotak *Keterangan singkatan*.

Hasil membandingkan dua rujukan, sama dengan naskah *Pediatric Drugs*: **Pedoman Rumah Sakit** (SPO RSUP Dr. M. Djamil untuk gentamisin & amikasin; kriteria UNC untuk vankomisin), yang **tidak** membedakan usia gestasi, dan **Literatur Terkini** (ANMF 2024), yang menyesuaikan interval menurut usia pascamenstruasi dan usia pascalahir.

**Pasien berikutnya:** klik **Pasien Baru** (formulir dikosongkan), atau langsung ubah data lalu klik **Hitung** lagi. Semua kolom isian **kosong sejak awal** dan antibiotik belum terpilih, sehingga kalkulator tidak pernah menghitung dengan angka yang tidak diisi pengguna; bila ada kolom wajib yang kosong, muncul kotak merah yang menyebutkan kolom tersebut. Angka desimal boleh memakai koma atau titik. Ringkasan setiap pasien yang sudah dihitung tetap tersimpan pada tabel *Riwayat* selama halaman tidak dimuat ulang. Bagian ini **tidak bergantung** pada Bagian 3, dan sebaliknya.


```python
# ---- Formulir data pasien: dibuat terpisah untuk Bagian 2 (Rekomendasi) dan Bagian 3 (Evaluasi) ----
_LAY = widgets.Layout(width="580px")   # perlebar agar isian angka tidak terpotong
_STY = {"description_width": "300px"}
NILAI_AWAL_FORM = {"antibiotik": None, "ga": "", "ga_hari": "", "pna": "", "bb": "", "ada_ginjal": False,
                   "scr": "", "uop": "", "hipotermia": False, "cox": False, "sepsis_berat": False}
PILIHAN_ANTIBIOTIK = [("-- pilih antibiotik --", None), ("Gentamisin", "Gentamisin"),
                      ("Amikasin", "Amikasin"), ("Vankomisin", "Vankomisin")]


def _isian_angka(deskripsi, contoh, disabled=False, sty=None, lay=None):
    """Kolom angka yang KOSONG sejak awal (contoh pengisian tampil samar sebagai petunjuk).
    Koma maupun titik desimal diterima."""
    return widgets.Text(value="", placeholder=contoh, description=deskripsi, disabled=disabled,
                        style=sty or _STY, layout=lay or _LAY, continuous_update=False)


def buat_form_pasien(kode_awal="Pasien 1"):
    f = {
        "kode": widgets.Text(value=kode_awal, description="Kode pasien:", style=_STY, layout=_LAY),
        "antibiotik": widgets.Dropdown(options=PILIHAN_ANTIBIOTIK, value=None,
                                       description="Antibiotik:", style=_STY, layout=_LAY),
        "ga": _isian_angka("Usia gestasi saat lahir -- minggu:", "wajib, minggu lengkap, mis. 32"),
        "ga_hari": _isian_angka("Usia gestasi saat lahir -- tambahan hari:", "opsional, 0-6, mis. 4 (kosong = 0)"),
        "pna": _isian_angka("Usia pascalahir (hari; hari lahir = 0):", "wajib, 0-28, mis. 3"),
        "bb": _isian_angka("Berat badan saat ini (kg):", "wajib, dalam kilogram, mis. 1,72"),
        "ada_ginjal": widgets.Checkbox(value=False, layout=widgets.Layout(width="760px"),
                                       description="Data fungsi ginjal tersedia (bila tidak dicentang, fungsi ginjal diasumsikan normal sesuai usia)"),
        "scr": _isian_angka("Kreatinin serum (mg/dL):", "wajib bila dicentang, mis. 0,6 (µmol/L ÷ 88,4)", disabled=True),
        "uop": _isian_angka("Produksi urin (mL/kg/jam):", "opsional, mis. 2,0", disabled=True),
        "hipotermia": widgets.Checkbox(value=False, layout=widgets.Layout(width="760px"),
                                       description="Hipotermia terapeutik (pada ensefalopati hipoksik-iskemik)"),
        "cox": widgets.Checkbox(value=False, layout=widgets.Layout(width="760px"),
                                description="Sedang mendapat indometasin / ibuprofen (penghambat siklooksigenase)"),
        "sepsis_berat": widgets.Checkbox(value=False, layout=widgets.Layout(width="760px"),
                                         description="Sepsis berat / dugaan Staphylococcus aureus resisten metisilin / meningitis / endokarditis (khusus Vankomisin)"),
    }

    def _toggle(change):
        f["scr"].disabled = not f["ada_ginjal"].value
        f["uop"].disabled = not f["ada_ginjal"].value

    f["ada_ginjal"].observe(_toggle, names="value")
    return f


def _baca_angka(teks):
    """'1,72' atau '1.72' -> 1.72; kosong -> None; bukan angka -> ValueError."""
    t = ("" if teks is None else str(teks)).strip().replace(",", ".")
    if t == "":
        return None
    return float(t)


class _PengumpulIsian:
    """Membaca beberapa kolom sekaligus dan mengumpulkan SEMUA kekurangan dalam satu pesan."""
    def __init__(self):
        self.kosong, self.salah = [], []

    def angka(self, widget, label, wajib=True, bulat=False):
        try:
            v = _baca_angka(widget.value)
        except ValueError:
            self.salah.append(label + " (\"" + str(widget.value) + "\" bukan angka)")
            return None
        if v is None:
            if wajib:
                self.kosong.append(label)
            return None
        if bulat:
            if abs(v - round(v)) > 1e-9:
                self.salah.append(label + " harus bilangan bulat")
                return None
            return int(round(v))
        return v

    def periksa(self):
        pesan = []
        if self.kosong:
            pesan.append("Isian berikut wajib diisi: " + ", ".join(self.kosong) + ".")
        if self.salah:
            pesan.append("Isian berikut belum benar: " + "; ".join(self.salah) + ".")
        if pesan:
            raise ValueError(" ".join(pesan))


# Rentang isian yang dapat dihitung. Isian di luar rentang ditolak dengan pesan yang jelas.
BATAS_ISIAN = {
    "ga": (22.0, 43.0, "Usia gestasi saat lahir", "minggu"),
    "bb": (0.3, 10.0, "Berat badan", "kg"),
    "scr": (0.0, 10.0, "Kreatinin serum", "mg/dL"),
    "uop": (0.0, 10.0, "Produksi urin", "mL/kg/jam"),
}


def _cek_isian(nama, nilai):
    lo, hi, label, satuan = BATAS_ISIAN[nama]
    if nilai is None or not (lo <= nilai <= hi):
        tambahan = ""
        if nama == "bb" and nilai is not None and nilai >= 300:
            tambahan = " Tampaknya berat diisi dalam gram; isi dalam kilogram (mis. 1720 g = 1,72 kg)."
        if nama == "scr" and nilai is not None and nilai > 10:
            tambahan = " Tampaknya kreatinin diisi dalam µmol/L; bagi dengan 88,4 untuk mendapatkan mg/dL."
        raise ValueError(label + " " + _fmt_num(nilai) + " " + satuan + " di luar rentang yang dapat dihitung ("
                         + _fmt_num(lo) + "-" + _fmt_num(hi) + " " + satuan + ")." + (tambahan or " Periksa kembali isian."))


def hitung_dari_form(f):
    p = _PengumpulIsian()
    if f["antibiotik"].value is None:
        p.kosong.append("Antibiotik")
    ga = p.angka(f["ga"], "Usia gestasi saat lahir (minggu)", bulat=True)
    ga_hari = p.angka(f["ga_hari"], "Usia gestasi saat lahir (tambahan hari)", wajib=False, bulat=True)
    pna = p.angka(f["pna"], "Usia pascalahir", bulat=True)
    bb = p.angka(f["bb"], "Berat badan")
    ada = f["ada_ginjal"].value
    scr = p.angka(f["scr"], "Kreatinin serum (karena 'Data fungsi ginjal tersedia' dicentang)") if ada else None
    uop = p.angka(f["uop"], "Produksi urin", wajib=False) if ada else None
    p.periksa()
    if ga_hari is not None and not (0 <= ga_hari <= 6):
        raise ValueError("Tambahan hari pada usia gestasi harus 0-6 (7 hari = 1 minggu).")
    ga = ga + (ga_hari or 0) / 7.0
    _cek_isian("ga", ga)
    _cek_isian("bb", bb)
    if scr is not None:
        _cek_isian("scr", scr)
    if uop is not None:
        _cek_isian("uop", uop)
    return bangun_rekomendasi(
        f["antibiotik"].value, ga, pna, bb, scr=scr, uop=uop,
        hipotermia=f["hipotermia"].value, cox_inhibitor=f["cox"].value,
        sepsis_berat=f["sepsis_berat"].value, kode_pasien=f["kode"].value)


def reset_form(f, kode_baru):
    for k, v in NILAI_AWAL_FORM.items():
        f[k].value = v
    f["kode"].value = kode_baru


def salin_form(dari, ke):
    for k in ["kode"] + list(NILAI_AWAL_FORM):
        ke[k].value = dari[k].value


def tampilkan_form(f):
    display(widgets.HTML("<b>Data Pasien</b> <span style='color:#666;font-size:0.9em;'>-- demi kerahasiaan, isi kode pasien "
                         "tanpa nama atau nomor rekam medis.</span>"))
    display(f["kode"], f["antibiotik"], f["ga"], f["ga_hari"], f["pna"], f["bb"])
    display(widgets.HTML("<br><b>Data tambahan (opsional, isi bila relevan)</b>"))
    display(f["ada_ginjal"], f["scr"], f["uop"], f["hipotermia"], f["cox"], f["sepsis_berat"])


def _tabel_riwayat(baris):
    if not baris:
        return "<div style='color:#777;font-size:0.9em;'>Belum ada perhitungan pada sesi ini.</div>"
    kolom = list(baris[0].keys())
    head = "".join("<th style='" + TH_STYLE + "'>" + k + "</th>" for k in kolom)
    isi = "".join("<tr>" + "".join("<td style='" + TD_STYLE + "'>" + str(r.get(k, "")) + "</td>" for k in kolom) + "</tr>" for r in baris)
    return "<div style='overflow-x:auto;'><table style='border-collapse:collapse;width:100%;margin:6px 0;'><tr>" + head + "</tr>" + isi + "</table></div>"


def _perbarui_riwayat(out, baris, judul):
    with out:
        clear_output()
        display(HTML("<h5 style='margin:12px 0 2px 0;'>" + judul + " (" + str(len(baris)) + " pasien)</h5>"
                     "<div style='color:#777;font-size:0.85em;'>Ringkasan tetap tersimpan selama halaman tidak dimuat ulang.</div>"
                     + _tabel_riwayat(baris)))


def _teks_mg(a, b):
    if a is None:
        return "-"
    return _fmt_num(a) + " mg" if a == b else _fmt_num(a) + "-" + _fmt_num(b) + " mg"


# ---- Bagian 2: Rekomendasi ----
form_rek = buat_form_pasien("Pasien 1")
# nama lama dipertahankan agar kompatibel
w_kode_pasien, w_antibiotik, w_ga, w_pna, w_bb = (form_rek["kode"], form_rek["antibiotik"], form_rek["ga"],
                                                  form_rek["pna"], form_rek["bb"])
w_ada_ginjal, w_scr, w_uop = form_rek["ada_ginjal"], form_rek["scr"], form_rek["uop"]
w_hipotermia, w_cox, w_sepsis_berat = form_rek["hipotermia"], form_rek["cox"], form_rek["sepsis_berat"]

RIWAYAT_REKOMENDASI = []
btn_hitung = widgets.Button(description="Hitung Rekomendasi Dosis", button_style="primary", icon="calculator",
                            layout=widgets.Layout(width="230px"))
btn_pasien_baru_rek = widgets.Button(description="Pasien Baru", icon="user-plus", layout=widgets.Layout(width="160px"),
                                     tooltip="Kembalikan formulir ke nilai awal untuk menghitung pasien lain")
out_hitung = widgets.Output()
out_riwayat_rek = widgets.Output()


def _ringkas_rekomendasi(hasil):
    return {"Kode pasien": hasil["kode_pasien"], "Antibiotik": hasil["antibiotik"],
            "Usia gestasi saat lahir": _fmt_usia_minggu(hasil["ga_lahir"]),
            "Usia pascalahir (hari)": hasil["pna_hari"], "Berat badan (kg)": _fmt_num(hasil["berat_kg"]),
            "Pedoman Rumah Sakit: dosis": _teks_mg(*hasil["rs_rentang_mg"]),
            "Pedoman Rumah Sakit: frekuensi": _fmt_q_rentang(hasil["rs"]["interval_min"], hasil["rs"]["interval_max"]),
            "Literatur Terkini: dosis": _teks_mg(*hasil["lit_rentang_mg"]),
            "Literatur Terkini: frekuensi": _fmt_q_rentang(*hasil["lit_interval_rentang"])}


def _on_hitung_clicked(b):
    with out_hitung:
        clear_output()
        try:
            hasil = hitung_dari_form(form_rek)
        except ValueError as e:
            display(HTML(_kotak("bahaya", "Data tidak dapat dihitung", str(e))))
            return
        tampilkan_rekomendasi(hasil["antibiotik"], hasil)
    RIWAYAT_REKOMENDASI.append(_ringkas_rekomendasi(hasil))
    _perbarui_riwayat(out_riwayat_rek, RIWAYAT_REKOMENDASI, "Riwayat rekomendasi pada sesi ini")


def _on_pasien_baru_rek(b):
    reset_form(form_rek, "Pasien " + str(len(RIWAYAT_REKOMENDASI) + 1))
    with out_hitung:
        clear_output()
        display(HTML(_kotak("info", "Formulir siap untuk pasien baru",
                            "Isi data pasien berikutnya, lalu klik <b>Hitung Rekomendasi Dosis</b>. "
                            "Ringkasan pasien sebelumnya tetap tersimpan pada Riwayat di bawah.")))


btn_hitung.on_click(_on_hitung_clicked)
btn_pasien_baru_rek.on_click(_on_pasien_baru_rek)

tampilkan_form(form_rek)
display(widgets.HTML("<br>"))
display(widgets.HBox([btn_hitung, btn_pasien_baru_rek]))
display(widgets.HTML("<div style='color:#555;font-size:0.9em;margin:4px 0;'>Semua kolom kosong sejak awal; kolom bertanda <i>wajib diisi</i> harus diisi sebelum menghitung. "
                     "Angka desimal boleh memakai koma atau titik (1,72 atau 1.72).</div>"))
display(widgets.HTML("<div style='color:#555;font-size:0.9em;margin:4px 0;'>Untuk pasien lain: klik <b>Pasien Baru</b> (formulir dikosongkan), "
                     "atau langsung ubah data lalu klik <b>Hitung Rekomendasi Dosis</b> lagi. Tidak perlu kembali atau memuat ulang halaman.</div>"))
display(out_hitung, out_riwayat_rek)
_perbarui_riwayat(out_riwayat_rek, RIWAYAT_REKOMENDASI, "Riwayat rekomendasi pada sesi ini")
```

## 3. Evaluasi Pemberian Antibiotik yang Sudah Dilakukan

Bagian ini **berdiri sendiri**: punya formulir data pasien sendiri, sehingga dapat dipakai tanpa mengisi bagian Rekomendasi Dosis. Tombol **Salin data pasien dari Rekomendasi Dosis** tersedia bila pasien yang sama sudah diisi di sana (opsional).

Yang dibandingkan:

1. **Pedoman Rumah Sakit** -- SPO RSUP Dr. M. Djamil (gentamisin, amikasin) dan kriteria UNC (vankomisin); tidak membedakan usia gestasi.
2. **Literatur Terkini** -- ANMF 2024 (dosis & interval dari nomogram usia pascamenstruasi/usia pascalahir).
3. **Hitung Manual (opsional)** -- centang *Saya punya hasil hitung manual* bila ada. Isi dosis per pemberian sebagai **batas bawah dan batas atas** (mg) dan frekuensi sebagai **paling sering** dan **paling jarang** (setiap ... jam); bila hasil hitungan satu nilai, isi batas bawah = batas atas. Bila tidak dicentang, kolom Hitung Manual tidak ditampilkan.
4. **Aktual Diberikan (wajib)** -- dosis per pemberian (mg) dan frekuensi yang benar-benar diberikan, diisi sebagai **diberikan setiap ... jam** (mis. 36). Kalkulator juga menampilkan dosis per kg (dosis aktual ÷ berat badan).

Hasil diawali **kotak ringkasan berwarna**: merah bila ada pemberian **berlebih**, kuning bila ada yang **kurang**, hijau bila semuanya **sesuai**. Kategori:
- **Sesuai** -- di dalam rentang rujukan; **Berlebih** -- di atas batas maksimal; **Kurang** -- di bawah batas minimal.
- Untuk Berlebih/Kurang ditulis selisihnya **dalam satuan dan %**: dosis dalam mg (mis. "+4,14 mg; +25,0%"), frekuensi dalam dosis/hari beserta selisih interval dalam jam (mis. "+0,333 dosis/hari; +50,0%; interval 12 jam lebih pendek dari interval 36 jam").
- Frekuensi dinilai pada skala dosis/hari (24 ÷ interval): **Berlebih** = lebih sering dari rujukan (interval lebih pendek), **Kurang** = lebih jarang (interval lebih panjang).
- **Tidak ada toleransi persentase**: perbedaan sekecil apa pun langsung dinyatakan Berlebih/Kurang.

Untuk pasien berikutnya klik **Pasien Baru** (formulir dikosongkan), atau langsung ubah data lalu klik **Evaluasi Pemberian** lagi; ringkasan setiap pasien tersimpan pada tabel *Riwayat*.


```python
# ---- Bagian 3: Evaluasi pemberian (berdiri sendiri; punya formulir data pasien sendiri) ----
_LAY3 = widgets.Layout(width="620px")   # perlebar agar isian angka tidak terpotong
_STY3 = {"description_width": "350px"}
form_eval = buat_form_pasien("Pasien 1")
w_eval_antibiotik = form_eval["antibiotik"]   # nama lama dipertahankan

NILAI_AWAL_EVAL = {"dm_min": "", "dm_max": "", "im_min": "", "im_max": "", "da": "", "ia": ""}


def _isian_eval(deskripsi, contoh, disabled=False):
    return _isian_angka(deskripsi, contoh, disabled=disabled, sty=_STY3, lay=_LAY3)


w_dosis_manual_min = _isian_eval("Dosis manual per pemberian -- batas bawah (mg):", "mis. 8,6", disabled=True)
w_dosis_manual_max = _isian_eval("Dosis manual per pemberian -- batas atas (mg):", "mis. 12,9", disabled=True)
w_interval_manual_min = _isian_eval("Frekuensi manual -- paling sering: setiap ... jam:", "mis. 24", disabled=True)
w_interval_manual_max = _isian_eval("Frekuensi manual -- paling jarang: setiap ... jam:", "mis. 48", disabled=True)
w_dosis_aktual = _isian_eval("Dosis aktual per pemberian (mg):", "wajib, mis. 9")
w_interval_aktual = _isian_eval("Frekuensi aktual -- diberikan setiap ... jam:", "wajib, mis. 24 (sekali sehari) atau 36")

# Hitung manual bersifat OPSIONAL: isian aktif hanya bila kotak ini dicentang.
w_ada_manual = widgets.Checkbox(value=False, layout=widgets.Layout(width="760px"),
                                description="Saya punya hasil hitung manual sebagai pembanding (opsional)")


def _toggle_manual(change):
    for _w in (w_dosis_manual_min, w_dosis_manual_max, w_interval_manual_min, w_interval_manual_max):
        _w.disabled = not w_ada_manual.value


w_ada_manual.observe(_toggle_manual, names="value")


def _baca_isian_evaluasi():
    """(dosis_manual_min, dosis_manual_max, interval_manual_min, interval_manual_max, dosis_aktual, interval_aktual).
    Hitung manual = None bila tidak dicentang. Isian kosong/salah -> ValueError berisi semua kekurangan."""
    p = _PengumpulIsian()
    da = p.angka(w_dosis_aktual, "Dosis aktual")
    ia = p.angka(w_interval_aktual, "Frekuensi aktual (setiap ... jam)", bulat=True)
    dm_min = dm_max = im_min = im_max = None
    if w_ada_manual.value:
        dm_min = p.angka(w_dosis_manual_min, "Dosis manual batas bawah")
        dm_max = p.angka(w_dosis_manual_max, "Dosis manual batas atas")
        im_min = p.angka(w_interval_manual_min, "Frekuensi manual paling sering (setiap ... jam)", bulat=True)
        im_max = p.angka(w_interval_manual_max, "Frekuensi manual paling jarang (setiap ... jam)", bulat=True)
    p.periksa()
    if da <= 0:
        raise ValueError("Dosis aktual harus lebih dari 0 mg.")
    if not (1 <= ia <= 96):
        raise ValueError("Frekuensi aktual: isi jarak antardosis 1-96 jam (mis. 24 untuk sekali sehari).")
    if w_ada_manual.value:
        if dm_min <= 0 or dm_max <= 0:
            raise ValueError("Dosis manual harus lebih dari 0 mg (atau hilangkan centang hitung manual).")
        if dm_min > dm_max:
            raise ValueError("Dosis manual batas bawah tidak boleh lebih besar dari batas atas.")
        if not (1 <= im_min <= 96 and 1 <= im_max <= 96):
            raise ValueError("Frekuensi manual harus 1-96 jam.")
        if im_min > im_max:
            raise ValueError("Frekuensi manual 'paling sering' (jam) tidak boleh lebih besar dari 'paling jarang' (jam).")
    return dm_min, dm_max, im_min, im_max, da, ia


def _baca_semua_evaluasi():
    """Membaca data pasien Bagian 3 dan isian evaluasi; semua kekurangan dilaporkan sekaligus."""
    pesan, hasil, isian = [], None, None
    try:
        hasil = hitung_dari_form(form_eval)
    except ValueError as e:
        pesan.append(str(e))
    try:
        isian = _baca_isian_evaluasi()
    except ValueError as e:
        pesan.append(str(e))
    if pesan:
        raise ValueError("<br>".join(pesan))
    return hasil, isian


RIWAYAT_EVALUASI = []
btn_evaluasi = widgets.Button(description="Evaluasi Pemberian", button_style="warning", icon="stethoscope",
                              layout=widgets.Layout(width="230px"))
btn_pasien_baru_eval = widgets.Button(description="Pasien Baru", icon="user-plus", layout=widgets.Layout(width="160px"),
                                      tooltip="Kembalikan formulir ke nilai awal untuk mengevaluasi pasien lain")
btn_salin_dari_rek = widgets.Button(description="Salin data pasien dari Rekomendasi Dosis", icon="copy",
                                    layout=widgets.Layout(width="300px"))
out_evaluasi = widgets.Output()
out_riwayat_eval = widgets.Output()

def _warna(status):
    return {"ok": "#198754", "warn": "#dc3545", "info": "#6c757d"}[status]

# --- Perbandingan interval (titik vs titik): dipakai untuk baris Frekuensi ---
def _cmp_interval(nilai, ref):
    if ref is None or nilai is None:
        return "-"
    if nilai == ref:
        return "<span style=\'color:#198754;font-weight:600;\'>Sesuai</span>"
    elif nilai < ref:
        return "<span style=\'color:#dc3545;font-weight:600;\'>Lebih pendek</span>"
    else:
        return "<span style=\'color:#fd7e14;font-weight:600;\'>Lebih panjang</span>"


def _cmp_interval_rs(nilai, imin, imax):
    """Perbandingan interval terhadap rentang interval RS (mis. gentamisin 24-48).
    Menghasilkan span berwarna: hijau=Sesuai, merah=Lebih pendek, oranye=Lebih panjang."""
    if imin is None or imax is None or nilai is None:
        return "-"
    if imin <= nilai <= imax:
        return "<span style=\'color:#198754;font-weight:600;\'>Sesuai</span>"
    elif nilai < imin:
        return "<span style=\'color:#dc3545;font-weight:600;\'>Lebih pendek</span>"
    else:
        return "<span style=\'color:#fd7e14;font-weight:600;\'>Lebih panjang</span>"

def _status_interval_rentang_list(diberikan, opsi_list):
    if diberikan in opsi_list:
        return ("ok", "Sesuai (" + _fmt_q(diberikan) + " termasuk pilihan tepat).")
    return ("warn", "TIDAK termasuk pilihan frekuensi yang dianggap tepat (" + str(opsi_list) + " jam).")

def _status_interval_rentang_minmax(diberikan, imin, imax):
    if imin <= diberikan <= imax:
        return ("ok", "Sesuai (dalam rentang " + _fmt_rentang(imin, imax) + " jam).")
    _, teks = _kategori_dosis(24.0 / diberikan, 24.0 / diberikan, 24.0 / imax, 24.0 / imin, "dosis/hari")
    if diberikan < imin:
        return ("warn", "LEBIH PENDEK dari rentang tepat (" + _fmt_rentang(imin, imax) + " jam): "
                + _fmt_num(imin - diberikan) + " jam lebih pendek; frekuensi berlebih " + teks + ".")
    return ("warn", "LEBIH PANJANG dari rentang tepat (" + _fmt_rentang(imin, imax) + " jam): "
            + _fmt_num(diberikan - imax) + " jam lebih panjang; frekuensi kurang " + teks + ".")


# --- Kategorisasi dosis TERPADU (rentang vs rentang) ---
# Sisi kiri (nilai_min, nilai_max) boleh berupa TITIK TUNGGAL (nilai_min==nilai_max,
# dipakai untuk Dosis Aktual) ATAU rentang genuine (dipakai untuk Hitung Manual),
# dibandingkan terhadap rentang rekomendasi (rekom_min, rekom_max):
#   - Sesuai   : nilai/rentang berada di DALAM rentang rekomendasi.
#   - Berlebih : nilai_max melebihi rekom_max -- %kelebihan dihitung dari rekom_max.
#   - Kurang   : nilai_min di bawah rekom_min -- %kekurangan dihitung dari rekom_min.
# Berlaku untuk dosis TETAP (rekom_min==rekom_max) MAUPUN rentang: tidak ada
# toleransi -- perbedaan sekecil apa pun langsung dikategorikan Berlebih/Kurang
# (hanya epsilon float 1e-6 untuk menghindari galat pembulatan biner).
_EPS_DOSIS = 1e-6

def _fmt_pct(pct):
    """Persentase 1 desimal; 2 desimal bila selisih sangat kecil (<0,1%)."""
    return ("{0:.2f}%" if abs(pct) < 0.1 else "{0:.1f}%").format(abs(pct)).replace(".", ",")


def _teks_selisih(tanda, selisih, pct, satuan=None):
    """Teks selisih: '+4.14 mg; +25.0%' (satuan diisi) atau '+25.0%' (tanpa satuan)."""
    teks_pct = tanda + _fmt_pct(pct)
    if not satuan:
        return teks_pct
    return tanda + _fmt_num(round(abs(selisih), 3)) + " " + satuan + "; " + teks_pct


def _kategori_dosis(nilai_min, nilai_max, rekom_min, rekom_max, satuan=None):
    """Mengembalikan (kategori, teks_selisih). Revisi 2026.11.01: teks selisih memuat
    angka dalam SATUAN (bila `satuan` diisi, mis. 'mg', 'mg/kg', 'dosis/hari') DAN %."""
    if rekom_min is None or rekom_max is None or nilai_min is None or nilai_max is None:
        return None, None
    if nilai_max > rekom_max + _EPS_DOSIS:
        d = nilai_max - rekom_max
        pct = d / rekom_max * 100 if rekom_max else 0
        return "Berlebih", _teks_selisih("+", d, pct, satuan)
    elif nilai_min < rekom_min - _EPS_DOSIS:
        d = rekom_min - nilai_min
        pct = d / rekom_min * 100 if rekom_min else 0
        return "Kurang", _teks_selisih("-", d, pct, satuan)
    else:
        return "Sesuai", None


def _teks_interval(kategori, n_imin, n_imax, r_imin, r_imax):
    """Selisih interval dalam JAM untuk kategori frekuensi.
    Berlebih (terlalu sering) -> interval tersering lebih pendek dari batas bawah rekomendasi;
    Kurang (terlalu jarang)   -> interval terjarang lebih panjang dari batas atas rekomendasi."""
    if kategori == "Berlebih" and None not in (n_imin, r_imin):
        return "interval " + _fmt_num(r_imin - n_imin) + " jam lebih pendek dari interval " + str(r_imin) + " jam"
    if kategori == "Kurang" and None not in (n_imax, r_imax):
        return "interval " + _fmt_num(n_imax - r_imax) + " jam lebih panjang dari interval " + str(r_imax) + " jam"
    return ""

_WARNA_KATEGORI = {"Sesuai": "#198754", "Berlebih": "#dc3545", "Kurang": "#fd7e14"}

def _span_kategori(kategori, teks, tambahan=""):
    html = "<span style=\'color:" + _WARNA_KATEGORI[kategori] + ";font-weight:600;\'>" + kategori + "</span>"
    if teks:
        html += "<br><span style=\'color:#555;font-size:0.8em;\'>(" + teks + ")</span>"
    if tambahan:
        html += "<br><span style=\'color:#555;font-size:0.8em;\'>" + tambahan + "</span>"
    return html


def _render_kategori_dosis(nilai_min, nilai_max, rekom_min, rekom_max, satuan="mg"):
    kategori, teks = _kategori_dosis(nilai_min, nilai_max, rekom_min, rekom_max, satuan)
    if kategori is None:
        return "-"
    return _span_kategori(kategori, teks)


def _cmp_batas(nilai, ref, label, satuan="mg", interval_nilai=None, interval_ref=None):
    """Bandingkan SATU batas manual (nilai) terhadap SATU batas rekomendasi (ref):
    arah + selisih dalam SATUAN + %, tanpa toleransi. Untuk frekuensi (satuan
    dosis/hari) disertakan juga selisih interval dalam jam."""
    if ref is None or nilai is None or ref == 0:
        return label + ": -"
    d = nilai - ref
    if abs(d) <= _EPS_DOSIS:
        return label + ": <span style=\'color:#198754;font-weight:600;\'>sama</span>"
    tanda = "+" if d > 0 else "-"
    teks = _teks_selisih(tanda, d, d / ref * 100, satuan)
    if interval_nilai is not None and interval_ref is not None and interval_nilai != interval_ref:
        teks += "; interval " + _fmt_num(abs(interval_nilai - interval_ref)) + " jam " + ("lebih pendek" if interval_nilai < interval_ref else "lebih panjang")
    if d > 0:
        return label + ": <span style=\'color:#dc3545;font-weight:600;\'>lebih besar (" + teks + ")</span>"
    return label + ": <span style=\'color:#fd7e14;font-weight:600;\'>lebih kecil (" + teks + ")</span>"


def _render_manual_vs_rekom(m_min, m_max, rekom_min, rekom_max, satuan="mg"):
    """Sel kolom 'Hitung Manual vs Pedoman RS/Literatur'.
    - Bila rekomendasi berupa RENTANG (rekom_min != rekom_max): tampilkan
      perbedaan manual-min vs rekom-min DAN manual-maks vs rekom-maks, masing-
      masing dengan arah + % (sesuai permintaan).
    - Bila rekomendasi berupa dosis TETAP (rekom_min == rekom_max): tampilkan
      kategori tunggal Sesuai/Berlebih/Kurang (tanpa toleransi)."""
    if rekom_min is None or rekom_max is None or m_min is None or m_max is None:
        return "-"
    if rekom_min == rekom_max:
        return _render_kategori_dosis(m_min, m_max, rekom_min, rekom_max, satuan)
    return (_cmp_batas(m_min, rekom_min, "Min", satuan) + "<br>" + _cmp_batas(m_max, rekom_max, "Maks", satuan))


# --- Kategorisasi FREKUENSI ---
# Frekuensi diperlakukan pada skala DOSIS/HARI (= 24 / interval jam) agar
# "lebih sering" = angka lebih besar, sehingga bisa memakai ULANG logika
# kategori dosis: Berlebih = lebih sering dari frekuensi MAKSIMAL rekomendasi
# (interval lebih pendek dari interval terpendek), Kurang = lebih jarang dari
# frekuensi MINIMAL (interval lebih panjang dari interval terpanjang). Persentase
# dinyatakan pada skala frekuensi (dosis/hari), sesuai permintaan.
def _freq_per_hari(interval):
    if interval is None or interval == 0:
        return None
    return 24.0 / float(interval)


def _freq_rentang_per_hari(imin, imax):
    """Rentang interval [imin, imax] jam (imin<=imax) -> rentang frekuensi/hari
    [24/imax, 24/imin] (dari paling jarang ke paling sering)."""
    if imin is None or imax is None:
        return None, None
    return 24.0 / float(imax), 24.0 / float(imin)


def _render_kategori_frekuensi(aktual_interval, rmin_interval, rmax_interval):
    """Kategori frekuensi AKTUAL (interval tunggal) vs rentang interval rekomendasi."""
    fa = _freq_per_hari(aktual_interval)
    fmin, fmax = _freq_rentang_per_hari(rmin_interval, rmax_interval)
    kategori, teks = _kategori_dosis(fa, fa, fmin, fmax, "dosis/hari")
    if kategori is None:
        return "-"
    return _span_kategori(kategori, teks, _teks_interval(kategori, aktual_interval, aktual_interval, rmin_interval, rmax_interval))


def _render_manual_vs_frekuensi(m_imin, m_imax, r_imin, r_imax):
    """Kategori frekuensi MANUAL (rentang interval) vs rentang interval rekomendasi.
    Dua-batas bila rekomendasi berupa rentang; kategori tunggal bila tetap."""
    if None in (m_imin, m_imax, r_imin, r_imax):
        return "-"
    fm_min, fm_max = _freq_rentang_per_hari(m_imin, m_imax)
    fr_min, fr_max = _freq_rentang_per_hari(r_imin, r_imax)
    if r_imin == r_imax:
        kategori, teks = _kategori_dosis(fm_min, fm_max, fr_min, fr_max, "dosis/hari")
        if kategori is None:
            return "-"
        return _span_kategori(kategori, teks, _teks_interval(kategori, m_imin, m_imax, r_imin, r_imax))
    # batas frekuensi MIN <-> interval MAKS, frekuensi MAKS <-> interval MIN
    return (_cmp_batas(fm_min, fr_min, "Min", "dosis/hari", m_imax, r_imax) + "<br>"
            + _cmp_batas(fm_max, fr_max, "Maks", "dosis/hari", m_imin, r_imin))


def _tabel_evaluasi(hasil, dm_min_mg, dm_max_mg, interval_manual_min, interval_manual_max, dosis_aktual_mg, interval_aktual):
    """TABEL UTAMA Bagian 3: Pedoman RS, Literatur Terkini, Hitung Manual (opsional), Aktual.
    Kolom Hitung Manual hanya ditampilkan bila hitung manual diisi (dm_min_mg tidak None)."""
    rs = hasil["rs"]
    rs_min_mg, rs_max_mg = hasil["rs_rentang_mg"]
    lit_min_mg, lit_max_mg = hasil["lit_rentang_mg"]
    rs_imin, rs_imax = rs.get("interval_min"), rs.get("interval_max")
    lit_imin, lit_imax = hasil.get("lit_interval_rentang", (None, None))
    ada_manual = dm_min_mg is not None

    def sel_rentang_mg(vmin, vmax):
        if vmin is None:
            return "-"
        return _fmt_num(vmax) + " mg" if vmin == vmax else _fmt_num(vmin) + "-" + _fmt_num(vmax) + " mg"

    berat = hasil.get("berat_kg")
    if dosis_aktual_mg is not None and berat:
        sel_aktual = ("<b>" + _fmt_num(dosis_aktual_mg) + " mg</b><br>"
                      "<span style='color:#777;font-size:0.85em;'>(" + _fmt_num(dosis_aktual_mg / berat) + " mg/kg/dosis)</span>")
    else:
        sel_aktual = _fmt_num(dosis_aktual_mg) + " mg"

    # (judul kolom, isi baris Dosis, isi baris Frekuensi, kolom manual?)
    kolom = [
        ("Pedoman Rumah Sakit", sel_rentang_mg(rs_min_mg, rs_max_mg), _fmt_q_rentang(rs_imin, rs_imax), False),
        ("Literatur Terkini", sel_rentang_mg(lit_min_mg, lit_max_mg), _fmt_q_rentang(lit_imin, lit_imax), False),
        ("Hitung Manual", sel_rentang_mg(dm_min_mg, dm_max_mg),
         _fmt_q_rentang(interval_manual_min, interval_manual_max) if ada_manual else "-", True),
        ("Aktual Diberikan", sel_aktual, _fmt_q(interval_aktual), False),
        ("Aktual vs Pedoman Rumah Sakit",
         _render_kategori_dosis(dosis_aktual_mg, dosis_aktual_mg, rs_min_mg, rs_max_mg),
         _render_kategori_frekuensi(interval_aktual, rs_imin, rs_imax), False),
        ("Aktual vs Literatur",
         _render_kategori_dosis(dosis_aktual_mg, dosis_aktual_mg, lit_min_mg, lit_max_mg),
         _render_kategori_frekuensi(interval_aktual, lit_imin, lit_imax), False),
        ("Manual vs Pedoman Rumah Sakit",
         _render_manual_vs_rekom(dm_min_mg, dm_max_mg, rs_min_mg, rs_max_mg),
         _render_manual_vs_frekuensi(interval_manual_min, interval_manual_max, rs_imin, rs_imax), True),
        ("Manual vs Literatur",
         _render_manual_vs_rekom(dm_min_mg, dm_max_mg, lit_min_mg, lit_max_mg),
         _render_manual_vs_frekuensi(interval_manual_min, interval_manual_max, lit_imin, lit_imax), True),
    ]
    kolom = [k for k in kolom if ada_manual or not k[3]]
    th = lambda t: "<th style='" + TH_STYLE + "'>" + t + "</th>"
    td = lambda t: "<td style='" + TD_STYLE + "'>" + t + "</td>"
    header = "<tr>" + th("Parameter") + "".join(th(k[0]) for k in kolom) + "</tr>"
    baris_dosis = "<tr>" + td("<b>Dosis per pemberian</b>") + "".join(td(k[1]) for k in kolom) + "</tr>"
    baris_freq = "<tr>" + td("<b>Frekuensi</b>") + "".join(td(k[2]) for k in kolom) + "</tr>"
    catatan_manual = ("" if ada_manual else
                      "Hitung manual tidak diisi, sehingga kolom Hitung Manual tidak ditampilkan. ")
    return ("<div style='overflow-x:auto;'><table style='border-collapse:collapse;width:100%;margin:10px 0;'>"
            + header + baris_dosis + baris_freq + "</table></div>"
            + "<div style='color:#777;font-size:0.85em;'>" + catatan_manual
            + "Pedoman Rumah Sakit tidak membedakan usia gestasi; Literatur Terkini (ANMF 2024) menyesuaikan interval menurut usia. "
              "Kategori DOSIS: <b>Sesuai</b> (di dalam rentang) &middot; <b>Berlebih</b> (di atas maksimal; selisih dalam mg dan % dari batas maksimal) &middot; <b>Kurang</b> (di bawah minimal; selisih dalam mg dan % dari batas minimal). "
              "Kategori FREKUENSI dihitung pada skala frekuensi (dosis/hari = 24/interval): <b>Berlebih</b> = lebih sering dari frekuensi maksimal rekomendasi (interval lebih pendek), <b>Kurang</b> = lebih jarang dari frekuensi minimal (interval lebih panjang); selisih ditulis dalam dosis/hari dan %, beserta selisih interval dalam jam. Tanpa toleransi. "
              "Geser tabel ke kanan bila terpotong di layar kecil.</div>")


def _blok_evaluasi_kriteria(hasil, dosis_min_per_kg, dosis_max_per_kg, interval_jam, label):
    kr = hasil["rs_kriteria"]
    kategori, pct = _kategori_dosis(dosis_min_per_kg, dosis_max_per_kg, kr["dosis_min"], kr["dosis_max"], "mg/kg")
    warna_dosis = _WARNA_KATEGORI.get(kategori, "#6c757d")
    msg_dosis = (kategori or "-") + (" (" + pct + ")" if pct else "")
    s_int, msg_int = _status_interval_rentang_minmax(interval_jam, kr["interval_min"], kr["interval_max"])
    return ("<div style=\'border-left:4px solid #0d6efd;background:#eef6ff;padding:10px 14px;margin:8px 0;border-radius:4px;\'>"
            + "<b>Kriteria Kesesuaian Pedoman RS</b> <span style=\'color:#777;font-size:0.85em;\'>(dibandingkan dengan " + label + "; sumber: " + kr["sumber"] + ")</span>"
            + "<ul style=\'margin:6px 0 0 0;\'>"
            + "<li><b>Dosis:</b> <span style=\'color:" + warna_dosis + ";font-weight:600;\'>" + msg_dosis + "</span> (rentang tepat: " + _fmt_rentang(kr["dosis_min"], kr["dosis_max"]) + " mg/kg)</li>"
            + "<li><b>Interval:</b> <span style=\'color:" + _warna(s_int) + ";font-weight:600;\'>" + msg_int + "</span></li>"
            + "</ul>" + _catatan_cakupan_kriteria(hasil) + "</div>")


def _blok_evaluasi_rentang(judul, data, dosis_min_per_kg, dosis_max_per_kg, interval_jam, label):
    dmin, dmax = data["dosis_rentang"]
    imin, imax = data["interval_rentang"]
    kategori, pct = _kategori_dosis(dosis_min_per_kg, dosis_max_per_kg, dmin, dmax, "mg/kg")
    warna_dosis = _WARNA_KATEGORI.get(kategori, "#6c757d")
    msg_dosis = (kategori or "-") + (" (" + pct + ")" if pct else "")
    s_int, msg_int = _status_interval_rentang_minmax(interval_jam, imin, imax)
    return ("<div style=\'border-left:4px solid #0dcaf0;background:#f9f9f9;padding:10px 14px;margin:8px 0;border-radius:4px;\'>"
            + "<b>" + judul + "</b> <span style=\'color:#777;font-size:0.85em;\'>(dibandingkan dengan " + label + "; sumber: " + data["sumber"] + ")</span>"
            + "<ul style=\'margin:6px 0 0 0;\'>"
            + "<li><b>Dosis:</b> <span style=\'color:" + warna_dosis + ";font-weight:600;\'>" + msg_dosis + "</span> (rentang tepat: " + _fmt_rentang(dmin, dmax) + " mg/kg)</li>"
            + "<li><b>Interval:</b> <span style=\'color:" + _warna(s_int) + ";font-weight:600;\'>" + msg_int + "</span></li>"
            + "</ul></div>")


def _ringkasan_evaluasi(hasil, da_mg, ia):
    """Kotak ringkasan berwarna untuk pemberian AKTUAL + baris ringkas untuk riwayat."""
    rs_min, rs_max = hasil["rs_rentang_mg"]
    lit_min, lit_max = hasil["lit_rentang_mg"]
    rs_imin, rs_imax = hasil["rs"].get("interval_min"), hasil["rs"].get("interval_max")
    lit_imin, lit_imax = hasil.get("lit_interval_rentang", (None, None))

    def kat_dosis(rmin, rmax):
        k, t = _kategori_dosis(da_mg, da_mg, rmin, rmax, "mg")
        if k is None:
            return None, "tidak dapat dibandingkan (pedoman tidak menetapkan dosis rutin)"
        return k, k + ((" (" + t + ")") if t else "")

    def kat_frek(rmin, rmax):
        if rmin is None or rmax is None or not ia:
            return None, "tidak dapat dibandingkan (pedoman tidak menetapkan interval rutin)"
        fa = 24.0 / ia
        k, t = _kategori_dosis(fa, fa, 24.0 / rmax, 24.0 / rmin, "dosis/hari")
        tambahan = _teks_interval(k, ia, ia, rmin, rmax)
        if t:
            return k, k + " (" + t + ((("; " + tambahan) if tambahan else "")) + ")"
        return k, k

    baris = [("Dosis aktual vs Pedoman RS",) + kat_dosis(rs_min, rs_max),
             ("Frekuensi aktual vs Pedoman RS",) + kat_frek(rs_imin, rs_imax),
             ("Dosis aktual vs Literatur Terkini",) + kat_dosis(lit_min, lit_max),
             ("Frekuensi aktual vs Literatur Terkini",) + kat_frek(lit_imin, lit_imax)]
    kats = [k for _, k, _ in baris]
    if "Berlebih" in kats and "Kurang" in kats:
        tingkat, judul = "bahaya", "Pemberian aktual tidak sesuai rujukan (ada yang berlebih dan ada yang kurang)"
    elif "Berlebih" in kats:
        tingkat, judul = "bahaya", "Pemberian aktual BERLEBIH dibanding rujukan"
    elif "Kurang" in kats:
        tingkat, judul = "perhatian", "Pemberian aktual KURANG dibanding rujukan"
    else:
        tingkat, judul = "sesuai", "Dosis dan frekuensi aktual sesuai dengan rujukan"
    isi = ("<ul style='margin:4px 0 0 18px;'>"
           + "".join("<li>" + label + ": <span style='color:" + _WARNA_KATEGORI.get(k, "#6c757d") + ";font-weight:600;'>"
                     + teks + "</span></li>" for label, k, teks in baris)
           + "</ul>")
    ringkas = {"Kode pasien": hasil["kode_pasien"], "Antibiotik": hasil["antibiotik"],
               "Dosis aktual": _fmt_num(da_mg) + " mg", "Frekuensi aktual": _fmt_q(ia)}
    nama_kolom = ["Dosis vs Pedoman Rumah Sakit", "Frekuensi vs Pedoman Rumah Sakit",
                  "Dosis vs Literatur Terkini", "Frekuensi vs Literatur Terkini"]
    for nama, (_, k, teks) in zip(nama_kolom, baris):
        ringkas[nama] = ("<span style='color:" + _WARNA_KATEGORI.get(k, "#6c757d") + ";font-weight:600;'>" + teks + "</span>")
    return _kotak(tingkat, judul, isi), ringkas


def _html_evaluasi(hasil, dm_min, dm_max, im_min, im_max, da, ia):
    """HTML hasil evaluasi (belum diberi penjelasan singkatan)."""
    berat = hasil["berat_kg"] if hasil["berat_kg"] else 1.0
    kotak_ringkas, _ = _ringkasan_evaluasi(hasil, da, ia)
    _kode = hasil.get("kode_pasien", "")
    _kode_html = ("<div style='font-weight:600;color:#0d6efd;margin-bottom:2px;'>Kode pasien: " + _kode + "</div>") if _kode else ""
    html = (_kode_html
            + "<h4>Evaluasi Pemberian " + hasil["antibiotik"] + " -- usia gestasi lahir " + _fmt_usia_minggu(hasil["ga_lahir"])
            + ", usia pascalahir " + str(hasil["pna_hari"]) + " hari, PMA " + _fmt_usia_minggu(hasil["ga_lahir"] + hasil["pna_hari"] / 7.0)
            + ", berat " + _fmt_num(hasil["berat_kg"]) + " kg</h4>"
            + kotak_ringkas
            + "".join(_kotak(*p) for p in _peringatan_pasien(hasil))
            + _tabel_evaluasi(hasil, dm_min, dm_max, im_min, im_max, da, ia)
            + "<h5 style='margin-top:14px;margin-bottom:2px;'>Rincian Kriteria Kesesuaian Pedoman RS</h5>"
            + (_blok_evaluasi_kriteria(hasil, dm_min / berat, dm_max / berat, im_min, "Hitung Manual") if dm_min is not None else "")
            + _blok_evaluasi_kriteria(hasil, da / berat, da / berat, ia, "Aktual Diberikan"))
    if TAMPILKAN_DRUGSCOM_TAMBAHAN and dm_min is not None:
        html += _blok_evaluasi_rentang("Drugs.com / ASHP (AAP &amp; IDSA)", hasil["drugscom"], dm_min / berat, dm_max / berat, im_min, "Hitung Manual")
        html += _blok_evaluasi_rentang("Drugs.com / ASHP (AAP &amp; IDSA)", hasil["drugscom"], da / berat, da / berat, ia, "Aktual Diberikan")
    return html + _legenda_warna() + _baris_disclaimer() + "<div style='color:#999;font-size:0.8em;margin-top:8px;'>Versi aturan " + hasil.get("versi_aturan", "") + "</div>"


def _on_evaluasi_clicked(b):
    with out_evaluasi:
        clear_output()
        try:
            hasil, isian = _baca_semua_evaluasi()
        except ValueError as e:
            display(HTML(_kotak("bahaya", "Data tidak dapat dihitung", str(e))))
            return
        display(HTML(_perjelas_singkatan(_html_evaluasi(hasil, *isian))))
    RIWAYAT_EVALUASI.append(_ringkasan_evaluasi(hasil, isian[4], isian[5])[1])
    _perbarui_riwayat(out_riwayat_eval, RIWAYAT_EVALUASI, "Riwayat evaluasi pada sesi ini")


def reset_eval(kode_baru):
    reset_form(form_eval, kode_baru)
    w_dosis_manual_min.value, w_dosis_manual_max.value = NILAI_AWAL_EVAL["dm_min"], NILAI_AWAL_EVAL["dm_max"]
    w_interval_manual_min.value, w_interval_manual_max.value = NILAI_AWAL_EVAL["im_min"], NILAI_AWAL_EVAL["im_max"]
    w_dosis_aktual.value, w_interval_aktual.value = NILAI_AWAL_EVAL["da"], NILAI_AWAL_EVAL["ia"]
    w_ada_manual.value = False


def _on_pasien_baru_eval(b):
    reset_eval("Pasien " + str(len(RIWAYAT_EVALUASI) + 1))
    with out_evaluasi:
        clear_output()
        display(HTML(_kotak("info", "Formulir siap untuk pasien baru",
                            "Isi data pasien berikutnya beserta dosis dan frekuensi yang diberikan, lalu klik <b>Evaluasi Pemberian</b>. "
                            "Ringkasan pasien sebelumnya tetap tersimpan pada Riwayat di bawah.")))


def _on_salin_dari_rek(b):
    salin_form(form_rek, form_eval)
    with out_evaluasi:
        clear_output()
        display(HTML(_kotak("info", "Data pasien disalin dari bagian Rekomendasi Dosis",
                            "Periksa kembali data pasien, isi dosis dan frekuensi yang diberikan, lalu klik <b>Evaluasi Pemberian</b>.")))


btn_evaluasi.on_click(_on_evaluasi_clicked)
btn_pasien_baru_eval.on_click(_on_pasien_baru_eval)
btn_salin_dari_rek.on_click(_on_salin_dari_rek)

display(widgets.HTML("<div style='color:#555;font-size:0.9em;margin-bottom:6px;'>Bagian ini dapat dipakai <b>tanpa</b> mengisi bagian Rekomendasi Dosis. "
                     "Isi data pasien di bawah. Bila pasien yang sama sudah diisi di bagian Rekomendasi Dosis, klik <b>Salin data pasien dari Rekomendasi Dosis</b> (opsional).</div>"))
display(btn_salin_dari_rek)
tampilkan_form(form_eval)
display(widgets.HTML("<div style='color:#555;font-size:0.9em;margin:4px 0;'>Semua kolom kosong sejak awal; kolom bertanda <i>wajib diisi</i> harus diisi. "
                     "Angka desimal boleh memakai koma atau titik.</div>"))
display(widgets.HTML("<br><b>Dosis &amp; Frekuensi Aktual yang Sudah Diberikan ke Pasien</b> (wajib) -- "
                     "frekuensi diisi sebagai jarak antardosis dalam jam (mis. diberikan tiap 36 jam: isi 36)."))
display(w_dosis_aktual, w_interval_aktual)
display(widgets.HTML("<br><b>Hasil Hitung Manual</b> (opsional) -- centang bila Anda punya hasil hitungan sendiri sebagai pembanding. "
                     "Bila hasil hitungan Anda satu nilai, isi batas bawah = batas atas (begitu pula frekuensi). Bila tidak dicentang, kolom Hitung Manual tidak ditampilkan."))
display(w_ada_manual)
display(w_dosis_manual_min, w_dosis_manual_max, w_interval_manual_min, w_interval_manual_max)
display(widgets.HTML("<br>"))
display(widgets.HBox([btn_evaluasi, btn_pasien_baru_eval]))
display(widgets.HTML("<div style='color:#555;font-size:0.9em;margin:4px 0;'>Untuk pasien lain: klik <b>Pasien Baru</b> (formulir dikosongkan), "
                     "atau langsung ubah data lalu klik <b>Evaluasi Pemberian</b> lagi. Tidak perlu kembali atau memuat ulang halaman.</div>"))
display(out_evaluasi, out_riwayat_eval)
_perbarui_riwayat(out_riwayat_eval, RIWAYAT_EVALUASI, "Riwayat evaluasi pada sesi ini")
```

## 4. Unduh Laporan (PDF)

Tersedia dua tombol yang **terpisah**:

- **Unduh PDF Rekomendasi Dosis** — memakai data pasien yang sedang terisi di bagian Rekomendasi Dosis.
- **Unduh PDF Evaluasi Pemberian** — memakai data pasien, hitung manual (bila dicentang), dan pemberian aktual yang sedang terisi di bagian Evaluasi Pemberian (memuat tabel rekomendasi pasien tersebut dan hasil evaluasinya).

Laporan memuat kotak peringatan berwarna yang sama dengan tampilan di layar, serta keterangan singkatan.

> Bila pembuatan PDF gagal di lingkungan Anda (mis. pustaka tidak tersedia), aplikasi otomatis menyediakan berkas **HTML** sebagai cadangan -- buka berkas itu di browser lalu gunakan *Print → Save as PDF*.


```python
# ---- Unduh laporan (PDF) ----
!pip install xhtml2pdf --quiet

import datetime, io, re as _re

def _laporan_html(hasil, antibiotik, sertakan_eval, dm_min_mg, dm_max_mg, i_man_min, i_man_max, da_mg, i_akt):
    """Menyusun dokumen HTML mandiri (A4 landscape) untuk dikonversi ke PDF.
    Memakai ulang _tabel_rekomendasi & _tabel_evaluasi agar tampilan konsisten
    dengan output di notebook (termasuk warna kategori)."""
    now = datetime.datetime.now().strftime("%d-%m-%Y %H:%M")
    kode = hasil.get("kode_pasien", "") or "-"
    if not hasil.get("data_ginjal_tersedia", True):
        ginjal = "Data tidak diisi -- diasumsikan normal sesuai usia (penyesuaian ginjal tidak dapat dinilai)"
    else:
        ginjal = "Normal (tidak ada indikasi gangguan)" if hasil["ginjal_normal"] else "Ada indikasi gangguan ginjal"
    css = ("@page { size: A4 landscape; margin: 1.2cm; }"
           "body { font-family: Helvetica, Arial, sans-serif; font-size: 9px; color: #222; }"
           "h2 { font-size: 15px; margin: 0 0 2px 0; }"
           "h3 { font-size: 11px; margin: 12px 0 4px 0; }"
           "table { border-collapse: collapse; width: 100%; margin: 6px 0; }"
           "th, td { border: 1px solid #999; padding: 4px 6px; text-align: left; }"
           "th { background-color: #e9ecef; }"
           ".info td { border: none; padding: 1px 6px 1px 0; }"
           ".kecil { color: #666; font-size: 8px; }")

    info = ("<table class=\'info\'>"
            "<tr><td><b>Kode pasien</b></td><td>" + kode + "</td>"
            "<td><b>Antibiotik</b></td><td>" + antibiotik + "</td></tr>"
            "<tr><td><b>Usia gestasi lahir</b></td><td>" + _fmt_usia_minggu(hasil["ga_lahir"]) + "</td>"
            "<td><b>Usia pascalahir</b></td><td>" + str(hasil["pna_hari"]) + " hari</td></tr>"
            "<tr><td><b>Usia pascamenstruasi (PMA) saat ini</b></td><td>" + _fmt_usia_minggu(hasil["ga_lahir"] + hasil["pna_hari"] / 7.0) + "</td>"
            "<td><b>Berat badan</b></td><td>" + _fmt_num(hasil["berat_kg"]) + " kg</td></tr>"
            "<tr><td><b>Fungsi ginjal</b></td><td colspan=\'3\'>" + ginjal + "</td></tr>"
            "</table>")

    peringatan = "".join(_kotak(*p) for p in _peringatan_rekomendasi(hasil))
    bagian_rekom = "<h3>Rekomendasi Dosis &amp; Frekuensi</h3>" + _tabel_rekomendasi(hasil)
    bagian_eval = ""
    if sertakan_eval:
        bagian_eval = ("<h3>Evaluasi Pemberian</h3>"
                       + _ringkasan_evaluasi(hasil, da_mg, i_akt)[0]
                       + _tabel_evaluasi(hasil, dm_min_mg, dm_max_mg, i_man_min, i_man_max, da_mg, i_akt))
    isi = _perjelas_singkatan(info + peringatan + bagian_rekom + bagian_eval + _legenda_warna())

    disclaimer = ("<p class=\'kecil\'>Laporan ini dibuat oleh alat bantu keputusan klinis (clinical decision support) dan "
                  "BUKAN pengganti penilaian apoteker klinis/DPJP. Gentamisin, Amikasin, dan Vankomisin adalah antibiotik "
                  "indeks terapi sempit -- selalu verifikasi ulang dengan SPO/formulary terbaru, data laboratorium (fungsi "
                  "ginjal, kadar obat), dan pemantauan kadar obat. Alat ini belum divalidasi secara klinis/regulatori. "
                  "Sumber: lihat Daftar Rujukan pada notebook.</p>")

    return ("<html><head><meta charset=\'utf-8\'><style>" + css + "</style></head><body>"
            "<h2>Laporan Dosis &amp; Evaluasi Antibiotik Neonatus</h2>"
            "<div class=\'kecil\'>Dibuat: " + now + "</div>"
            + isi + disclaimer
            + "</body></html>")


def _nama_file_aman(teks):
    teks = (teks or "pasien").strip()
    teks = _re.sub(r"[^A-Za-z0-9_-]+", "_", teks)
    return teks or "pasien"


def _simpan_dan_unduh(html, dasar):
    """Coba buat PDF; bila gagal, sediakan HTML sebagai cadangan. Lalu unduh."""
    pdf_ok = False
    nama_pdf = dasar + ".pdf"
    try:
        from xhtml2pdf import pisa
        with open(nama_pdf, "wb") as f:
            status = pisa.CreatePDF(html, dest=f)
        pdf_ok = not status.err
    except Exception as e:
        print("Info: pembuatan PDF tidak berhasil (" + type(e).__name__ + "). Menyediakan berkas HTML sebagai cadangan.")
    if pdf_ok:
        berkas = nama_pdf
        print("Laporan PDF berhasil dibuat: " + berkas)
    else:
        berkas = dasar + ".html"
        with open(berkas, "w", encoding="utf-8") as f:
            f.write(html)
        print("Berkas HTML dibuat: " + berkas + " -- buka di browser lalu 'Print > Save as PDF'.")
    try:
        from google.colab import files
        files.download(berkas)
    except Exception:
        try:
            from IPython.display import FileLink
            display(FileLink(berkas))
        except Exception:
            print("Berkas tersimpan di: " + berkas)
    return berkas


btn_unduh_rek = widgets.Button(description="Unduh PDF Rekomendasi Dosis", button_style="success", icon="download",
                               layout=widgets.Layout(width="300px"))
btn_unduh_eval = widgets.Button(description="Unduh PDF Evaluasi Pemberian", button_style="success", icon="download",
                                layout=widgets.Layout(width="300px"))
out_pdf = widgets.Output()


def _on_unduh_rek(b):
    with out_pdf:
        clear_output()
        try:
            hasil = hitung_dari_form(form_rek)
        except ValueError as e:
            display(HTML(_kotak("bahaya", "Data tidak dapat dihitung", str(e))))
            return
        html = _laporan_html(hasil, hasil["antibiotik"], False, None, None, None, None, None, None)
        stamp = datetime.datetime.now().strftime("%Y%m%d_%H%M%S")
        _simpan_dan_unduh(html, "laporan_rekomendasi_" + _nama_file_aman(hasil["kode_pasien"]) + "_" + hasil["antibiotik"] + "_" + stamp)


def _on_unduh_eval(b):
    with out_pdf:
        clear_output()
        try:
            hasil, isian = _baca_semua_evaluasi()
        except ValueError as e:
            display(HTML(_kotak("bahaya", "Data tidak dapat dihitung", str(e))))
            return
        html = _laporan_html(hasil, hasil["antibiotik"], True, *isian)
        stamp = datetime.datetime.now().strftime("%Y%m%d_%H%M%S")
        _simpan_dan_unduh(html, "laporan_evaluasi_" + _nama_file_aman(hasil["kode_pasien"]) + "_" + hasil["antibiotik"] + "_" + stamp)


btn_unduh_rek.on_click(_on_unduh_rek)
btn_unduh_eval.on_click(_on_unduh_eval)

display(widgets.HTML("<b>Unduh Laporan</b> -- PDF Rekomendasi memakai data di bagian Rekomendasi Dosis; PDF Evaluasi memakai data di bagian Evaluasi Pemberian."))
display(widgets.HBox([btn_unduh_rek, btn_unduh_eval]))
display(out_pdf)
```

## 5. Pemrosesan Massal (Batch) — Evaluasi Retrospektif Banyak Pasien

Bagian ini memungkinkan Anda mengevaluasi **banyak pasien sekaligus** (mis. untuk kajian retrospektif penggunaan antibiotik).

**Alur kerja:**
1. **Unduh Template Excel** — berisi sheet `Petunjuk` (keterangan tiap kolom) dan sheet `Data Pasien` (tempat mengisi data). Dua baris CONTOH disertakan sebagai panduan format dan akan **diabaikan otomatis** saat diproses.
2. **Isi data** — satu baris = satu pasien. Kolom wajib: kode pasien, antibiotik, usia gestasi, usia pascalahir, berat badan, dosis & frekuensi aktual. Kolom lain opsional.
3. **Unggah berkas** yang sudah diisi, lalu klik **Proses Data & Unduh Hasil**.

**Keluaran:**
- **Excel (.xlsx)** berisi sheet: `Hasil per Pasien` (tabel lengkap perhitungan & evaluasi **dosis dan frekuensi**, dengan pewarnaan kategori), `Ringkasan Kesesuaian` (distribusi Sesuai/Berlebih/Kurang **untuk dosis DAN frekuensi**, pada perhitungan **AKTUAL maupun MANUAL**, keseluruhan & per antibiotik), `Analisis Statistik` (**dosis dan frekuensi**), **`Interpretasi & Kesimpulan`** (narasi otomatis atas hasil evaluasi & uji statistik, termasuk penilaian reliabilitas aplikasi vs hitung manual, dengan poin-poin penting disorot), `Peringatan` (baris yang dilewati beserta alasannya), dan `Keterangan Singkatan`.
- **PDF ringkasan** berisi distribusi kesesuaian, hasil uji statistik, dan kesimpulannya.

**Analisis statistik otomatis yang dijalankan** (untuk **DOSIS dan FREKUENSI**):
- **Uji beda berpasangan** antara nilai aktual dan titik tengah rentang rekomendasi, dijalankan terpisah per antibiotik: untuk **dosis** memakai skala mg/kg, untuk **frekuensi** memakai skala dosis/hari (24 ÷ interval). Aplikasi otomatis memilih **paired t-test** bila selisih terdistribusi normal (uji Shapiro-Wilk, p>0.05) atau **Wilcoxon signed-rank** bila tidak normal. Dilewati bila n<3.
- **Uji McNemar** untuk membandingkan proporsi pasien berkategori "Sesuai" menurut Pedoman RS dibanding menurut Literatur (berpasangan pada pasien yang sama), dijalankan untuk **dosis** dan **frekuensi**. Memakai uji eksak binomial bila pasangan diskordan <25, atau khi-kuadrat dengan koreksi kontinuitas bila >=25. (Kesesuaian frekuensi memakai rentang interval Pedoman RS, yang tidak membedakan usia.)

> Statistik dihasilkan otomatis, namun **interpretasinya tetap memerlukan penilaian klinis** dan ukuran sampel yang memadai. Nilai p bukan pengganti pertimbangan klinis.


```python
# Instalasi pustaka untuk pemrosesan massal & statistik
!pip install openpyxl scipy xhtml2pdf --quiet

# -*- coding: utf-8 -*-
# Mesin pemrosesan massal (batch). Fungsi-fungsi ini mengandalkan fungsi dari
# sel sebelumnya di notebook: bangun_rekomendasi, _kategori_dosis, _fmt_num, hitung_pma.
import io
import datetime
import pandas as pd
import numpy as np

# ---------------------------------------------------------------------------
# TEMPLATE
# ---------------------------------------------------------------------------
KOLOM_TEMPLATE = [
    "kode_pasien", "antibiotik", "usia_gestasi_lahir_minggu", "usia_pascalahir_hari",
    "berat_badan_kg", "kreatinin_serum_mg_dl", "produksi_urin_ml_kg_jam",
    "hipotermia_terapeutik", "terapi_cox_inhibitor", "sepsis_berat",
    "dosis_manual_min_mg", "dosis_manual_max_mg",
    "frekuensi_manual_min_jam", "frekuensi_manual_max_jam",
    "dosis_aktual_mg", "frekuensi_aktual_jam",
    "usia_gestasi_lahir_hari",
]

CONTOH_BARIS = [
    ["CONTOH-hapus baris ini", "Gentamisin", 32, 3, 1.5, "", "", "Tidak", "Tidak", "Tidak", 7.5, 11.25, 24, 48, 10, 24, 4],
    ["CONTOH-hapus baris ini", "Vankomisin", 30, 5, 1.2, 0.4, 2.0, "Tidak", "Tidak", "Ya", 15, 18, 12, 12, 18, 12, ""],
]

PETUNJUK = [
    ("kode_pasien", "Kode pasien unik, tanpa nama atau nomor rekam medis (mis. Pasien 1). WAJIB. Baris yang diawali 'CONTOH' akan diabaikan otomatis."),
    ("antibiotik", "Salah satu dari: Gentamisin, Amikasin, Vankomisin. WAJIB (persis, huruf besar di awal)."),
    ("usia_gestasi_lahir_minggu", "Usia gestasi saat lahir, minggu LENGKAP (mis. 32 untuk 32 minggu 4 hari). WAJIB."),
    ("usia_pascalahir_hari", "Usia pascalahir dalam hari, 0-28 (mis. 3). WAJIB. Baris di atas 28 hari dilewati."),
    ("berat_badan_kg", "Berat badan saat ini dalam KILOGRAM (mis. 1,5; bukan 1500 gram). WAJIB."),
    ("kreatinin_serum_mg_dl", "Kreatinin serum (mg/dL). OPSIONAL - kosongkan bila tidak ada (fungsi ginjal lalu diasumsikan normal dan ditandai)."),
    ("produksi_urin_ml_kg_jam", "Produksi urin (mL/kg/jam). OPSIONAL - kosongkan bila tidak ada."),
    ("hipotermia_terapeutik", "Ya / Tidak. Default Tidak bila kosong."),
    ("terapi_cox_inhibitor", "Ya / Tidak (indometasin/ibuprofen). Default Tidak bila kosong."),
    ("sepsis_berat", "Ya / Tidak (khusus Vankomisin: MRSA/meningitis/endokarditis). Default Tidak bila kosong."),
    ("dosis_manual_min_mg", "Batas MIN dosis hitung manual (mg total). OPSIONAL."),
    ("dosis_manual_max_mg", "Batas MAKS dosis hitung manual (mg total). OPSIONAL (isi sama dengan MIN bila nilai tunggal)."),
    ("frekuensi_manual_min_jam", "Interval frekuensi manual MIN/tersering (jam, mis. 24). OPSIONAL."),
    ("frekuensi_manual_max_jam", "Interval frekuensi manual MAKS/terjarang (jam, mis. 48). OPSIONAL (isi sama dengan MIN bila nilai tunggal)."),
    ("dosis_aktual_mg", "Dosis aktual yang diberikan (mg total). WAJIB untuk evaluasi."),
    ("frekuensi_aktual_jam", "Frekuensi aktual: diberikan setiap ... jam (mis. 24). WAJIB untuk evaluasi."),
    ("usia_gestasi_lahir_hari", "Tambahan hari usia gestasi saat lahir, 0-6 (mis. 4 untuk 32 minggu 4 hari). OPSIONAL, kosong = 0."),
]


def _style_header(ws, nrow, ncol, openpyxl_styles):
    Font, PatternFill, Border, Side, Alignment = openpyxl_styles
    fill = PatternFill("solid", fgColor="1F4E78")
    thin = Side(style="thin", color="AAAAAA")
    for c in range(1, ncol + 1):
        cell = ws.cell(row=nrow, column=c)
        cell.font = Font(name="Arial", bold=True, color="FFFFFF", size=10)
        cell.fill = fill
        cell.alignment = Alignment(horizontal="center", vertical="center", wrap_text=True)
        cell.border = Border(left=thin, right=thin, top=thin, bottom=thin)


def buat_template_excel(nama_file=None):
    """Membuat berkas Excel template yang harus diisi pengguna."""
    import openpyxl
    from openpyxl.styles import Font, PatternFill, Border, Side, Alignment
    styles = (Font, PatternFill, Border, Side, Alignment)
    thin = Side(style="thin", color="CCCCCC")

    wb = openpyxl.Workbook()
    # --- Sheet Petunjuk ---
    ws0 = wb.active
    ws0.title = "Petunjuk"
    ws0["A1"] = "TEMPLATE DATA EVALUASI DOSIS ANTIBIOTIK NEONATUS (BATCH)"
    ws0["A1"].font = Font(name="Arial", bold=True, size=13)
    ws0["A3"] = ("Isi data pada sheet 'Data Pasien'. Satu baris = satu pasien. "
                 "Hapus dua baris CONTOH sebelum memproses (atau biarkan - baris CONTOH diabaikan otomatis). "
                 "Kolom OPSIONAL boleh dikosongkan.")
    ws0["A3"].font = Font(name="Arial", size=10)
    ws0["A3"].alignment = Alignment(wrap_text=True, vertical="top")
    ws0.merge_cells("A3:D5")
    ws0["A7"] = "Kolom"
    ws0["B7"] = "Keterangan"
    for c in ("A7", "B7"):
        ws0[c].font = Font(name="Arial", bold=True, color="FFFFFF")
        ws0[c].fill = PatternFill("solid", fgColor="1F4E78")
        ws0[c].border = Border(left=thin, right=thin, top=thin, bottom=thin)
    r = 8
    for nama, ket in PETUNJUK:
        ws0.cell(row=r, column=1, value=nama).font = Font(name="Arial", bold=True, size=9)
        ws0.cell(row=r, column=2, value=ket).font = Font(name="Arial", size=9)
        ws0.cell(row=r, column=2).alignment = Alignment(wrap_text=True, vertical="top")
        for cc in (1, 2):
            ws0.cell(row=r, column=cc).border = Border(left=thin, right=thin, top=thin, bottom=thin)
        r += 1
    ws0.column_dimensions["A"].width = 30
    ws0.column_dimensions["B"].width = 95

    # --- Sheet Data Pasien ---
    ws = wb.create_sheet("Data Pasien")
    for j, nama in enumerate(KOLOM_TEMPLATE, start=1):
        ws.cell(row=1, column=j, value=nama)
    _style_header(ws, 1, len(KOLOM_TEMPLATE), styles)
    for i, contoh in enumerate(CONTOH_BARIS, start=2):
        for j, val in enumerate(contoh, start=1):
            cell = ws.cell(row=i, column=j, value=val)
            cell.font = Font(name="Arial", italic=True, size=9, color="888888")
    ws.freeze_panes = "A2"
    ws.auto_filter.ref = "A1:" + openpyxl.utils.get_column_letter(len(KOLOM_TEMPLATE)) + "1"
    for j, nama in enumerate(KOLOM_TEMPLATE, start=1):
        ws.column_dimensions[openpyxl.utils.get_column_letter(j)].width = max(14, min(26, len(nama) + 2))

    if nama_file is None:
        nama_file = "template_evaluasi_dosis_neonatus.xlsx"
    wb.save(nama_file)
    return nama_file


# ---------------------------------------------------------------------------
# PARSING & PEMROSESAN
# ---------------------------------------------------------------------------
def _ya(x):
    return str(x).strip().lower() in ("ya", "yes", "y", "true", "1", "benar")


def _num(x):
    if x is None:
        return None
    if isinstance(x, float) and np.isnan(x):
        return None
    s = str(x).strip().replace(",", ".")
    if s == "" or s.lower() in ("nan", "none"):
        return None
    try:
        return float(s)
    except ValueError:
        return None


def _kat_dosis_teks(nilai_min, nilai_max, rmin, rmax, fn_kategori, satuan="mg"):
    k, p = fn_kategori(nilai_min, nilai_max, rmin, rmax, satuan)
    if k is None:
        return "", ""
    return k, (p or "")


def _manual_vs_teks(mmin, mmax, rmin, rmax, fn_kategori, satuan="mg", interval=None):
    """Teks Manual vs rekomendasi. `interval` = (m_imin, m_imax, r_imin, r_imax) untuk
    frekuensi, agar selisih interval dalam jam ikut ditulis."""
    if rmin is None or rmax is None or mmin is None or mmax is None:
        return ""
    if rmin == rmax:
        k, p = fn_kategori(mmin, mmax, rmin, rmax, satuan)
        if k is None:
            return ""
        if interval is not None:
            p = ((p + "; ") if p else "") + _teks_interval_batch(k, *interval)
            p = p.rstrip("; ")
        return k + ((" (" + p + ")") if p else "")

    def bound(v, ref, iv=None, iref=None):
        if ref == 0:
            return "n/a"
        d = v - ref
        if abs(d) <= 1e-6:
            return "sama"
        tanda = "+" if d > 0 else "-"
        teks = (tanda + fn_fmt_batch(abs(d)) + " " + satuan + "; " + tanda
                + _fmt_pct_batch(d / ref * 100))
        if iv is not None and iref is not None and iv != iref:
            teks += "; interval " + fn_fmt_batch(abs(iv - iref)) + " jam " + ("lebih pendek" if iv < iref else "lebih panjang")
        return ("lebih besar" if d > 0 else "lebih kecil") + " (" + teks + ")"

    if interval is not None:
        m_imin, m_imax, r_imin, r_imax = interval
        return "Min: " + bound(mmin, rmin, m_imax, r_imax) + " | Maks: " + bound(mmax, rmax, m_imin, r_imin)
    return "Min: " + bound(mmin, rmin) + " | Maks: " + bound(mmax, rmax)


def fn_fmt_batch(x):
    x = round(float(x), 3)
    if abs(x - round(x)) < 1e-9:
        return str(int(round(x)))
    return ("%.3f" % x).rstrip("0").rstrip(".").replace(".", ",")


def _fmt_pct_batch(pct):
    return ("{0:.2f}%" if abs(pct) < 0.1 else "{0:.1f}%").format(abs(pct)).replace(".", ",")


def _teks_interval_batch(kategori, n_imin, n_imax, r_imin, r_imax):
    if kategori == "Berlebih" and None not in (n_imin, r_imin):
        return "interval " + fn_fmt_batch(r_imin - n_imin) + " jam lebih pendek dari interval " + str(r_imin) + " jam"
    if kategori == "Kurang" and None not in (n_imax, r_imax):
        return "interval " + fn_fmt_batch(n_imax - r_imax) + " jam lebih panjang dari interval " + str(r_imax) + " jam"
    return ""


def _freq_teks(aktual, rekom):
    if rekom is None or aktual is None:
        return ""
    if aktual == rekom:
        return "Sesuai"
    return "Lebih pendek" if aktual < rekom else "Lebih panjang"


def _fph(iv):
    """Frekuensi per hari = 24 / interval jam."""
    if iv is None or iv == 0:
        return None
    return 24.0 / float(iv)


def _kat_freq_aktual_teks(a_interval, imin, imax, fn_kategori):
    """Kategori frekuensi AKTUAL (interval tunggal) vs rentang interval rekomendasi,
    dihitung pada skala dosis/hari. Mengembalikan (kategori, pct)."""
    if a_interval is None or imin is None or imax is None:
        return "", ""
    fa = _fph(a_interval)
    fmin, fmax = _fph(imax), _fph(imin)  # interval max -> freq min, dst.
    k, p = fn_kategori(fa, fa, fmin, fmax, "dosis/hari")
    if k in ("Berlebih", "Kurang"):
        p = (p or "") + "; " + _teks_interval_batch(k, a_interval, a_interval, imin, imax)
    return (k or ""), (p or "")


def _manual_freq_teks(m_imin, m_imax, r_imin, r_imax, fn_kategori):
    """Kategori frekuensi MANUAL (rentang interval) vs rentang interval rekomendasi,
    pada skala dosis/hari. Reuse _manual_vs_teks."""
    if None in (m_imin, m_imax, r_imin, r_imax):
        return ""
    fm_min, fm_max = _fph(m_imax), _fph(m_imin)
    fr_min, fr_max = _fph(r_imax), _fph(r_imin)
    return _manual_vs_teks(fm_min, fm_max, fr_min, fr_max, fn_kategori, "dosis/hari",
                           (m_imin, m_imax, r_imin, r_imax))


def _kat_manual_dosis(dm_min, dm_max, rmin, rmax, fn_kategori):
    """Kategori TUNGGAL (Sesuai/Berlebih/Kurang) rentang dosis manual vs rentang
    rekomendasi -- untuk keperluan ringkasan/statistik (bukan teks dua-batas)."""
    if None in (dm_min, dm_max, rmin, rmax):
        return ""
    k, _ = fn_kategori(dm_min, dm_max, rmin, rmax)
    return k or ""


def _kat_manual_freq(m_imin, m_imax, r_imin, r_imax, fn_kategori):
    """Kategori TUNGGAL frekuensi manual (rentang interval) vs rentang rekomendasi,
    pada skala dosis/hari -- untuk ringkasan/statistik."""
    if None in (m_imin, m_imax, r_imin, r_imax):
        return ""
    fm_min, fm_max = _fph(m_imax), _fph(m_imin)
    fr_min, fr_max = _fph(r_imax), _fph(r_imin)
    k, _ = fn_kategori(fm_min, fm_max, fr_min, fr_max)
    return k or ""


def _freq_teks_rentang(aktual, imin, imax):
    """Perbandingan frekuensi aktual terhadap rentang interval RS (mis. 24-48)."""
    if imin is None or imax is None or aktual is None:
        return ""
    if imin <= aktual <= imax:
        return "Sesuai"
    return "Lebih pendek" if aktual < imin else "Lebih panjang"


def _fmt_q_rentang_teks(imin, imax):
    if imin is None and imax is None:
        return "Dosis tunggal"
    if imin == imax:
        return "tiap " + str(imin) + " jam"
    return "tiap " + str(imin) + "-" + str(imax) + " jam"


def proses_dataframe(df, fn_bangun, fn_kategori, fn_fmt):
    """Memproses DataFrame input -> (records list, warnings list).
    fn_bangun=bangun_rekomendasi, fn_kategori=_kategori_dosis, fn_fmt=_fmt_num."""
    df = df.copy()
    df.columns = [str(c).strip() for c in df.columns]
    records = []
    warnings = []
    for idx, row in df.iterrows():
        kode = str(row.get("kode_pasien", "")).strip()
        if kode == "" or kode.lower().startswith("contoh") or kode.lower() == "nan":
            continue
        antibiotik = str(row.get("antibiotik", "")).strip().capitalize()
        if antibiotik not in ("Gentamisin", "Amikasin", "Vankomisin"):
            warnings.append("Baris '" + kode + "': antibiotik '" + str(row.get('antibiotik')) + "' tidak dikenali - dilewati.")
            continue
        ga = _num(row.get("usia_gestasi_lahir_minggu"))
        ga_hari = _num(row.get("usia_gestasi_lahir_hari"))
        if ga is not None and ga_hari is not None:
            if not (0 <= ga_hari <= 6) or abs(ga - round(ga)) > 1e-9:
                warnings.append("Baris '" + kode + "': usia gestasi harus diisi minggu lengkap + hari 0-6 - dilewati.")
                continue
            ga = ga + ga_hari / 7.0
        pna = _num(row.get("usia_pascalahir_hari"))
        bb = _num(row.get("berat_badan_kg"))
        if ga is None or pna is None or bb is None or bb <= 0:
            warnings.append("Baris '" + kode + "': data usia gestasi / pascalahir / berat tidak lengkap - dilewati.")
            continue
        pna = int(round(pna))
        if pna < 0 or pna > USIA_PASCALAHIR_MAKS_HARI:
            warnings.append("Baris '" + kode + "': usia pascalahir " + str(pna) + " hari di luar cakupan kalkulator (0-28 hari) - dilewati.")
            continue
        if not (22 <= ga <= 43) or not (0.3 <= bb <= 10):
            warnings.append("Baris '" + kode + "': usia gestasi (22-43 minggu) atau berat badan (0,3-10 kg) di luar rentang - dilewati. "
                            "Periksa satuan (berat dalam kilogram).")
            continue
        scr = _num(row.get("kreatinin_serum_mg_dl"))
        uop = _num(row.get("produksi_urin_ml_kg_jam"))
        hipo = _ya(row.get("hipotermia_terapeutik"))
        cox = _ya(row.get("terapi_cox_inhibitor"))
        berat_sepsis = _ya(row.get("sepsis_berat"))
        dm_min = _num(row.get("dosis_manual_min_mg"))
        dm_max = _num(row.get("dosis_manual_max_mg"))
        f_man_min = _num(row.get("frekuensi_manual_min_jam"))
        f_man_max = _num(row.get("frekuensi_manual_max_jam"))
        da = _num(row.get("dosis_aktual_mg"))
        f_akt = _num(row.get("frekuensi_aktual_jam"))

        hasil = fn_bangun(antibiotik, ga, pna, bb, scr=scr, uop=uop,
                          hipotermia=hipo, cox_inhibitor=cox, sepsis_berat=berat_sepsis,
                          kode_pasien=kode)
        rs_min_mg, rs_max_mg = hasil["rs_rentang_mg"]
        lit_min_mg, lit_max_mg = hasil["lit_rentang_mg"]
        rs_min_kg, rs_max_kg = hasil["rs_rentang_per_kg"]
        lit_min_kg, lit_max_kg = hasil["lit_rentang_per_kg"]
        rs_freq_min = hasil["rs"].get("interval_min")
        rs_freq_max = hasil["rs"].get("interval_max")
        rs_ga_freq = hasil["rs_ga"]["interval_jam"]
        lit_ifmin, lit_ifmax = hasil.get("lit_interval_rentang", (None, None))

        # kategori dosis aktual (titik) vs rentang
        ka_rs, pa_rs = _kat_dosis_teks(da, da, rs_min_mg, rs_max_mg, fn_kategori) if da is not None else ("", "")
        ka_lit, pa_lit = _kat_dosis_teks(da, da, lit_min_mg, lit_max_mg, fn_kategori) if da is not None else ("", "")
        # kategori manual dosis (rentang) vs rentang
        man_rs = _manual_vs_teks(dm_min, dm_max, rs_min_mg, rs_max_mg, fn_kategori)
        man_lit = _manual_vs_teks(dm_min, dm_max, lit_min_mg, lit_max_mg, fn_kategori)

        f_akt_int = int(f_akt) if f_akt is not None else None
        f_man_min_int = int(f_man_min) if f_man_min is not None else None
        f_man_max_int = int(f_man_max) if f_man_max is not None else None

        # kategori FREKUENSI aktual (Sesuai/Berlebih/Kurang, skala dosis/hari)
        kfa_rs, pfa_rs = _kat_freq_aktual_teks(f_akt_int, rs_freq_min, rs_freq_max, fn_kategori)
        kfa_ga, pfa_ga = _kat_freq_aktual_teks(f_akt_int, rs_ga_freq, rs_ga_freq, fn_kategori)
        kfa_lit, pfa_lit = _kat_freq_aktual_teks(f_akt_int, lit_ifmin, lit_ifmax, fn_kategori)
        # kategori FREKUENSI manual (rentang) vs rentang -- TEKS dua-batas utk tabel
        mf_rs = _manual_freq_teks(f_man_min_int, f_man_max_int, rs_freq_min, rs_freq_max, fn_kategori)
        mf_ga = _manual_freq_teks(f_man_min_int, f_man_max_int, rs_ga_freq, rs_ga_freq, fn_kategori)
        mf_lit = _manual_freq_teks(f_man_min_int, f_man_max_int, lit_ifmin, lit_ifmax, fn_kategori)
        # kategori TUNGGAL manual (untuk ringkasan Sesuai/Berlebih/Kurang)
        kma_rs = _kat_manual_dosis(dm_min, dm_max, rs_min_mg, rs_max_mg, fn_kategori)
        kma_lit = _kat_manual_dosis(dm_min, dm_max, lit_min_mg, lit_max_mg, fn_kategori)
        kmf_rs = _kat_manual_freq(f_man_min_int, f_man_max_int, rs_freq_min, rs_freq_max, fn_kategori)
        kmf_ga = _kat_manual_freq(f_man_min_int, f_man_max_int, rs_ga_freq, rs_ga_freq, fn_kategori)
        kmf_lit = _kat_manual_freq(f_man_min_int, f_man_max_int, lit_ifmin, lit_ifmax, fn_kategori)

        def _kt(k, p):
            return (k + ((" (" + p + ")") if p else "")) if k else "-"

        def rentang_txt(a, b):
            if a is None:
                return "-"
            return fn_fmt(b) + " mg" if a == b else fn_fmt(a) + "-" + fn_fmt(b) + " mg"

        rec = {
            "Kode Pasien": kode,
            "Antibiotik": antibiotik,
            "Usia Gestasi (mgg)": ga,
            "Usia Pascalahir (hari)": pna,
            "PMA (mgg)": hasil["pma_minggu"],
            "Berat (kg)": bb,
            "Fungsi Ginjal": ("Tidak tersedia (diasumsikan normal)" if not hasil["data_ginjal_tersedia"]
                              else ("Normal" if hasil["ginjal_normal"] else "Terganggu")),
            "Dosis RS (mg)": rentang_txt(rs_min_mg, rs_max_mg),
            "Frek RS": _fmt_q_rentang_teks(rs_freq_min, rs_freq_max),
            "Dosis Literatur (mg)": rentang_txt(lit_min_mg, lit_max_mg),
            "Frek Literatur": _fmt_q_rentang_teks(lit_ifmin, lit_ifmax),
            "Dosis Manual (mg)": rentang_txt(dm_min, dm_max) if dm_min is not None else "-",
            "Frek Manual": _fmt_q_rentang_teks(f_man_min_int, f_man_max_int) if f_man_min_int is not None else "-",
            "Dosis Aktual (mg)": (fn_fmt(da) + " mg") if da is not None else "-",
            "Dosis Aktual (mg/kg/dosis)": (fn_fmt(da / bb) + " mg/kg") if (da is not None and bb) else "-",
            "Frek Aktual": ("tiap " + str(f_akt_int) + " jam") if f_akt_int is not None else "-",
            "Aktual vs RS (dosis)": _kt(ka_rs, pa_rs),
            "Aktual vs RS (frek)": _kt(kfa_rs, pfa_rs),
            "Aktual vs Literatur (dosis)": _kt(ka_lit, pa_lit),
            "Aktual vs Literatur (frek)": _kt(kfa_lit, pfa_lit),
            "Manual vs RS (dosis)": man_rs or "-",
            "Manual vs RS (frek)": mf_rs or "-",
            "Manual vs Literatur (dosis)": man_lit or "-",
            "Manual vs Literatur (frek)": mf_lit or "-",
            # data numerik untuk statistik (tidak ditampilkan sbg kolom utama)
            "_da_perkg": (da / bb) if da is not None else None,
            "_rs_mid_perkg": (rs_min_kg + rs_max_kg) / 2 if rs_min_kg is not None else None,
            "_lit_mid_perkg": (lit_min_kg + lit_max_kg) / 2 if lit_min_kg is not None else None,
            "_ka_rs": ka_rs, "_ka_lit": ka_lit,
            # frekuensi (skala dosis/hari) untuk statistik
            "_fa_perhari": _fph(f_akt_int),
            "_rs_freq_mid": ((_fph(rs_freq_max) + _fph(rs_freq_min)) / 2.0) if (rs_freq_min is not None and rs_freq_max is not None) else None,
            "_lit_freq_mid": ((_fph(lit_ifmax) + _fph(lit_ifmin)) / 2.0) if (lit_ifmin is not None and lit_ifmax is not None) else None,
            # titik tengah MANUAL untuk uji beda (dosis mg/kg & frekuensi dosis/hari)
            "_dm_mid_perkg": (((dm_min + dm_max) / 2.0) / bb) if (dm_min is not None and dm_max is not None and bb) else None,
            "_fm_mid_perhari": ((_fph(f_man_max_int) + _fph(f_man_min_int)) / 2.0) if (f_man_min_int is not None and f_man_max_int is not None) else None,
            "_kfa_rs": kfa_rs, "_kfa_ga": kfa_ga, "_kfa_lit": kfa_lit,
            # kategori tunggal MANUAL (dosis & frekuensi) untuk ringkasan
            "_kma_rs": kma_rs, "_kma_lit": kma_lit,
            "_kmf_rs": kmf_rs, "_kmf_ga": kmf_ga, "_kmf_lit": kmf_lit,
        }
        records.append(rec)
    return records, warnings


# ---------------------------------------------------------------------------
# STATISTIK OTOMATIS
# ---------------------------------------------------------------------------
def _uji_beda_berpasangan(a, b):
    """Uji beda berpasangan a vs b (mis. dosis aktual per kg vs titik tengah
    rekomendasi per kg). Otomatis memilih paired t-test (bila selisih normal)
    atau Wilcoxon signed-rank (bila tidak normal)."""
    import warnings as _w
    from scipy import stats
    a = np.asarray(a, dtype=float)
    b = np.asarray(b, dtype=float)
    n = len(a)
    out = {"n": n}
    if n < 3:
        out.update(cukup=False, pesan="n<3, uji tidak dilakukan")
        return out
    diff = a - b
    out["median_selisih"] = float(np.median(diff))
    out["rerata_selisih"] = float(np.mean(diff))
    if np.allclose(diff, 0):
        out.update(cukup=True, uji="(semua selisih = 0)", statistik=None, p=1.0, signifikan=False)
        return out
    with _w.catch_warnings():
        _w.simplefilter("ignore")
        try:
            _, p_norm = stats.shapiro(diff)
        except Exception:
            p_norm = 0.0
        try:
            if p_norm > 0.05:
                stat, p = stats.ttest_rel(a, b)
                uji = "Paired t-test"
            else:
                stat, p = stats.wilcoxon(a, b)
                uji = "Wilcoxon signed-rank"
        except Exception as e:
            out.update(cukup=False, pesan="uji gagal: " + str(e))
            return out
    out.update(cukup=True, uji=uji, statistik=float(stat), p=float(p),
               signifikan=bool(p < 0.05), p_normalitas=float(p_norm))
    return out


def _uji_mcnemar(rs_ok, lit_ok):
    """McNemar: apakah proporsi 'Sesuai' berbeda antara penilaian vs Pedoman RS
    dan vs Literatur (berpasangan pada pasien yang sama)."""
    from scipy import stats
    rs_ok = np.asarray(rs_ok, dtype=bool)
    lit_ok = np.asarray(lit_ok, dtype=bool)
    n = len(rs_ok)
    b = int(np.sum(rs_ok & ~lit_ok))
    c = int(np.sum(~rs_ok & lit_ok))
    out = {"n": n, "b_RS_only": b, "c_Lit_only": c}
    nd = b + c
    if nd == 0:
        out.update(cukup=True, p=1.0, signifikan=False, metode="tidak ada pasangan diskordan")
        return out
    try:
        if nd < 25:
            p = min(1.0, 2 * stats.binom.cdf(min(b, c), nd, 0.5))
            metode = "McNemar eksak (binomial)"
        else:
            chi2 = (abs(b - c) - 1) ** 2 / nd
            p = float(stats.chi2.sf(chi2, 1))
            metode = "McNemar (koreksi kontinuitas)"
    except Exception as e:
        out.update(cukup=False, pesan="uji gagal: " + str(e))
        return out
    out.update(cukup=True, p=float(p), signifikan=bool(p < 0.05), metode=metode)
    return out


def _blok_uji_per_antibiotik(df, kol_akt, kol_man, kol_rs_mid, kol_lit_mid):
    """Uji beda berpasangan (aktual & manual vs titik tengah rekomendasi) per
    antibiotik, untuk kolom numerik tertentu (dosis mg/kg ATAU frekuensi dosis/hari).
    Menghasilkan blok berisi: aktual_vs_RS, aktual_vs_Literatur, manual_vs_RS,
    manual_vs_Literatur (yang datanya memadai, n>=3)."""
    per_ab = {}
    for ab in ["Gentamisin", "Amikasin", "Vankomisin"]:
        sub = df[df["Antibiotik"] == ab]
        blok = {}
        # AKTUAL
        sa = sub[sub[kol_akt].notna()]
        if len(sa) >= 3:
            s_rs = sa[sa[kol_rs_mid].notna()]
            if len(s_rs) >= 3:
                blok["aktual_vs_RS"] = _uji_beda_berpasangan(s_rs[kol_akt].values, s_rs[kol_rs_mid].values)
            s_lit = sa[sa[kol_lit_mid].notna()]
            if len(s_lit) >= 3:
                blok["aktual_vs_Literatur"] = _uji_beda_berpasangan(s_lit[kol_akt].values, s_lit[kol_lit_mid].values)
        # MANUAL
        sm = sub[sub[kol_man].notna()]
        if len(sm) >= 3:
            s_rs = sm[sm[kol_rs_mid].notna()]
            if len(s_rs) >= 3:
                blok["manual_vs_RS"] = _uji_beda_berpasangan(s_rs[kol_man].values, s_rs[kol_rs_mid].values)
            s_lit = sm[sm[kol_lit_mid].notna()]
            if len(s_lit) >= 3:
                blok["manual_vs_Literatur"] = _uji_beda_berpasangan(s_lit[kol_man].values, s_lit[kol_lit_mid].values)
        if not blok:
            blok["pesan"] = "n<3, uji beda tidak dilakukan."
        per_ab[ab] = blok
    return per_ab


def _mcnemar_dari_kolom(df, kol_rs, kol_lit):
    valid = df[df[kol_rs].astype(str).ne("") & df[kol_lit].astype(str).ne("")]
    if len(valid) >= 1:
        rs_ok = valid[kol_rs].eq("Sesuai").values
        lit_ok = valid[kol_lit].eq("Sesuai").values
        return _uji_mcnemar(rs_ok, lit_ok)
    return None


def hitung_statistik(records):
    """Menjalankan analisis statistik untuk DOSIS dan FREKUENSI.
    Struktur keluaran:
      { 'dosis': {'per_antibiotik':..., 'mcnemar':...},
        'frekuensi': {'per_antibiotik':..., 'mcnemar':...},
        'catatan': [...] }
    - Uji beda: aktual (mg/kg untuk dosis, dosis/hari untuk frekuensi) vs titik
      tengah rentang rekomendasi RS & Literatur, per antibiotik.
    - McNemar: proporsi 'Sesuai' vs Pedoman RS dibanding vs Literatur (berpasangan).
    """
    hasil = {"dosis": {"per_antibiotik": {}, "mcnemar": None},
             "frekuensi": {"per_antibiotik": {}, "mcnemar": None},
             "catatan": []}
    df = pd.DataFrame(records)
    if df.empty:
        hasil["catatan"].append("Tidak ada data valid untuk dianalisis.")
        return hasil

    # DOSIS
    hasil["dosis"]["per_antibiotik"] = _blok_uji_per_antibiotik(df, "_da_perkg", "_dm_mid_perkg", "_rs_mid_perkg", "_lit_mid_perkg")
    hasil["dosis"]["mcnemar_aktual"] = _mcnemar_dari_kolom(df, "_ka_rs", "_ka_lit")
    hasil["dosis"]["mcnemar_manual"] = _mcnemar_dari_kolom(df, "_kma_rs", "_kma_lit")
    # FREKUENSI (skala dosis/hari)
    hasil["frekuensi"]["per_antibiotik"] = _blok_uji_per_antibiotik(df, "_fa_perhari", "_fm_mid_perhari", "_rs_freq_mid", "_lit_freq_mid")
    hasil["frekuensi"]["mcnemar_aktual"] = _mcnemar_dari_kolom(df, "_kfa_rs", "_kfa_lit")
    hasil["frekuensi"]["mcnemar_manual"] = _mcnemar_dari_kolom(df, "_kmf_rs", "_kmf_lit")
    return hasil


def ringkasan_kesesuaian(records):
    """Distribusi kategori (Sesuai/Berlebih/Kurang) untuk DOSIS & FREKUENSI, pada
    perhitungan AKTUAL maupun MANUAL, terhadap Pedoman RS & Literatur, keseluruhan
    & per antibiotik.
    Setiap baris: [Jenis, Perhitungan, Kelompok, Pembanding, Total, Sesuai n, %, Berlebih n, %, Kurang n, %]."""
    df = pd.DataFrame(records)
    baris = []

    # (Jenis, Perhitungan, kolom_kategori, Pembanding)
    pasangan = [
        ("Dosis", "Aktual", "_ka_rs", "vs Pedoman RS"),
        ("Dosis", "Aktual", "_ka_lit", "vs Literatur"),
        ("Dosis", "Manual", "_kma_rs", "vs Pedoman RS"),
        ("Dosis", "Manual", "_kma_lit", "vs Literatur"),
        ("Frekuensi", "Aktual", "_kfa_rs", "vs Pedoman RS"),
        ("Frekuensi", "Aktual", "_kfa_lit", "vs Literatur"),
        ("Frekuensi", "Manual", "_kmf_rs", "vs Pedoman RS"),
        ("Frekuensi", "Manual", "_kmf_lit", "vs Literatur"),
    ]

    def hitung(subdf, label):
        for jenis, perhitungan, kolom, ref in pasangan:
            if kolom not in subdf.columns:
                continue
            vals = subdf[kolom].astype(str)
            total = int((vals != "").sum())
            s = int((vals == "Sesuai").sum())
            b = int((vals == "Berlebih").sum())
            k = int((vals == "Kurang").sum())
            def pct(x):
                return "{0:.0f}%".format(100 * x / total) if total else "-"
            baris.append([jenis, perhitungan, label, ref, total, s, pct(s), b, pct(b), k, pct(k)])

    if not df.empty:
        hitung(df, "SEMUA")
        for ab in ["Gentamisin", "Amikasin", "Vankomisin"]:
            sub = df[df["Antibiotik"] == ab]
            if not sub.empty:
                hitung(sub, ab)
    return baris


# ---------------------------------------------------------------------------
# INTERPRETASI & KESIMPULAN OTOMATIS
# ---------------------------------------------------------------------------
def _distribusi_kol(df, kol):
    if kol not in df.columns:
        return 0, 0, 0, 0
    vals = df[kol].astype(str)
    total = int((vals != "").sum())
    s = int((vals == "Sesuai").sum())
    b = int((vals == "Berlebih").sum())
    k = int((vals == "Kurang").sum())
    return total, s, b, k


def _pct(x, t):
    return (100.0 * x / t) if t else 0.0


def buat_interpretasi(records, stats):
    """Menghasilkan interpretasi naratif + kesimpulan dari hasil evaluasi & uji
    statistik. Mengembalikan daftar section: {judul, isi:[(tipe, teks)]} dengan
    tipe 'penting' (disorot), 'poin' (butir), 'sub' (subjudul), 'normal'."""
    df = pd.DataFrame(records)
    n = len(df)
    seksi = []
    if df.empty:
        return [{"judul": "Interpretasi", "isi": [("normal", "Tidak ada data valid untuk diinterpretasikan.")]}]

    def arah_kata(b, k):
        if b == 0 and k == 0:
            return "seluruhnya dalam rentang rekomendasi"
        if b > k:
            return "cenderung BERLEBIH (dosis/frekuensi lebih dari rekomendasi)"
        if k > b:
            return "cenderung KURANG (dosis/frekuensi kurang dari rekomendasi)"
        return "tersebar seimbang antara berlebih dan kurang"

    # ---------- SECTION 1: Ringkasan kesesuaian pemberian aktual ----------
    isi = [("normal", "Total " + str(n) + " pasien dianalisis. Ringkasan berikut menilai kesesuaian pemberian NYATA (aktual) terhadap perhitungan aplikasi.")]
    for jenis, kol, acuan in [("Dosis", "_ka_rs", "Pedoman RS"), ("Dosis", "_ka_lit", "Literatur"),
                              ("Frekuensi", "_kfa_rs", "Pedoman RS"), ("Frekuensi", "_kfa_lit", "Literatur")]:
        total, s, b, k = _distribusi_kol(df, kol)
        if total == 0:
            continue
        ps = _pct(s, total)
        teks = (jenis + " aktual vs " + acuan + ": " + "{0:.0f}%".format(ps) + " Sesuai "
                + "(" + str(s) + "/" + str(total) + "), " + "{0:.0f}% Berlebih, ".format(_pct(b, total))
                + "{0:.0f}% Kurang".format(_pct(k, total)) + "; " + arah_kata(b, k) + ".")
        seksi_tipe = "penting" if ps < 50 else "poin"
        isi.append((seksi_tipe, teks))
    # antibiotik dengan kesesuaian dosis terendah (vs RS)
    worst = None
    for ab in ["Gentamisin", "Amikasin", "Vankomisin"]:
        sub = df[df["Antibiotik"] == ab]
        total, s, b, k = _distribusi_kol(sub, "_ka_rs")
        if total:
            ps = _pct(s, total)
            if worst is None or ps < worst[1]:
                worst = (ab, ps, total)
    if worst:
        isi.append(("penting", "Kesesuaian dosis aktual (vs RS) terendah pada " + worst[0] + " (" + "{0:.0f}%".format(worst[1]) + " sesuai, n=" + str(worst[2]) + ") -- perlu perhatian khusus pada peresepan antibiotik ini."))
    seksi.append({"judul": "1. Ringkasan Kesesuaian Pemberian Aktual", "isi": isi})

    # ---------- SECTION 2: Interpretasi uji beda aktual vs rekomendasi ----------
    def _desc(u, subjek, acuan, satuan):
        if not u or not u.get("cukup"):
            return None
        md = u.get("median_selisih", 0)
        arah = "lebih tinggi" if md > 0 else ("lebih rendah" if md < 0 else "setara")
        if u.get("signifikan"):
            return ("penting", subjek + " BERBEDA SIGNIFIKAN dari " + acuan + " (" + u.get("uji", "-") + ", p=" + "{0:.4f}".format(u["p"]) + "); median selisih " + "{0:+.3f}".format(md) + " " + satuan + " (" + arah + ").")
        return ("poin", subjek + " tidak berbeda signifikan dari " + acuan + " (p=" + "{0:.4f}".format(u["p"]) + "); median selisih " + "{0:+.3f}".format(md) + " " + satuan + ".")

    isi = [("normal", "Uji beda menilai apakah nilai AKTUAL menyimpang bermakna dari titik tengah rekomendasi aplikasi. Uji dipilih otomatis (paired t-test bila selisih normal; Wilcoxon bila tidak).")]
    ada_signif = False
    for satuan, kunci_blok, jenis in [("mg/kg", "dosis", "Dosis"), ("dosis/hari", "frekuensi", "Frekuensi")]:
        for ab, blok in stats[kunci_blok]["per_antibiotik"].items():
            for key, acuan in [("aktual_vs_RS", "rekomendasi RS"), ("aktual_vs_Literatur", "rekomendasi Literatur")]:
                d = _desc(blok.get(key), jenis + " aktual " + ab + " vs " + acuan, acuan, satuan)
                if d:
                    if d[0] == "penting":
                        ada_signif = True
                    isi.append(d)
    if not ada_signif:
        isi.append(("poin", "Secara umum tidak ditemukan perbedaan signifikan yang menonjol antara pemberian aktual dan rekomendasi (dengan ukuran sampel saat ini)."))
    seksi.append({"judul": "2. Interpretasi Uji Beda: Pemberian Aktual vs Rekomendasi Aplikasi", "isi": isi})

    # ---------- SECTION 3: RELIABILITAS APLIKASI vs HITUNG MANUAL ----------
    isi = [("normal", "Bagian ini menilai seberapa konsisten (andal) perhitungan APLIKASI dibandingkan HITUNG MANUAL tenaga kesehatan. Aplikasi memakai berat badan eksak (tanpa pembulatan) dan dapat menyertakan/tidak menyertakan usia gestasi & pascalahir; hitung manual umumnya membulatkan berat badan.")]
    for satuan, kunci_blok, jenis in [("mg/kg", "dosis", "Dosis"), ("dosis/hari", "frekuensi", "Frekuensi")]:
        n_uji = 0
        n_signif = 0
        mds = []
        detail = []
        for ab, blok in stats[kunci_blok]["per_antibiotik"].items():
            u = blok.get("manual_vs_RS")
            if u and u.get("cukup"):
                n_uji += 1
                mds.append(u.get("median_selisih", 0))
                if u.get("signifikan"):
                    n_signif += 1
                    detail.append(ab + " (p=" + "{0:.4f}".format(u["p"]) + ", selisih " + "{0:+.3f}".format(u.get("median_selisih", 0)) + " " + satuan + ")")
        if n_uji == 0:
            isi.append(("poin", jenis + ": data hitung manual tidak memadai untuk uji reliabilitas (perlu >=3 pasien per antibiotik)."))
            continue
        if n_signif == 0:
            rentang_md = "{0:+.3f} s/d {1:+.3f}".format(min(mds), max(mds)) if mds else "-"
            isi.append(("penting", jenis + ": hitung manual TIDAK berbeda signifikan dari perhitungan aplikasi (RS) pada seluruh " + str(n_uji) + " antibiotik yang diuji (median selisih " + rentang_md + " " + satuan + ") -> aplikasi ANDAL & konsisten dengan hitung manual berbasis pedoman."))
        else:
            isi.append(("penting", jenis + ": terdapat perbedaan signifikan antara hitung manual dan aplikasi pada " + str(n_signif) + " dari " + str(n_uji) + " antibiotik: " + "; ".join(detail) + " -> tinjau penyebab (mis. pembulatan berat/usia atau versi pedoman)."))
    # concordance kategorikal manual vs RS (dosis)
    total, s, b, k = _distribusi_kol(df, "_kma_rs")
    if total:
        ps = _pct(s, total)
        if ps >= 80:
            isi.append(("poin", "Secara kategorikal, " + "{0:.0f}%".format(ps) + " hasil hitung manual berada dalam rentang dosis rekomendasi aplikasi (RS) -> kesepakatan tinggi."))
        else:
            isi.append(("penting", "Secara kategorikal, hanya " + "{0:.0f}%".format(ps) + " hasil hitung manual yang PERSIS dalam rentang dosis aplikasi (RS). Karena aplikasi TIDAK membulatkan & TIDAK memberi toleransi, selisih pembulatan kecil pada hitung manual dapat terkategori Berlebih/Kurang meski secara statistik selisihnya tidak bermakna. Ini justru menunjukkan aplikasi lebih presisi terhadap berat badan aktual."))
    seksi.append({"judul": "3. Reliabilitas Aplikasi vs Hitung Manual", "isi": isi})

    # ---------- SECTION 4: Kesimpulan & tindak lanjut ----------
    isi = []
    total_d, s_d, b_d, k_d = _distribusi_kol(df, "_ka_rs")
    if total_d:
        isi.append(("poin", "Kesesuaian dosis aktual terhadap pedoman RS: " + "{0:.0f}%".format(_pct(s_d, total_d)) + " sesuai. " + ("Perlu audit/edukasi peresepan." if _pct(s_d, total_d) < 70 else "Tergolong baik.")))
    isi.append(("poin", "Aplikasi dapat dijadikan alat bantu verifikasi dosis: konsisten dengan hitung manual (uji beda) namun lebih presisi karena memakai berat badan eksak dan mempertimbangkan (opsional) usia gestasi/pascalahir."))
    isi.append(("penting", "TINDAK LANJUT: fokuskan telaah pada pemberian berkategori Berlebih (risiko toksisitas) dan Kurang (risiko subterapeutik), terutama antibiotik indeks terapi sempit ini, dan konfirmasi dengan kadar obat (TDM) bila tersedia."))
    seksi.append({"judul": "4. Kesimpulan & Tindak Lanjut", "isi": isi})

    # ---------- SECTION 5: Keterbatasan ----------
    isi = [
        ("poin", "Nilai p menunjukkan kemaknaan statistik, BUKAN kemaknaan klinis; interpretasi akhir tetap memerlukan penilaian profesional."),
        ("poin", "Uji beda per antibiotik memerlukan minimal 3 pasien; kelompok dengan n<3 tidak diuji."),
        ("poin", "Titik tengah rentang dipakai sebagai acuan numerik uji beda. Kesesuaian frekuensi terhadap Pedoman RS memakai rentang interval pedoman (pedoman RS tidak membedakan usia)."),
        ("poin", "Alat ini adalah clinical decision support, BUKAN pengganti penilaian apoteker/DPJP. Rekomendasi berbasis pedoman yang diunggah & literatur yang diverifikasi; selalu cek versi terbaru."),
    ]
    seksi.append({"judul": "5. Catatan Metodologis & Keterbatasan", "isi": isi})
    return seksi


# ---------------------------------------------------------------------------
# OUTPUT EXCEL
# ---------------------------------------------------------------------------
KOLOM_HASIL = [
    "Kode Pasien", "Antibiotik", "Usia Gestasi (mgg)", "Usia Pascalahir (hari)", "PMA (mgg)",
    "Berat (kg)", "Fungsi Ginjal",
    "Dosis RS (mg)", "Frek RS",
    "Dosis Literatur (mg)", "Frek Literatur",
    "Dosis Manual (mg)", "Frek Manual", "Dosis Aktual (mg)", "Dosis Aktual (mg/kg/dosis)", "Frek Aktual",
    "Aktual vs RS (dosis)", "Aktual vs RS (frek)",
    "Aktual vs Literatur (dosis)", "Aktual vs Literatur (frek)",
    "Manual vs RS (dosis)", "Manual vs RS (frek)",
    "Manual vs Literatur (dosis)", "Manual vs Literatur (frek)",
]
KOLOM_KATEGORI = ["Aktual vs RS (dosis)", "Aktual vs RS (frek)",
                  "Aktual vs Literatur (dosis)", "Aktual vs Literatur (frek)",
                  "Manual vs RS (dosis)", "Manual vs RS (frek)",
                  "Manual vs Literatur (dosis)", "Manual vs Literatur (frek)"]

_FILL_KATEGORI = {"Sesuai": "C6EFCE", "Berlebih": "FFC7CE", "Kurang": "FFEB9C",
                  "Lebih pendek": "FFC7CE", "Lebih panjang": "FFEB9C"}


def _warna_kategori_teks(teks):
    t = str(teks)
    for kunci, warna in _FILL_KATEGORI.items():
        if t.startswith(kunci):
            return warna
    return None


def buat_output_excel(records, stats, ringkasan, warnings, nama_file=None):
    import openpyxl
    from openpyxl.styles import Font, PatternFill, Border, Side, Alignment
    from openpyxl.utils import get_column_letter
    styles = (Font, PatternFill, Border, Side, Alignment)
    thin = Side(style="thin", color="CCCCCC")

    wb = openpyxl.Workbook()

    # ---- Sheet 1: Hasil per Pasien ----
    ws = wb.active
    ws.title = "Hasil per Pasien"
    for j, nama in enumerate(KOLOM_HASIL, start=1):
        ws.cell(row=1, column=j, value=nama)
    _style_header(ws, 1, len(KOLOM_HASIL), styles)
    for i, rec in enumerate(records, start=2):
        for j, nama in enumerate(KOLOM_HASIL, start=1):
            val = rec.get(nama, "")
            cell = ws.cell(row=i, column=j, value=val)
            cell.font = Font(name="Arial", size=9)
            cell.border = Border(left=thin, right=thin, top=thin, bottom=thin)
            cell.alignment = Alignment(vertical="center", wrap_text=True)
            if nama in KOLOM_KATEGORI:
                w = _warna_kategori_teks(val)
                if w:
                    cell.fill = PatternFill("solid", fgColor=w)
    ws.freeze_panes = "A2"
    if records:
        ws.auto_filter.ref = "A1:" + get_column_letter(len(KOLOM_HASIL)) + str(len(records) + 1)
    for j, nama in enumerate(KOLOM_HASIL, start=1):
        ws.column_dimensions[get_column_letter(j)].width = max(12, min(24, len(nama) + 2))

    # ---- Sheet 2: Ringkasan (Dosis & Frekuensi; Aktual & Manual) ----
    ws2 = wb.create_sheet("Ringkasan Kesesuaian")
    ws2["A1"] = "Ringkasan Kesesuaian Dosis & Frekuensi (Aktual & Manual)"
    ws2["A1"].font = Font(name="Arial", bold=True, size=12)
    hdr = ["Jenis", "Perhitungan", "Kelompok", "Pembanding", "Total", "Sesuai (n)", "Sesuai (%)", "Berlebih (n)", "Berlebih (%)", "Kurang (n)", "Kurang (%)"]
    for j, h in enumerate(hdr, start=1):
        ws2.cell(row=3, column=j, value=h)
    _style_header(ws2, 3, len(hdr), styles)
    for i, brs in enumerate(ringkasan, start=4):
        for j, v in enumerate(brs, start=1):
            cell = ws2.cell(row=i, column=j, value=v)
            cell.font = Font(name="Arial", size=9)
            cell.border = Border(left=thin, right=thin, top=thin, bottom=thin)
            if j == 1 and str(v) == "Frekuensi":
                cell.fill = PatternFill("solid", fgColor="EDF2FB")
            if j == 2 and str(v) == "Manual":
                cell.fill = PatternFill("solid", fgColor="FBF3E7")
    ws2.freeze_panes = "A4"
    lebar = [11, 12, 12, 26, 8, 10, 10, 11, 11, 10, 10]
    for j, w in enumerate(lebar, start=1):
        ws2.column_dimensions[get_column_letter(j)].width = w

    # ---- Sheet 3: Analisis Statistik (Dosis & Frekuensi) ----
    ws3 = wb.create_sheet("Analisis Statistik")
    r = [1]  # pakai list agar bisa dimutasi di dalam fungsi bersarang

    def _tulis_blok(judul_beda, satuan_selisih, blok_stats):
        ws3.cell(row=r[0], column=1, value=judul_beda).font = Font(name="Arial", bold=True, size=10); r[0] += 1
        hdr3 = ["Antibiotik", "Perbandingan", "n", "Uji", "Statistik", "p-value", satuan_selisih, "Kesimpulan"]
        for j, h in enumerate(hdr3, start=1):
            ws3.cell(row=r[0], column=j, value=h)
        _style_header(ws3, r[0], len(hdr3), styles); r[0] += 1
        urutan = [("aktual_vs_RS", "Aktual vs Pedoman RS"), ("aktual_vs_Literatur", "Aktual vs Literatur"),
                  ("manual_vs_RS", "Manual vs Pedoman RS"), ("manual_vs_Literatur", "Manual vs Literatur")]
        for ab, blok in blok_stats["per_antibiotik"].items():
            ada = any(k in blok for k, _ in urutan)
            if not ada:
                ws3.cell(row=r[0], column=1, value=ab).font = Font(name="Arial", size=9)
                ws3.cell(row=r[0], column=2, value=blok.get("pesan", "-")).font = Font(name="Arial", size=9, italic=True)
                r[0] += 1
                continue
            for key, lbl in urutan:
                u = blok.get(key)
                if not u:
                    continue
                if not u.get("cukup"):
                    vals = [ab, lbl, u.get("n", "-"), u.get("pesan", "-"), "-", "-", "-", u.get("pesan", "-")]
                else:
                    p = u.get("p")
                    kesimp = ("Berbeda signifikan (p<0.05)" if u.get("signifikan") else "Tidak berbeda signifikan (p>=0.05)")
                    vals = [ab, lbl, u["n"], u.get("uji", "-"),
                            round(u["statistik"], 3) if u.get("statistik") is not None else "-",
                            round(p, 4) if p is not None else "-",
                            round(u.get("median_selisih", 0), 3), kesimp]
                for j, v in enumerate(vals, start=1):
                    cell = ws3.cell(row=r[0], column=j, value=v)
                    cell.font = Font(name="Arial", size=9)
                    cell.border = Border(left=thin, right=thin, top=thin, bottom=thin)
                    if j == 2 and str(v).startswith("Manual"):
                        cell.fill = PatternFill("solid", fgColor="FBF3E7")
                r[0] += 1
        r[0] += 1

    def _tulis_mcnemar(judul, mc):
        ws3.cell(row=r[0], column=1, value=judul).font = Font(name="Arial", bold=True, size=10); r[0] += 1
        if mc and mc.get("cukup"):
            pasang = [("Metode", mc.get("metode", "-")), ("n pasien", mc.get("n")),
                      ("Sesuai hanya oleh RS (b)", mc.get("b_RS_only")),
                      ("Sesuai hanya oleh Literatur (c)", mc.get("c_Lit_only")),
                      ("p-value", round(mc.get("p", 1), 4)),
                      ("Kesimpulan", "Proporsi kesesuaian BERBEDA signifikan antara RS & Literatur (p<0.05)" if mc.get("signifikan") else "Proporsi kesesuaian TIDAK berbeda signifikan (p>=0.05)")]
            for lab, val in pasang:
                ws3.cell(row=r[0], column=1, value=lab).font = Font(name="Arial", size=9)
                ws3.cell(row=r[0], column=2, value=val).font = Font(name="Arial", size=9)
                r[0] += 1
        else:
            ws3.cell(row=r[0], column=1, value=(mc or {}).get("pesan", "Tidak dapat dihitung.")).font = Font(name="Arial", size=9, italic=True); r[0] += 1
        r[0] += 1

    ws3.cell(row=r[0], column=1, value="Analisis Statistik Otomatis").font = Font(name="Arial", bold=True, size=12); r[0] += 2
    _tulis_blok("A. Uji beda DOSIS (mg/kg) vs titik tengah rentang rekomendasi -- Aktual & Manual, berpasangan", "Median selisih (mg/kg)", stats["dosis"])
    _tulis_mcnemar("B1. McNemar kesesuaian DOSIS AKTUAL: proporsi 'Sesuai' vs Pedoman RS dibanding vs Literatur", stats["dosis"]["mcnemar_aktual"])
    _tulis_mcnemar("B2. McNemar kesesuaian DOSIS MANUAL: proporsi 'Sesuai' vs Pedoman RS dibanding vs Literatur", stats["dosis"]["mcnemar_manual"])
    _tulis_blok("C. Uji beda FREKUENSI (dosis/hari = 24/interval) vs titik tengah rekomendasi -- Aktual & Manual, berpasangan", "Median selisih (dosis/hari)", stats["frekuensi"])
    _tulis_mcnemar("D1. McNemar kesesuaian FREKUENSI AKTUAL: proporsi 'Sesuai' vs Pedoman RS dibanding vs Literatur", stats["frekuensi"]["mcnemar_aktual"])
    _tulis_mcnemar("D2. McNemar kesesuaian FREKUENSI MANUAL: proporsi 'Sesuai' vs Pedoman RS dibanding vs Literatur", stats["frekuensi"]["mcnemar_manual"])

    ws3.cell(row=r[0], column=1, value="Catatan: 'Aktual/Manual vs Pedoman RS/Literatur' = uji beda nilai aktual/manual terhadap titik tengah rentang rekomendasi aplikasi (dosis: mg/kg; frekuensi: dosis/hari) -- memperlihatkan seberapa jauh pemberian nyata & hitung manual menyimpang dari perhitungan aplikasi. Uji otomatis memilih paired t-test bila selisih normal (Shapiro-Wilk p>0.05), selain itu Wilcoxon signed-rank. Kesesuaian frekuensi terhadap Pedoman RS memakai rentang interval pedoman (pedoman RS tidak membedakan usia). Interpretasi tetap memerlukan penilaian klinis & ukuran sampel memadai.").font = Font(name="Arial", size=8, italic=True, color="666666")
    ws3.column_dimensions["A"].width = 34
    for col in "BCDEFGH":
        ws3.column_dimensions[col].width = 20
    ws3.column_dimensions["H"].width = 36

    # ---- Sheet 4: Peringatan (bila ada) ----
    # ---- Sheet: Interpretasi & Kesimpulan ----
    seksi_interp = buat_interpretasi(records, stats)
    wsi = wb.create_sheet("Interpretasi & Kesimpulan")
    wsi.sheet_view.showGridLines = False
    wsi["A1"] = "INTERPRETASI HASIL & KESIMPULAN"
    wsi["A1"].font = Font(name="Arial", bold=True, size=14, color="1F4E78")
    wsi["A2"] = ("Halaman ini membantu tenaga kesehatan menelaah pengobatan yang telah diberikan secara seksama & dapat dipertanggungjawabkan. "
                 "Poin yang disorot (kuning) adalah temuan penting.")
    wsi["A2"].font = Font(name="Arial", size=9, italic=True, color="666666")
    wsi["A2"].alignment = Alignment(wrap_text=True, vertical="top")
    wsi.merge_cells("A2:A3")
    rr = 5
    fill_penting = PatternFill("solid", fgColor="FFF3CD")
    fill_judul = PatternFill("solid", fgColor="1F4E78")
    for sec in seksi_interp:
        c = wsi.cell(row=rr, column=1, value=sec["judul"])
        c.font = Font(name="Arial", bold=True, size=11, color="FFFFFF")
        c.fill = fill_judul
        c.alignment = Alignment(vertical="center")
        wsi.row_dimensions[rr].height = 20
        rr += 1
        for tipe, teks in sec["isi"]:
            if tipe == "penting":
                cell = wsi.cell(row=rr, column=1, value="\u2605 " + teks)  # bintang
                cell.font = Font(name="Arial", size=10, bold=True, color="7A5C00")
                cell.fill = fill_penting
            elif tipe == "poin":
                cell = wsi.cell(row=rr, column=1, value="\u2022 " + teks)
                cell.font = Font(name="Arial", size=10)
            elif tipe == "sub":
                cell = wsi.cell(row=rr, column=1, value=teks)
                cell.font = Font(name="Arial", size=10, bold=True)
            else:
                cell = wsi.cell(row=rr, column=1, value=teks)
                cell.font = Font(name="Arial", size=10)
            cell.alignment = Alignment(wrap_text=True, vertical="top")
            # perkiraan tinggi baris berdasarkan panjang teks (~90 char/baris pada lebar 150)
            baris_teks = max(1, (len(teks) // 90) + 1)
            wsi.row_dimensions[rr].height = 15 * baris_teks
            rr += 1
        rr += 1
    wsi.column_dimensions["A"].width = 150
    # pindahkan sheet interpretasi ke posisi ke-2 (setelah Hasil per Pasien)
    wb.move_sheet(wsi, offset=-(len(wb.sheetnames) - 2))

    if warnings:
        ws4 = wb.create_sheet("Peringatan")
        ws4["A1"] = "Baris yang dilewati / peringatan"
        ws4["A1"].font = Font(name="Arial", bold=True, size=11)
        for i, w in enumerate(warnings, start=3):
            ws4.cell(row=i, column=1, value=w).font = Font(name="Arial", size=9)
        ws4.column_dimensions["A"].width = 110

    # Sheet keterangan singkatan & istilah (revisi 2026.11.01)
    wsk = wb.create_sheet("Keterangan Singkatan")
    wsk["A1"] = "Keterangan singkatan dan istilah"
    wsk["A1"].font = Font(name="Arial", bold=True, size=11)
    for i, (sk, panjang, ket) in enumerate(SINGKATAN, start=3):
        wsk.cell(row=i, column=1, value=sk).font = Font(name="Arial", bold=True, size=9)
        wsk.cell(row=i, column=2, value=panjang + ((" (" + ket + ")") if ket else "")).font = Font(name="Arial", size=9)
    _r = len(SINGKATAN) + 4
    for teks in ["Frekuensi ditulis 'tiap X jam' (interval antardosis).",
                 "Berlebih = di atas batas maksimal rujukan; Kurang = di bawah batas minimal rujukan. Selisih ditulis dalam satuan (mg, mg/kg, atau dosis/hari) dan %.",
                 "Frekuensi Berlebih = diberikan lebih sering (interval lebih pendek); frekuensi Kurang = lebih jarang (interval lebih panjang)."]:
        wsk.cell(row=_r, column=1, value=teks).font = Font(name="Arial", size=9, italic=True)
        _r += 1
    wsk.column_dimensions["A"].width = 14
    wsk.column_dimensions["B"].width = 110

    if nama_file is None:
        stamp = datetime.datetime.now().strftime("%Y%m%d_%H%M%S")
        nama_file = "hasil_evaluasi_batch_" + stamp + ".xlsx"
    wb.save(nama_file)
    return nama_file


# ---------------------------------------------------------------------------
# OUTPUT PDF (ringkasan eksekutif)
# ---------------------------------------------------------------------------
def _batch_pdf_html(records, stats, ringkasan, warnings):
    now = datetime.datetime.now().strftime("%d-%m-%Y %H:%M")
    n = len(records)
    css = ("@page { size: A4 landscape; margin: 1.2cm; }"
           "body { font-family: Helvetica, Arial, sans-serif; font-size: 9px; color:#222; }"
           "h2 { font-size:15px; margin:0 0 2px 0;} h3{font-size:11px;margin:12px 0 4px 0;}"
           "table{border-collapse:collapse;width:100%;margin:6px 0;} th,td{border:1px solid #999;padding:3px 5px;text-align:left;} th{background-color:#e9ecef;}"
           ".kecil{color:#666;font-size:8px;}")
    # Ringkasan table (dengan kolom Jenis)
    rows_r = ""
    for brs in ringkasan:
        rows_r += "<tr>" + "".join("<td>" + str(x) + "</td>" for x in brs) + "</tr>"
    tabel_ring = ("<table><tr><th>Jenis</th><th>Perhitungan</th><th>Kelompok</th><th>Pembanding</th><th>Total</th><th>Sesuai (n)</th><th>Sesuai (%)</th>"
                  "<th>Berlebih (n)</th><th>Berlebih (%)</th><th>Kurang (n)</th><th>Kurang (%)</th></tr>" + rows_r + "</table>")

    def _tabel_uji(blok_stats):
        rows_s = ""
        urutan = [("aktual_vs_RS", "Aktual vs RS"), ("aktual_vs_Literatur", "Aktual vs Literatur"),
                  ("manual_vs_RS", "Manual vs RS"), ("manual_vs_Literatur", "Manual vs Literatur")]
        for ab, blok in blok_stats["per_antibiotik"].items():
            for key, lbl in urutan:
                u = blok.get(key)
                if not u:
                    continue
                if not u.get("cukup"):
                    rows_s += "<tr><td>" + ab + "</td><td>" + lbl + "</td><td colspan='4'>" + str(u.get("pesan", "-")) + "</td></tr>"
                else:
                    kes = "Berbeda signifikan (p&lt;0.05)" if u.get("signifikan") else "Tidak berbeda signifikan"
                    rows_s += ("<tr><td>" + ab + "</td><td>" + lbl + "</td><td>" + str(u["n"]) + "</td><td>" + str(u.get("uji", "-")) +
                               "</td><td>" + ("{0:.4f}".format(u["p"]) if u.get("p") is not None else "-") + "</td><td>" + kes + "</td></tr>")
        if not rows_s:
            rows_s = "<tr><td colspan='6'>Tidak ada uji beda yang dapat dijalankan (n per antibiotik &lt; 3).</td></tr>"
        return ("<table><tr><th>Antibiotik</th><th>Perbandingan</th><th>n</th><th>Uji</th><th>p-value</th><th>Kesimpulan</th></tr>" + rows_s + "</table>")

    def _mc_txt(mc):
        if mc and mc.get("cukup"):
            return (mc.get("metode", "-") + "; p=" + "{0:.4f}".format(mc.get("p", 1)) + " -> "
                    + ("proporsi kesesuaian BERBEDA signifikan antara RS & Literatur." if mc.get("signifikan") else "proporsi kesesuaian tidak berbeda signifikan."))
        return (mc or {}).get("pesan", "tidak dapat dihitung.")

    tabel_stat_dosis = _tabel_uji(stats["dosis"])
    tabel_stat_freq = _tabel_uji(stats["frekuensi"])
    mc_dosis_a = _mc_txt(stats["dosis"]["mcnemar_aktual"])
    mc_dosis_m = _mc_txt(stats["dosis"]["mcnemar_manual"])
    mc_freq_a = _mc_txt(stats["frekuensi"]["mcnemar_aktual"])
    mc_freq_m = _mc_txt(stats["frekuensi"]["mcnemar_manual"])

    peringatan_html = ""
    if warnings:
        peringatan_html = "<h3>Peringatan</h3><ul>" + "".join("<li class='kecil'>" + w + "</li>" for w in warnings) + "</ul>"

    # Interpretasi & kesimpulan
    seksi_interp = buat_interpretasi(records, stats)
    interp_html = "<h2 style='margin-top:14px;'>Interpretasi Hasil &amp; Kesimpulan</h2>"
    interp_html += "<p class='kecil'>Membantu telaah pengobatan yang telah diberikan secara seksama &amp; dapat dipertanggungjawabkan. Poin disorot = temuan penting.</p>"
    for sec in seksi_interp:
        interp_html += "<h3>" + sec["judul"] + "</h3>"
        butir = ""
        for tipe, teks in sec["isi"]:
            t = (teks.replace("&", "&amp;").replace("<", "&lt;").replace(">", "&gt;"))
            if tipe == "penting":
                interp_html += "<p style='background:#fff3cd;border-left:4px solid #e0a800;padding:6px 8px;margin:4px 0;font-weight:bold;'>\u2605 " + t + "</p>"
            elif tipe == "poin":
                butir += "<li>" + t + "</li>"
            elif tipe == "sub":
                interp_html += "<p style='font-weight:bold;margin:4px 0;'>" + t + "</p>"
            else:
                interp_html += "<p class='kecil' style='color:#333;'>" + t + "</p>"
        if butir:
            interp_html += "<ul>" + butir + "</ul>"

    return ("<html><head><meta charset='utf-8'><style>" + css + "</style></head><body>"
            "<h2>Ringkasan Evaluasi Dosis &amp; Frekuensi Antibiotik Neonatus (Batch)</h2>"
            "<div class='kecil'>Dibuat: " + now + " &middot; Jumlah pasien dianalisis: " + str(n) + "</div>"
            "<h3>Distribusi Kesesuaian Dosis &amp; Frekuensi (Aktual &amp; Manual)</h3>" + tabel_ring +
            "<h3>Uji Beda DOSIS (Aktual &amp; Manual mg/kg vs Titik Tengah Rekomendasi)</h3>" + tabel_stat_dosis +
            "<p class='kecil'><b>McNemar dosis aktual:</b> " + mc_dosis_a + "<br><b>McNemar dosis manual:</b> " + mc_dosis_m + "</p>" +
            "<h3>Uji Beda FREKUENSI (Aktual &amp; Manual dosis/hari vs Titik Tengah Rekomendasi)</h3>" + tabel_stat_freq +
            "<p class='kecil'><b>McNemar frekuensi aktual:</b> " + mc_freq_a + "<br><b>McNemar frekuensi manual:</b> " + mc_freq_m + "</p>" +
            peringatan_html +
            interp_html +
            "<p class='kecil'>Alat bantu keputusan klinis - bukan pengganti penilaian apoteker/DPJP. Frekuensi dianalisis pada skala dosis/hari (24/interval); Kesesuaian frekuensi terhadap Pedoman RS memakai rentang interval pedoman (pedoman RS tidak membedakan usia). Interpretasi statistik memerlukan konteks klinis & ukuran sampel memadai.</p>"
            + _legenda_singkatan() + "</body></html>")


def buat_output_pdf(records, stats, ringkasan, warnings, nama_dasar=None):
    if nama_dasar is None:
        nama_dasar = "ringkasan_evaluasi_batch_" + datetime.datetime.now().strftime("%Y%m%d_%H%M%S")
    html = _batch_pdf_html(records, stats, ringkasan, warnings).replace("★", "•")  # font PDF tidak memiliki simbol bintang
    nama_pdf = nama_dasar + ".pdf"
    try:
        from xhtml2pdf import pisa
        with open(nama_pdf, "wb") as f:
            status = pisa.CreatePDF(html, dest=f)
        if not status.err:
            return nama_pdf, "pdf"
    except Exception:
        pass
    nama_html = nama_dasar + ".html"
    with open(nama_html, "w", encoding="utf-8") as f:
        f.write(html)
    return nama_html, "html"

print("Mesin pemrosesan massal siap.")
```

```python
# ---- UI: Template, Upload, Proses, Unduh Hasil ----
import os

def _unduh_berkas(path):
    """Unduh berkas (Colab); fallback ke tautan (Jupyter lain)."""
    try:
        from google.colab import files as _colab_files
        _colab_files.download(path)
        return True
    except Exception:
        try:
            from IPython.display import FileLink
            display(FileLink(path))
            return True
        except Exception:
            print("Berkas tersimpan di: " + path)
            return False


# --- 1. Unduh template ---
btn_template = widgets.Button(description="Unduh Template Excel", button_style="info", icon="file-excel-o")
out_template = widgets.Output()

def _on_template_clicked(b):
    with out_template:
        clear_output()
        try:
            f = buat_template_excel()
            print("Template berhasil dibuat: " + f)
            print("Isi sheet 'Data Pasien' (lihat sheet 'Petunjuk' untuk keterangan tiap kolom), lalu unggah kembali di bawah.")
            _unduh_berkas(f)
        except Exception as e:
            print("Gagal membuat template: " + type(e).__name__ + " - " + str(e))

btn_template.on_click(_on_template_clicked)


# --- 2. Unggah berkas terisi ---
w_upload = widgets.FileUpload(accept=".xlsx,.xls,.csv", multiple=False,
                               description="Unggah berkas terisi")

# --- 3. Proses & unduh hasil ---
w_format_out = widgets.SelectMultiple(
    options=["Excel (.xlsx)", "PDF ringkasan"],
    value=("Excel (.xlsx)", "PDF ringkasan"),
    description="Format hasil:", style={"description_width": "120px"}, rows=2)

btn_proses = widgets.Button(description="Proses Data & Unduh Hasil", button_style="success", icon="cogs")
out_proses = widgets.Output()


def _baca_upload(w):
    """Membaca konten FileUpload lintas versi ipywidgets (v7 dict / v8 tuple)."""
    val = w.value
    if not val:
        return None, None
    if isinstance(val, dict):  # ipywidgets v7
        nama = list(val.keys())[0]
        konten = val[nama]["content"]
    else:  # ipywidgets v8 -> tuple of dicts
        item = val[0]
        nama = item.get("name", "upload.xlsx")
        konten = item.get("content")
        konten = bytes(konten) if not isinstance(konten, bytes) else konten
    return nama, konten


def _on_proses_clicked(b):
    with out_proses:
        clear_output()
        nama, konten = _baca_upload(w_upload)
        if konten is None:
            print("Belum ada berkas yang diunggah. Klik 'Unggah berkas terisi' terlebih dahulu.")
            return
        print("Membaca berkas: " + nama + " ...")
        try:
            buf = io.BytesIO(konten)
            if nama.lower().endswith(".csv"):
                df = pd.read_csv(buf)
            else:
                try:
                    df = pd.read_excel(buf, sheet_name="Data Pasien")
                except Exception:
                    buf.seek(0)
                    df = pd.read_excel(buf)  # sheet pertama
        except Exception as e:
            print("Gagal membaca berkas: " + type(e).__name__ + " - " + str(e))
            return

        try:
            records, warnings = proses_dataframe(df, bangun_rekomendasi, _kategori_dosis, _fmt_num)
        except Exception as e:
            print("Gagal memproses data: " + type(e).__name__ + " - " + str(e))
            return

        if not records:
            print("Tidak ada baris data valid yang dapat diproses.")
            for w in warnings:
                print(" - " + w)
            return

        stats = hitung_statistik(records)
        ringkasan = ringkasan_kesesuaian(records)

        print("Berhasil memproses " + str(len(records)) + " pasien.")
        if warnings:
            print("Peringatan (" + str(len(warnings)) + " baris dilewati):")
            for w in warnings:
                print(" - " + w)

        # Tampilkan ringkasan singkat di layar (Dosis & Frekuensi)
        try:
            df_ring = pd.DataFrame(ringkasan, columns=["Jenis", "Perhitungan", "Kelompok", "Pembanding", "Total", "Sesuai (n)", "Sesuai (%)",
                                                        "Berlebih (n)", "Berlebih (%)", "Kurang (n)", "Kurang (%)"])
            display(HTML("<h5 style='margin-bottom:2px;'>Ringkasan Kesesuaian Dosis &amp; Frekuensi (Aktual &amp; Manual)</h5>"
                         + df_ring.to_html(index=False, border=1)))
        except Exception:
            pass

        # Ringkasan statistik singkat (Dosis & Frekuensi; Aktual & Manual)
        def _tabel_stat_layar(blok_stats, judul):
            baris_stat = []
            urutan = [("aktual_vs_RS", "Aktual vs Pedoman RS"), ("aktual_vs_Literatur", "Aktual vs Literatur"),
                      ("manual_vs_RS", "Manual vs Pedoman RS"), ("manual_vs_Literatur", "Manual vs Literatur")]
            for ab, blok in blok_stats["per_antibiotik"].items():
                for key, lbl in urutan:
                    u = blok.get(key)
                    if not u:
                        continue
                    if u.get("cukup"):
                        baris_stat.append([ab, lbl, u["n"], u.get("uji", "-"), round(u["p"], 4),
                                           "Berbeda signifikan" if u.get("signifikan") else "Tidak signifikan"])
                    else:
                        baris_stat.append([ab, lbl, u.get("n", "-"), u.get("pesan", "-"), "-", "-"])
            if baris_stat:
                df_stat = pd.DataFrame(baris_stat, columns=["Antibiotik", "Perbandingan", "n", "Uji", "p-value", "Kesimpulan"])
                display(HTML("<h5 style='margin-bottom:2px;'>" + judul + "</h5>" + df_stat.to_html(index=False, border=1)))

        def _mcnemar_layar(mc, label):
            if mc and mc.get("cukup"):
                display(HTML("<p><b>McNemar " + label + "</b>: "
                             + mc.get("metode", "-") + ", p=" + "{0:.4f}".format(mc.get("p", 1)) + " &rarr; "
                             + ("<span style='color:#dc3545;font-weight:600;'>berbeda signifikan</span>" if mc.get("signifikan")
                                else "<span style='color:#198754;font-weight:600;'>tidak berbeda signifikan</span>") + ".</p>"))

        _tabel_stat_layar(stats["dosis"], "Uji Beda DOSIS (Aktual &amp; Manual mg/kg vs Titik Tengah Rekomendasi)")
        _mcnemar_layar(stats["dosis"]["mcnemar_aktual"], "dosis aktual")
        _mcnemar_layar(stats["dosis"]["mcnemar_manual"], "dosis manual")
        _tabel_stat_layar(stats["frekuensi"], "Uji Beda FREKUENSI (Aktual &amp; Manual dosis/hari vs Titik Tengah Rekomendasi)")
        _mcnemar_layar(stats["frekuensi"]["mcnemar_aktual"], "frekuensi aktual")
        _mcnemar_layar(stats["frekuensi"]["mcnemar_manual"], "frekuensi manual")

        # Interpretasi & kesimpulan otomatis (disorot)
        try:
            seksi_interp = buat_interpretasi(records, stats)
            html_i = "<h4 style='margin:10px 0 2px 0;'>Interpretasi Hasil &amp; Kesimpulan</h4>"
            for sec in seksi_interp:
                html_i += "<h5 style='margin:8px 0 2px 0;color:#1F4E78;'>" + sec["judul"] + "</h5>"
                butir = ""
                for tipe, teks in sec["isi"]:
                    t = teks.replace("&", "&amp;").replace("<", "&lt;").replace(">", "&gt;")
                    if tipe == "penting":
                        html_i += ("<div style='background:#fff3cd;border-left:4px solid #e0a800;padding:5px 8px;margin:3px 0;font-weight:600;'>&#9733; " + t + "</div>")
                    elif tipe == "poin":
                        butir += "<li>" + t + "</li>"
                    else:
                        html_i += "<div style='color:#333;margin:2px 0;'>" + t + "</div>"
                if butir:
                    html_i += "<ul style='margin:2px 0 2px 18px;'>" + butir + "</ul>"
            display(HTML(html_i))
        except Exception as e:
            print("(Interpretasi tidak dapat dibuat: " + str(e) + ")")

        # Buat & unduh berkas hasil
        pilihan = list(w_format_out.value)
        stamp = datetime.datetime.now().strftime("%Y%m%d_%H%M%S")
        if "Excel (.xlsx)" in pilihan:
            try:
                fx = buat_output_excel(records, stats, ringkasan, warnings,
                                        "hasil_evaluasi_batch_" + stamp + ".xlsx")
                print("Berkas Excel dibuat: " + fx)
                _unduh_berkas(fx)
            except Exception as e:
                print("Gagal membuat Excel: " + type(e).__name__ + " - " + str(e))
        if "PDF ringkasan" in pilihan:
            try:
                fp, tipe = buat_output_pdf(records, stats, ringkasan, warnings,
                                            "ringkasan_evaluasi_batch_" + stamp)
                if tipe == "pdf":
                    print("Berkas PDF dibuat: " + fp)
                else:
                    print("PDF tidak tersedia; berkas HTML dibuat: " + fp + " (buka di browser, lalu 'Print > Save as PDF').")
                _unduh_berkas(fp)
            except Exception as e:
                print("Gagal membuat PDF: " + type(e).__name__ + " - " + str(e))


btn_proses.on_click(_on_proses_clicked)

display(widgets.HTML("<b>Langkah 1 &mdash; Unduh template dan isi data pasien</b>"))
display(btn_template, out_template)
display(widgets.HTML("<br><b>Langkah 2 &mdash; Unggah berkas Excel yang sudah diisi</b>"))
display(w_upload)
display(widgets.HTML("<br><b>Langkah 3 &mdash; Proses &amp; unduh hasil</b>"))
display(w_format_out)
display(btn_proses, out_proses)
```

```python
import pandas as pd

def jalankan_uji():
    hasil = []
    def cek(nama, kondisi, detail=""):
        hasil.append({"Uji": nama, "Status": "LULUS" if kondisi else "GAGAL", "Detail": detail})

    h = bangun_rekomendasi("Gentamisin", 27, 3, 0.90)
    cek("R-GEN-1 Pedoman RS gentamisin 5-7,5 mg/kg", h["rs_rentang_per_kg"] == (5.0, 7.5), str(h["rs_rentang_per_kg"]))
    cek("R-GEN-2 Pedoman RS gentamisin q24-48h", (h["rs"]["interval_min"], h["rs"]["interval_max"]) == (24, 48))
    cek("R-GEN-3 Rentang mg = BB x mg/kg (0,9 kg -> 4,5-6,75 mg)", h["rs_rentang_mg"] == (4.5, 6.75), str(h["rs_rentang_mg"]))
    kr = rs_kriteria_kesesuaian("Gentamisin", 3)
    cek("R2 Kriteria gentamisin: interval 36 jam termasuk Sesuai", kr["interval_min"] <= 36 <= kr["interval_max"])
    cek("R2 Evaluasi frekuensi aktual q36h vs RS = Sesuai", _kategori_dosis(24/36, 24/36, 24/48, 24/24)[0] == "Sesuai")

    h = bangun_rekomendasi("Amikasin", 30, 4, 1.38)
    cek("R-AMK-1 Pedoman RS amikasin 15 mg/kg q24h",
        h["rs_rentang_per_kg"] == (15.0, 15.0) and (h["rs"]["interval_min"], h["rs"]["interval_max"]) == (24, 24))
    rmin, rmax = h["rs_rentang_mg"]
    cek("R-AMK-2 Amikasin tepat 15 mg/kg: 20,7 mg = Sesuai", _kategori_dosis(20.7, 20.7, rmin, rmax)[0] == "Sesuai")
    cek("R-AMK-3 Amikasin 21 mg = Berlebih (tanpa toleransi)", _kategori_dosis(21, 21, rmin, rmax)[0] == "Berlebih")
    cek("R-AMK-4 Amikasin 20,5 mg = Kurang (tanpa toleransi)", _kategori_dosis(20.5, 20.5, rmin, rmax)[0] == "Kurang")

    h = bangun_rekomendasi("Vankomisin", 30, 10, 1.45, scr=0.5, uop=2.0)
    cek("R1 Vankomisin RS UPL 0-14 hari: 10-15 mg/kg q12h",
        h["rs_rentang_per_kg"] == (10.0, 15.0) and (h["rs"]["interval_min"], h["rs"]["interval_max"]) == (12, 12))
    h = bangun_rekomendasi("Vankomisin", 30, 15, 1.45, scr=0.5, uop=2.0)
    cek("R1 Vankomisin RS UPL 15-28 hari: q8h", (h["rs"]["interval_min"], h["rs"]["interval_max"]) == (8, 8))
    cek("R1 (internal, tidak ditampilkan) nomogram UNC berbasis usia untuk vankomisin tetap terhitung (PMA 32,1; UPL 15 -> tiap 8 jam)", h["rs_ga"]["interval_jam"] == 8)
    h = bangun_rekomendasi("Vankomisin", 38, 20, 3.0, scr=1.5)
    cek("Vankomisin gangguan ginjal (UNC): dosis tunggal 10 mg/kg", h["rs_rentang_per_kg"] == (10.0, 10.0) and h["rs"]["interval_min"] is None)

    kr = rs_kriteria_kesesuaian("Vankomisin", 20)
    cek("R4 Kriteria vankomisin 15-28 hari: q8h", (kr["interval_min"], kr["interval_max"]) == (8, 8))

    h = bangun_rekomendasi("Gentamisin", 40, 12, 3.45)
    cek("R6 SCr kosong: ditandai & diasumsikan normal", h["data_ginjal_tersedia"] is False and h["ginjal_normal"] is True)
    cek("R6 Catatan data ginjal muncul", "ASUMSI fungsi ginjal normal" in _blok_data_ginjal(h))
    cek("R6 Tanpa peringatan kuning (cukup bulan, 3,45 kg, gentamisin)", "Risiko gangguan ginjal tinggi" not in _blok_data_ginjal(h))
    h = bangun_rekomendasi("Vankomisin", 26, 2, 0.74)
    cek("R6 Peringatan kuning (UG 26 minggu, vankomisin)", "Risiko gangguan ginjal tinggi" in _blok_data_ginjal(h))
    h = bangun_rekomendasi("Gentamisin", 33, 8, 1.72, scr=0.6, uop=2.8)
    cek("R6 Data ginjal diisi: tanpa catatan asumsi", _blok_data_ginjal(h) == "")

    # ---- R9: Literatur Terkini = ANMF 2024 saja ----
    cek("R9 Konfigurasi: GABUNG_DRUGSCOM_KE_LITERATUR = False", GABUNG_DRUGSCOM_KE_LITERATUR is False)
    cek("R9 Konfigurasi: TAMPILKAN_DRUGSCOM_TAMBAHAN = False", TAMPILKAN_DRUGSCOM_TAMBAHAN is False)
    h = bangun_rekomendasi("Gentamisin", 27, 3, 0.90)
    cek("R9 Lit gentamisin PMA 27,4: 5 mg/kg (4,5 mg) q48h",
        h["lit_rentang_per_kg"] == (5.0, 5.0) and h["lit_rentang_mg"] == (4.5, 4.5) and h["lit_interval_rentang"] == (48, 48),
        f'{h["lit_rentang_per_kg"]} {h["lit_rentang_mg"]} {h["lit_interval_rentang"]}')
    h = bangun_rekomendasi("Gentamisin", 33, 8, 1.72, scr=0.6, uop=2.8)
    cek("R9 Lit gentamisin PMA 34,1: 5 mg/kg (8,6 mg) q36h",
        h["lit_rentang_mg"] == (8.6, 8.6) and h["lit_interval_rentang"] == (36, 36),
        f'{h["lit_rentang_mg"]} {h["lit_interval_rentang"]}')
    h = bangun_rekomendasi("Gentamisin", 39, 2, 3.2, hipotermia=True)
    cek("R9 Lit gentamisin hipotermia terapeutik: q36h", h["lit_interval_rentang"] == (36, 36), str(h["lit_interval_rentang"]))
    h = bangun_rekomendasi("Amikasin", 30, 4, 1.38)
    cek("R9 Lit amikasin PMA 30,6, UPL 4: 12 mg/kg (16,56 mg) q36h",
        h["lit_rentang_per_kg"] == (12.0, 12.0) and h["lit_rentang_mg"] == (16.56, 16.56) and h["lit_interval_rentang"] == (36, 36),
        f'{h["lit_rentang_mg"]} {h["lit_interval_rentang"]}')
    h = bangun_rekomendasi("Amikasin", 26, 3, 0.80)
    cek("R9 Lit amikasin PMA 26,4, UPL 3: 14 mg/kg (11,2 mg) q48h",
        h["lit_rentang_mg"] == (11.2, 11.2) and h["lit_interval_rentang"] == (48, 48),
        f'{h["lit_rentang_mg"]} {h["lit_interval_rentang"]}')
    h = bangun_rekomendasi("Vankomisin", 30, 15, 1.45, scr=0.5, uop=2.0)
    cek("R9 Lit vankomisin PMA 32,1, UPL 15: 15 mg/kg (21,75 mg) q8h",
        h["lit_rentang_mg"] == (21.75, 21.75) and h["lit_interval_rentang"] == (8, 8),
        f'{h["lit_rentang_mg"]} {h["lit_interval_rentang"]}')
    lmin, lmax = h["lit_rentang_mg"]
    cek("R9 Vankomisin aktual 20 mg vs Literatur = Kurang (tanpa toleransi)", _kategori_dosis(20, 20, lmin, lmax)[0] == "Kurang")
    _tangkap = []
    _disp_asli = globals()["display"]
    globals()["display"] = lambda *a, **k: _tangkap.extend(getattr(x, "data", str(x)) for x in a)
    try:
        tampilkan_rekomendasi("Vankomisin", h)
    finally:
        globals()["display"] = _disp_asli
    _html = "".join(_tangkap)
    cek("R9 Tampilan rekomendasi tanpa kartu Drugs.com", "Drugs.com" not in _html and "ANMF" in _html)

    # ---- R10: selisih dalam satuan & % ----
    k, t = _kategori_dosis(21, 21, 20.7, 20.7, "mg")
    cek("R10 Amikasin 21 mg vs 20,7 mg: Berlebih +0,3 mg; +1,4%", (k, t) == ("Berlebih", "+0,3 mg; +1,4%"), str(t))
    k, t = _kategori_dosis(20, 20, 21.75, 21.75, "mg")
    cek("R10 Vankomisin 20 mg vs 21,75 mg: Kurang -1,75 mg; -8,0%", (k, t) == ("Kurang", "-1,75 mg; -8,0%"), str(t))
    _f = _render_kategori_frekuensi(24, 36, 36)
    cek("R10 Frekuensi q24h vs q36h: Berlebih +0,333 dosis/hari; +50,0%; interval 12 jam lebih pendek",
        "Berlebih" in _f and "+0,333 dosis/hari; +50,0%" in _f and "12 jam lebih pendek" in _f, _f)
    _f = _render_kategori_frekuensi(48, 24, 36)
    cek("R10 Frekuensi q48h vs q24-36h: Kurang -0,167 dosis/hari; -25,0%; interval 12 jam lebih panjang",
        "Kurang" in _f and "-0,167 dosis/hari; -25,0%" in _f and "12 jam lebih panjang" in _f, _f)
    _m = _render_manual_vs_rekom(8.0, 13.0, 8.6, 12.9)
    cek("R10 Manual 8-13 mg vs 8,6-12,9 mg: Min -0,6 mg; -7,0% | Maks +0,1 mg; +0,8%",
        "-0,6 mg; -7,0%" in _m and "+0,1 mg; +0,8%" in _m, _m)
    _m = _render_manual_vs_rekom(8.65, 12.9, 8.6, 12.9)
    cek("R10 Manual tanpa toleransi: selisih 0,05 mg (0,6%) tetap dilaporkan", "+0,05 mg; +0,6%" in _m and "Maks: <span" in _m and "sama" in _m, _m)
    _s = _status_interval_rentang_minmax(12, 24, 48)[1]
    cek("R10 Kriteria RS interval q12h vs 24-48 jam: 12 jam lebih pendek; +1 dosis/hari; +100,0%",
        "12 jam lebih pendek" in _s and "+1 dosis/hari; +100,0%" in _s, _s)

    # ---- R11: singkatan dijelaskan ----
    h = bangun_rekomendasi("Gentamisin", 27, 3, 0.90)
    _r = _perjelas_singkatan(_html_rekomendasi("Gentamisin", h))
    cek("R11 Singkatan dijelaskan: Rumah Sakit (RS), usia pascamenstruasi (PMA), Standar Prosedur Operasional (SPO)",
        "Rumah Sakit (RS)" in _r and "usia pascamenstruasi (PMA)" in _r and "Standar Prosedur Operasional (SPO)" in _r)
    cek("R11 Kotak 'Keterangan singkatan' ada di hasil", "Keterangan singkatan" in _r)
    cek("R11 Tidak ada notasi qXh atau GA/PNA pada hasil",
        _re_sk.search(r"(?<![A-Za-z])q\d+h", _r) is None and "GA/PNA" not in _r)
    h2 = bangun_rekomendasi("Amikasin", 30, 4, 1.38)
    _e = _perjelas_singkatan(_html_evaluasi(h2, 20, 21, 24, 36, 21, 24))
    cek("R11 Hasil evaluasi: singkatan dijelaskan & ada keterangan singkatan",
        "Rumah Sakit (RS)" in _e and "Keterangan singkatan" in _e and _re_sk.search(r"(?<![A-Za-z])q\d+h", _e) is None)

    # ---- R12: peringatan berwarna ----
    h = bangun_rekomendasi("Gentamisin", 40, 12, 3.45)
    _r = _html_rekomendasi("Gentamisin", h)
    cek("R12 Data ginjal kosong (risiko rendah): kotak kuning PERHATIAN di atas tabel",
        "PERHATIAN: Data fungsi ginjal tidak diisi" in _r and "#fff3cd" in _r and _r.index("PERHATIAN") < _r.index("<table"))
    h = bangun_rekomendasi("Vankomisin", 26, 2, 0.74)
    cek("R12 Data ginjal kosong + risiko tinggi: kotak merah PERINGATAN",
        "PERINGATAN: Data fungsi ginjal tidak diisi pada pasien berisiko tinggi" in _html_rekomendasi("Vankomisin", h))
    h = bangun_rekomendasi("Gentamisin", 38, 5, 3.0, scr=1.2, uop=2.0)
    _r = _html_rekomendasi("Gentamisin", h)
    cek("R12 Gangguan ginjal: kotak merah di atas tabel",
        "PERINGATAN: Kemungkinan gangguan fungsi ginjal" in _r and _r.index("PERINGATAN") < _r.index("<table"))
    k = _ringkasan_evaluasi(bangun_rekomendasi("Amikasin", 30, 4, 1.38), 21, 24)[0]
    cek("R12 Ringkasan evaluasi amikasin 21 mg: merah + Berlebih (+0,3 mg; +1,4%)",
        "PERINGATAN" in k and "Berlebih (+0,3 mg; +1,4%)" in k, k[:200])
    k = _ringkasan_evaluasi(bangun_rekomendasi("Vankomisin", 30, 15, 1.45, scr=0.5, uop=2.0), 21.75, 8)[0]
    cek("R12 Ringkasan evaluasi vankomisin 21,75 mg tiap 8 jam: hijau SESUAI", "SESUAI: Dosis dan frekuensi aktual sesuai" in k)

    # ---- R13 & R14: Bagian 3 mandiri; Pasien Baru ----
    if "form_eval" in globals() and "form_rek" in globals():
        n_rek0, n_eval0 = len(RIWAYAT_REKOMENDASI), len(RIWAYAT_EVALUASI)
        reset_form(form_rek, "Pasien 1")
        reset_eval("Uji Mandiri")
        form_eval["antibiotik"].value = "Amikasin"
        form_eval["ga"].value, form_eval["pna"].value, form_eval["bb"].value = "30", "4", "1,38"
        w_dosis_aktual.value, w_interval_aktual.value = "21", "24"
        _on_evaluasi_clicked(None)
        cek("R13 Evaluasi berjalan tanpa Bagian 2 dan memakai data pasien Bagian 3",
            len(RIWAYAT_EVALUASI) == n_eval0 + 1 and RIWAYAT_EVALUASI[-1]["Kode pasien"] == "Uji Mandiri"
            and RIWAYAT_EVALUASI[-1]["Antibiotik"] == "Amikasin" and form_rek["antibiotik"].value is None)
        form_rek["kode"].value = "Uji A"
        form_rek["antibiotik"].value = "Gentamisin"
        form_rek["ga"].value, form_rek["pna"].value, form_rek["bb"].value = "27", "3", "0.9"
        _on_hitung_clicked(None)
        _on_pasien_baru_rek(None)
        cek("R14 Pasien Baru: formulir kembali ke nilai awal dengan kode pasien baru",
            form_rek["ga"].value == "" and form_rek["bb"].value == "" and form_rek["antibiotik"].value is None
            and form_rek["kode"].value == "Pasien " + str(n_rek0 + 2))
        cek("R14 Ringkasan pasien sebelumnya tetap tersimpan di riwayat",
            len(RIWAYAT_REKOMENDASI) == n_rek0 + 1 and RIWAYAT_REKOMENDASI[-1]["Kode pasien"] == "Uji A")
        form_rek["antibiotik"].value = "Amikasin"
        form_rek["ga"].value, form_rek["pna"].value, form_rek["bb"].value = "30", "4", "1.38"
        _on_hitung_clicked(None)
        cek("R14 Pasien berikutnya langsung dapat dihitung", len(RIWAYAT_REKOMENDASI) == n_rek0 + 2)
        # kembalikan keadaan formulir & riwayat seperti semula
        del RIWAYAT_REKOMENDASI[n_rek0:]
        del RIWAYAT_EVALUASI[n_eval0:]
        reset_form(form_rek, "Pasien 1")
        reset_eval("Pasien 1")
        for _o in (out_hitung, out_evaluasi):
            _o.clear_output()
        _perbarui_riwayat(out_riwayat_rek, RIWAYAT_REKOMENDASI, "Riwayat rekomendasi pada sesi ini")
        _perbarui_riwayat(out_riwayat_eval, RIWAYAT_EVALUASI, "Riwayat evaluasi pada sesi ini")

    # ---- R15: hitung manual opsional ----
    h = bangun_rekomendasi("Amikasin", 30, 4, 1.38)
    _e = _html_evaluasi(h, None, None, None, None, 21, 24)
    cek("R15 Tanpa hitung manual: evaluasi tetap berjalan, kolom Hitung Manual tidak tampil",
        "Hitung Manual</th>" not in _e and "Manual vs" not in _e and "Hitung manual tidak diisi" in _e)
    _e = _html_evaluasi(h, 20, 21, 24, 36, 21, 24)
    cek("R15 Dengan hitung manual: kolom Hitung Manual & Manual vs tampil", "Hitung Manual</th>" in _e and "Manual vs Literatur" in _e)

    # ---- R16: isian di luar rentang ditolak dengan pesan, bukan dipotong diam-diam ----
    if "form_rek" in globals():
        _f = buat_form_pasien("Uji")
        _f["antibiotik"].value = "Gentamisin"
        _f["ga"].value, _f["pna"].value, _f["bb"].value = "34", "29", "2.1"
        try:
            hitung_dari_form(_f); _ok = False; _pesan = ""
        except ValueError as _ex:
            _ok = True; _pesan = str(_ex)
        cek("R16 Usia pascalahir 29 hari (data lengkap) ditolak dengan pesan cakupan", _ok and "0-28 hari" in _pesan, _pesan)
        _f["pna"].value, _f["bb"].value = "3", "0.2"
        try:
            hitung_dari_form(_f); _ok = False; _pesan = ""
        except ValueError as _ex:
            _ok = True; _pesan = str(_ex)
        cek("R16 Berat 0,2 kg ditolak dengan pesan rentang (tidak dibulatkan ke 0,3 kg)", _ok and "Berat badan 0,2 kg" in _pesan, _pesan)

    # ---- R17: kolom isian kosong sejak awal & wajib diisi ----
    if "form_rek" in globals():
        _f = buat_form_pasien("Uji")
        cek("R17 Formulir baru: antibiotik belum dipilih & kolom angka kosong",
            _f["antibiotik"].value is None and _f["ga"].value == "" and _f["pna"].value == "" and _f["bb"].value == "")
        try:
            hitung_dari_form(_f); _pesan = ""
        except ValueError as _ex:
            _pesan = str(_ex)
        cek("R17 Formulir kosong ditolak: Antibiotik, Usia gestasi, Usia pascalahir, Berat badan wajib diisi",
            all(x in _pesan for x in ["wajib diisi", "Antibiotik", "Usia gestasi saat lahir", "Usia pascalahir", "Berat badan"]), _pesan)
        _f["antibiotik"].value = "Gentamisin"
        _f["ga"].value, _f["pna"].value, _f["bb"].value = "33", "8", "1,72"
        cek("R17 Desimal dengan koma diterima (1,72 kg)", hitung_dari_form(_f)["berat_kg"] == 1.72)
        _f["bb"].value = "1.7.2"
        try:
            hitung_dari_form(_f); _pesan = ""
        except ValueError as _ex:
            _pesan = str(_ex)
        cek("R17 Isian bukan angka ditolak", "bukan angka" in _pesan, _pesan)
        _f["bb"].value = "1,72"
        _f["pna"].value = "8,5"
        try:
            hitung_dari_form(_f); _pesan = ""
        except ValueError as _ex:
            _pesan = str(_ex)
        cek("R17 Usia pascalahir harus bilangan bulat", "bilangan bulat" in _pesan, _pesan)
        _f["pna"].value = "8"
        _f["ada_ginjal"].value = True
        try:
            hitung_dari_form(_f); _pesan = ""
        except ValueError as _ex:
            _pesan = str(_ex)
        cek("R17 Data ginjal dicentang tetapi kreatinin kosong ditolak", "Kreatinin serum" in _pesan, _pesan)
        reset_eval("Uji")
        try:
            _baca_isian_evaluasi(); _pesan = ""
        except ValueError as _ex:
            _pesan = str(_ex)
        cek("R17 Evaluasi: dosis & interval aktual kosong ditolak", "Dosis aktual" in _pesan and "Frekuensi aktual" in _pesan, _pesan)
        reset_eval("Pasien 1")

    # ---- R18: usia gestasi minggu + hari; pita usia memakai minggu LENGKAP ----
    if "form_rek" in globals():
        _f = buat_form_pasien("Uji")
        _f["antibiotik"].value = "Amikasin"
        _f["ga"].value, _f["ga_hari"].value, _f["pna"].value, _f["bb"].value = "29", "3", "3", "1,1"
        _h = hitung_dari_form(_f)
        cek("R18 Usia gestasi 29 minggu 3 hari dibaca 29 + 3/7 minggu", abs(_h["ga_lahir"] - (29 + 3 / 7.0)) < 1e-9)
        cek("R18 Amikasin ANMF PMA 29 minggu 6 hari (pita <30) = 14 mg/kg tiap 48 jam",
            _h["lit_rentang_per_kg"] == (14.0, 14.0) and _h["lit_interval_rentang"] == (48, 48), str(_h["lit_interval_rentang"]))
        _f["ga"].value = "32,4"
        try:
            hitung_dari_form(_f); _pesan = ""
        except ValueError as _ex:
            _pesan = str(_ex)
        cek("R18 Minggu usia gestasi harus bilangan bulat (hari di kolom tambahan)", "bilangan bulat" in _pesan, _pesan)
    cek("R18 Vankomisin UNC: PMA 29 minggu 4 hari termasuk pita <30 minggu (tiap 18 jam bila usia pascalahir <=14 hari)",
        rs_vankomisin(29 + 4 / 7.0, 5)[4] == 18)
    cek("R18 Format usia: 32,571 minggu -> '32 minggu 4 hari'", _fmt_usia_minggu(32 + 4 / 7.0) == "32 minggu 4 hari")

    # ---- R19: hipotermia tidak memperpendek interval bayi sangat prematur ----
    cek("R19 Gentamisin ANMF PMA 27 minggu + hipotermia: tetap tiap 48 jam", lit_gentamisin(27.4, hipotermia=True)[1] == 48)
    cek("R19 Gentamisin ANMF PMA 39 minggu + hipotermia: tiap 36 jam", lit_gentamisin(39.3, hipotermia=True)[1] == 36)

    # ---- R20: peringatan klinis tambahan ----
    h = bangun_rekomendasi("Gentamisin", 33, 8, 1.72, scr=0.7)
    cek("R20 Kreatinin 0,5-<1 tanpa produksi urin: kotak kuning 'Produksi urin tidak diisi'",
        "PERHATIAN: Produksi urin tidak diisi" in _html_rekomendasi("Gentamisin", h))
    h = bangun_rekomendasi("Vankomisin", 26, 2, 0.74, scr=1.1, uop=1.8)
    cek("R20 Kreatinin tinggi pada 72 jam pertama: kotak kuning kreatinin ibu",
        "Kreatinin serum pada 72 jam pertama" in _html_rekomendasi("Vankomisin", h))
    h = bangun_rekomendasi("Gentamisin", 30, 2, 1.3, hipotermia=True)
    cek("R20 Hipotermia pada bayi <35 minggu: kotak kuning periksa isian", "Periksa isian hipotermia terapeutik" in _html_rekomendasi("Gentamisin", h))
    if "form_rek" in globals():
        _f = buat_form_pasien("Uji")
        _f["antibiotik"].value = "Gentamisin"
        _f["ga"].value, _f["pna"].value, _f["bb"].value = "33", "8", "1720"
        try:
            hitung_dari_form(_f); _pesan = ""
        except ValueError as _ex:
            _pesan = str(_ex)
        cek("R20 Berat 1720 ditolak dengan petunjuk gram -> kilogram", "gram" in _pesan and "1,72 kg" in _pesan, _pesan)
        _f["bb"].value = "1,72"
        _f["ada_ginjal"].value = True
        _f["scr"].value = "70"
        try:
            hitung_dari_form(_f); _pesan = ""
        except ValueError as _ex:
            _pesan = str(_ex)
        cek("R20 Kreatinin 70 ditolak dengan petunjuk µmol/L", "µmol/L" in _pesan, _pesan)

    # ---- R21: tampilan Indonesia & kejelasan ----
    h = bangun_rekomendasi("Gentamisin", 33, 8, 1.72, scr=0.6, uop=2.8)
    _r = _perjelas_singkatan(_html_rekomendasi("Gentamisin", h))
    cek("R21 Angka memakai koma desimal (8,6-12,9 mg)", "8,6-12,9 mg" in _r and "8.6" not in _r)
    cek("R21 Kriteria dinyatakan per pemberian (bukan per 24 jam)", "mg/kg per pemberian" in _r and "mg/kg/24 jam" not in _r)
    cek("R21 Simbol >= tidak tampil pada hasil (memakai ≥)", ">=" not in _r)
    _hv = bangun_rekomendasi("Vankomisin", 26, 2, 0.74, scr=1.1, uop=1.8)
    _rv = _html_rekomendasi("Vankomisin", _hv)
    cek("R21 Vankomisin + gangguan ginjal: catatan bahwa kolom RS mengikuti aturan ginjal UNC",
        "mengikuti aturan gangguan ginjal UNC" in _rv and "prematur dan dengan indikasi gangguan ginjal" in _rv and ">=" not in _rv)
    cek("R21 Usia ditampilkan minggu + hari & ada pernyataan alat bantu",
        "34 minggu 1 hari" in _r and "alat bantu keputusan klinis" in _r)

    # ---- R23: disamakan dengan naskah -- hanya Pedoman RS vs Literatur Terkini ----
    h = bangun_rekomendasi("Gentamisin", 27, 3, 0.90)
    _r = _html_rekomendasi("Gentamisin", h)
    _e = _html_evaluasi(h, 4, 6, 24, 48, 4.5, 24)
    cek("R23 Rekomendasi: tidak ada kolom Pedoman RS 'dengan usia gestasi'", "dengan usia gestasi" not in _r and _r.count("<th") == 3, str(_r.count("<th")))
    cek("R23 Evaluasi: tidak ada kolom 'dengan usia gestasi'", "dengan usia gestasi" not in _e and "tanpa usia gestasi" not in _e)
    _k, _rg = _ringkasan_evaluasi(h, 4.5, 24)
    cek("R23 Ringkasan evaluasi: 4 baris (dosis & frekuensi vs Pedoman RS dan Literatur)", _k.count("<li>") == 4 and len(_rg) == 8, str(_k.count("<li>")))

    try:
        bangun_rekomendasi("Gentamisin", 34, 29, 2.1)
        cek("R5 Usia pascalahir 29 hari ditolak", False)
    except ValueError:
        cek("R5 Usia pascalahir 29 hari ditolak", True)

    if "proses_dataframe" in globals():
        df = pd.DataFrame([
            ["U1", "Gentamisin", 33, 8, 1.72, "", "", "Tidak", "Tidak", "Tidak", "", "", "", "", 8.6, 36, ""],
            ["U2", "Amikasin", 30, 4, 1.38, "", "", "Tidak", "Tidak", "Tidak", "", "", "", "", 20.7, 24, ""],
            ["U3", "Vankomisin", 30, 15, 1.45, 0.5, 2.0, "Tidak", "Tidak", "Tidak", "", "", "", "", 20, 8, ""],
            ["U4", "Gentamisin", 34, 35, 2.1, "", "", "Tidak", "Tidak", "Tidak", "", "", "", "", 10, 24, ""],
            ["U5", "Gentamisin", 33, 8, 1720, "", "", "Tidak", "Tidak", "Tidak", "", "", "", "", 8.6, 36, ""],
            ["U6", "Amikasin", 29, 3, 1.1, "", "", "Tidak", "Tidak", "Tidak", "", "", "", "", 16.5, 24, 3],
        ], columns=KOLOM_TEMPLATE)
        rec, warn = proses_dataframe(df, bangun_rekomendasi, _kategori_dosis, _fmt_num)
        r = {x["Kode Pasien"]: x for x in rec}
        cek("Batch U1 gentamisin 8,6 mg q36h: dosis & frek RS Sesuai",
            r["U1"]["Aktual vs RS (dosis)"].startswith("Sesuai") and r["U1"]["Aktual vs RS (frek)"].startswith("Sesuai"))
        cek("Batch U1 SCr kosong -> 'Tidak tersedia (diasumsikan normal)'", r["U1"]["Fungsi Ginjal"] == "Tidak tersedia (diasumsikan normal)")
        cek("Batch U2 amikasin 20,7 mg q24h: Sesuai",
            r["U2"]["Aktual vs RS (dosis)"].startswith("Sesuai") and r["U2"]["Aktual vs RS (frek)"].startswith("Sesuai"))
        cek("Batch U3 vankomisin UPL 15 hari, 20 mg q8h: Sesuai",
            r["U3"]["Aktual vs RS (dosis)"].startswith("Sesuai") and r["U3"]["Aktual vs RS (frek)"].startswith("Sesuai"))
        cek("R9 Batch U1: Aktual vs Literatur (ANMF 5 mg/kg q36h) dosis & frek Sesuai",
            r["U1"]["Aktual vs Literatur (dosis)"].startswith("Sesuai") and r["U1"]["Aktual vs Literatur (frek)"].startswith("Sesuai"),
            r["U1"]["Dosis Literatur (mg)"] + " / " + r["U1"]["Frek Literatur"])
        cek("R9 Batch U2: amikasin 20,7 mg q24h vs Literatur (12 mg/kg q36h) = Berlebih",
            r["U2"]["Aktual vs Literatur (dosis)"].startswith("Berlebih") and r["U2"]["Aktual vs Literatur (frek)"].startswith("Berlebih"),
            r["U2"]["Aktual vs Literatur (dosis)"] + " / " + r["U2"]["Aktual vs Literatur (frek)"])
        cek("R10 Batch U2 amikasin vs Literatur: selisih mg & %, frek dosis/hari & jam",
            r["U2"]["Aktual vs Literatur (dosis)"] == "Berlebih (+4,14 mg; +25,0%)"
            and r["U2"]["Aktual vs Literatur (frek)"] == "Berlebih (+0,333 dosis/hari; +50,0%; interval 12 jam lebih pendek dari interval 36 jam)",
            r["U2"]["Aktual vs Literatur (dosis)"] + " / " + r["U2"]["Aktual vs Literatur (frek)"])
        cek("R22 Batch: berat 1720 (gram) dilewati dengan peringatan satuan", "U5" not in r and any("U5" in w and "kilogram" in w for w in warn))
        cek("R18 Batch: kolom tambahan hari usia gestasi dipakai (29 minggu 3 hari)", abs(r["U6"]["Usia Gestasi (mgg)"] - (29 + 3 / 7.0)) < 1e-9)
        cek("R23 Batch: tidak ada kolom 'dengan usia gestasi'", not any("usia gestasi & pascalahir" in k for k in r["U1"]))
        cek("Batch U4 usia pascalahir 35 hari dilewati dengan peringatan", "U4" not in r and any("U4" in w for w in warn))

    t = pd.DataFrame(hasil)
    print(f"{(t['Status'] == 'LULUS').sum()} dari {len(t)} uji LULUS  |  versi aturan {VERSI_ATURAN}")
    return t

hasil_uji = jalankan_uji()
hasil_uji
```

## Catatan Akhir & Batasan

- Kalkulator ini adalah **alat bantu**; keputusan akhir dosis, interval, dan pemantauan tetap berada pada apoteker klinis dan DPJP.
- **Rujukan:** hasil membandingkan **Pedoman Rumah Sakit** (gentamisin 5-7,5 mg/kg per pemberian tiap 24-48 jam; amikasin tepat 15 mg/kg tiap 24 jam; vankomisin 10-15 mg/kg per pemberian tiap 12 jam pada usia pascalahir 0-14 hari dan tiap 8 jam pada 15-28 hari) dengan **Literatur Terkini** (ANMF 2024 saja, nilai titik dosis & interval menurut usia). Pedoman RS tidak membedakan usia gestasi, sama dengan naskah *Pediatric Drugs*. Drugs.com/ASHP tidak dipakai; kodenya disimpan dan dapat diaktifkan lewat `GABUNG_DRUGSCOM_KE_LITERATUR` / `TAMPILKAN_DRUGSCOM_TAMBAHAN` atas instruksi peneliti.
- **Tampilan:** dosis total per pemberian (mg) dengan mg/kg sebagai keterangan; frekuensi sebagai "tiap X jam"; koma desimal; usia sebagai "minggu + hari"; singkatan dijelaskan pada kemunculan pertama dan dirangkum dalam kotak *Keterangan singkatan*; peringatan berwarna (merah, kuning, biru, hijau) tampil di atas tabel.
- **Isian:** semua kolom kosong sejak awal dan wajib diisi sebelum menghitung; usia gestasi diisi minggu lengkap + tambahan hari; usia pascalahir dengan hari lahir = 0; berat dalam kilogram; kreatinin dalam mg/dL. Isian di luar rentang ditolak dengan pesan, tidak dipotong diam-diam. **Kode pasien** diisi tanpa nama atau nomor rekam medis.
- **Kriteria Kesesuaian Pedoman RS** berlaku untuk semua neonatus usia pascalahir 0-28 hari. Untuk bayi prematur atau dengan indikasi gangguan ginjal, kalkulator menampilkan catatan agar hasil juga dibandingkan dengan kolom Literatur Terkini.
- **Data fungsi ginjal tidak diisi:** fungsi ginjal diasumsikan normal sesuai usia, dengan kotak kuning (merah bila pasien berisiko tinggi: usia gestasi <29 minggu, berat <1,5 kg, hipotermia terapeutik, indometasin/ibuprofen, atau vankomisin). Produksi urin opsional; diperlukan hanya bila kreatinin 0,5 sampai <1 mg/dL.
- **Penyesuaian ginjal otomatis** hanya untuk vankomisin (aturan UNC). Untuk gentamisin/amikasin pada gangguan ginjal, kalkulator memberi peringatan; penyesuaian interval dilakukan bersama DPJP berdasarkan kadar obat.
- **Catatan Pustaka Tambahan** (WHO 2024, Swiss Society of Neonatology 2024, JAID/JSC 2021, ACOG 2020/AAP) bersifat kualitatif dan tidak dihitung otomatis.
- **Cakupan:** usia pascalahir 0-28 hari. Input di atas 28 hari ditolak (mode tunggal) atau dilewati dengan peringatan (batch).
- SPO rumah sakit dapat direvisi. Perbarui fungsi `rs_gentamisin`, `rs_amikasin`, `rs_vankomisin`, dan `rs_kriteria_kesesuaian` bila SPO/kriteria terbaru tersedia (lihat penanda `# EDIT DI SINI`), lalu naikkan `VERSI_ATURAN`.
- Notebook ini tidak divalidasi secara klinis/regulatori dan tidak dimaksudkan sebagai alat medis (*medical device*) bersertifikasi.
